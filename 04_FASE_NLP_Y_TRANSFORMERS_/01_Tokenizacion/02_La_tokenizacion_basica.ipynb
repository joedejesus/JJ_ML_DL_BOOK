```python
# Bloque 1 — Definición del texto
"""Este bloque define un texto de entrada que será segmentado mediante tokenización básica.
La variable "texto" contiene una cadena que representa una secuencia lingüística continua.
El contenido será dividido en unidades discretas siguiendo reglas simples."""

texto = "La tokenización básica divide el texto en unidades simples."

texto # Muestra el texto original para verificar su contenido.
```

```python
# Bloque 2 — Tokenización por espacios
"""Este bloque aplica tokenización básica mediante separación por espacios.
Aplicamos el método ".split()" a la variable "texto" para obtener una lista de fragmentos separados por espacios en blanco.
La variable "tokens" contendrá la lista de fragmentos resultante de la tokenización.
Esta técnica es directa y no separa la puntuación de las palabras ni considera variaciones morfológicas."""

tokens = texto.split()

tokens # Muestra la lista de tokens para verificar el resultado de la tokenización.
```

```python
# Bloque 3 — Construcción de un vocabulario básico
"""Este bloque construye un vocabulario asignando un identificador numérico a cada token.
Aplicamos el constructor "set()" a la lista de tokens "tokens" para obtener un conjunto de tokens únicos.
Aplicamos el constructor "list()" al conjunto resultante de tokens únicos para obtener una lista de tokens sin duplicados.
La variable "tokens_unicos" contiene la lista de tokens sin duplicados.

Utilizamos una comprensión de diccionario "{}" para asignar un índice único a cada token en la lista de tokens sin duplicados.
La subexpresión "token: idx" asigna a cada token su índice en la lista de tokens sin duplicados.
La subexpresión "for idx, token in enumerate(tokens_unicos)" recorre cada token junto con su índice en la lista de tokens sin duplicados.
La clase "enumerate()" enumera los elementos y proporciona un índice para cada elemento en la secuencia.
La variable "vocabulario" contiene el diccionario que mapea cada token único a su índice correspondiente.
El orden del conjunto no está garantizado entre procesos, por lo que las asignaciones de índices pueden variar."""

tokens_unicos = list(set(tokens))
vocabulario = {token: idx for idx, token in enumerate(tokens_unicos)}

tokens_unicos, vocabulario # Muestra la lista de tokens únicos y el diccionario de vocabulario.
```

```python
# Bloque 4 — Conversión de tokens a índices
"""Este bloque convierte cada token del texto en su identificador numérico correspondiente.
Utilizamos una comprensión de lista "[]" para convertir cada token en su índice correspondiente en el vocabulario.
La subexpresión "vocabulario[token]" obtiene el índice correspondiente a cada token en el vocabulario definido previamente.
La subexpresión "for token in tokens" recorre cada token en la lista de tokens del texto original.
Esta comprensión de lista permite transformar el texto tokenizado en una secuencia de índices de manera eficiente.

La variable "secuencia_indices" almacena la secuencia de números enteros correspondiente al texto tokenizado.
Esta representación es compatible con modelos que operan sobre tensores."""

secuencia_indices = [vocabulario[token] for token in tokens]

secuencia_indices # Muestra la secuencia de índices correspondiente al texto tokenizado.
```

```python
# Bloque 5 — Padding para longitud fija
"""Este bloque aplica truncado y padding para generar una secuencia de longitud fija.
La variable "longitud_objetivo" define la longitud deseada; en este caso, 12.

Aplicamos slicing sobre la variable "secuencia_indices" para limitar su longitud a "longitud_objetivo".
El rango [:longitud_objetivo] indica que se tomarán los elementos desde el índice cero hasta el índice "longitud_objetivo" (excluyendo este último).
La variable "secuencia_fija" almacena la secuencia ajustada con la longitud deseada; en este caso, 12 elementos.

Utilizamos el operador de asignación "+=" para añadir ceros hasta alcanzar la longitud deseada.
La expresión "[0] * (longitud_objetivo - len(secuencia_fija))" genera una lista de ceros cuya longitud es la diferencia entre la longitud objetivo 
y la longitud actual de la secuencia.
La variable "secuencia_fija" almacena la secuencia final con longitud uniforme.
El índice 0 también corresponde a un token del vocabulario, por lo que no distingue el padding sin información adicional."""

longitud_objetivo = 12
secuencia_fija = secuencia_indices[:longitud_objetivo]
secuencia_fija += [0] * (longitud_objetivo - len(secuencia_fija))

secuencia_fija # Muestra la secuencia ajustada con padding hasta la longitud deseada.
```

```python
# Bloque 6 — Resultados finales
"""Este bloque organiza los resultados del proceso de tokenización básica.
La variable "resultado" almacena en forma de diccionario los tokens generados, el vocabulario, la secuencia de índices y la secuencia con padding.
La estructura permite visualizar las características esenciales de la tokenización básica."""

resultado = {
    "tokens": tokens,
    "vocabulario": vocabulario,
    "secuencia_indices": secuencia_indices,
    "secuencia_fija": secuencia_fija
}

resultado # Muestra el diccionario con los resultados finales de la tokenización básica.
```
