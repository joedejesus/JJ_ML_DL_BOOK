```python
# Bloque 1 — Definición de un texto de entrada
"""Este bloque define un texto de entrada que será procesado mediante tokenización.
La variable "texto" contiene una cadena que representa una secuencia lingüística.
El contenido será segmentado en unidades discretas para su posterior representación numérica."""

texto = "La tokenización convierte texto en unidades procesables por un modelo."

texto # Muestra el texto de entrada para verificar su contenido.
```

```python
# Bloque 2 — Tokenización basada en palabras
"""Este bloque aplica tokenización basada en palabras utilizando una división por espacios.
Aplicamos el método ".split()" a la variable "texto" para obtener una lista de fragmentos separados por espacios en blanco.
La variable "tokens_palabras" contendrá la lista de fragmentos resultante de la tokenización.
Esta técnica no separa la puntuación de las palabras ni maneja variaciones morfológicas."""

tokens_palabras = texto.split()

tokens_palabras # Muestra la lista de fragmentos tokenizados para verificar su contenido.
```

```python
# Bloque 3 — Construcción de un vocabulario simple
"""Este bloque construye un vocabulario asignando un identificador numérico a cada token.
Aplicamos "dict.fromkeys()" a la lista "tokens_palabras" para eliminar duplicados conservando el orden de aparición.
Aplicamos el constructor "list()" para obtener la lista de tokens sin duplicados.
La variable "tokens_unicos" contiene la lista de tokens sin duplicados.

Utilizamos una comprensión de diccionario "{}" para asignar un índice único a cada token en la lista de tokens sin duplicados.
La subexpresión "token: idx" asigna a cada token su índice en la lista de tokens sin duplicados.
La subexpresión "for idx, token in enumerate(tokens_unicos)" recorre cada token junto con su índice en la lista de tokens sin duplicados.
La clase "enumerate()" enumera los elementos y proporciona un índice para cada elemento en la secuencia.
La variable "vocabulario" contiene el diccionario que mapea cada token único a su índice correspondiente.
La asignación de índices es estable para el mismo texto y su orden de tokens."""

tokens_unicos = list(dict.fromkeys(tokens_palabras))
vocabulario = {token: idx for idx, token in enumerate(tokens_unicos)}

tokens_unicos, vocabulario # Muestra la lista de tokens únicos y el diccionario del vocabulario con los índices correspondientes.
```

```python
# Bloque 4 — Conversión de tokens a índices
"""Este bloque convierte cada token del texto original en su identificador numérico.
Utilizamos una comprensión de lista "[]" para convertir cada token en su índice correspondiente en el vocabulario.
La subexpresión "vocabulario[token]" obtiene el índice correspondiente a cada token en el vocabulario definido previamente.
La subexpresión "for token in tokens_palabras" recorre cada token en la lista de tokens del texto original.
Esta comprensión de lista permite transformar el texto tokenizado en una secuencia de índices de manera eficiente.

La variable "secuencia_indices" almacena la secuencia de números enteros correspondiente al texto tokenizado.
Esta representación es compatible con modelos que operan sobre tensores."""

secuencia_indices = [vocabulario[token] for token in tokens_palabras]

secuencia_indices # Muestra la secuencia de índices correspondiente al texto tokenizado.
```

```python
# Bloque 5 — Tokenización subword con HuggingFace
"""Este bloque aplica tokenización subword utilizando un tokenizador preentrenado de HuggingFace.
Este bloque importa "AutoTokenizer" de "transformers" para utilizar un tokenizador preentrenado de HuggingFace.

Aplicamos la clase "AutoTokenizer" acompañada del método ".from_pretrained()" a la variable "tokenizer" para cargar un tokenizador preentrenado.
El argumento "bert-base-uncased" especifica el modelo preentrenado que se desea utilizar.
La variable "tokenizer" almacena el objeto tokenizador preentrenado cargado.

Aplicamos el método ".tokenize()" al objeto tokenizador preentrenado para obtener la lista de tokens subword correspondientes al texto de entrada.
El argumento "texto" es el texto de entrada que se desea tokenizar.
La variable "tokens_subword" almacena los tokens generados por el tokenizador.
Esta técnica permite manejar palabras desconocidas mediante fragmentación en unidades más pequeñas."""

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

tokens_subword = tokenizer.tokenize(texto)
    
tokens_subword # Muestra la lista de tokens subword generados por el tokenizador.
```

```python
# Bloque 6 — Conversión de tokens subword a IDs
"""Este bloque codifica el texto completo con el método "encode()" del tokenizador.
El argumento "texto" es la secuencia original que se desea codificar.
El parámetro "add_special_tokens=True" agrega los tokens especiales de BERT.

La variable "ids_subword" almacena una única secuencia de IDs correspondiente al texto original.
Para obtener solamente los IDs de los tokens ya generados puede utilizarse "convert_tokens_to_ids()"."""

ids_subword = tokenizer.encode(texto, add_special_tokens=True)

ids_subword # Muestra los IDs del texto con sus tokens especiales.
```

```python
# Bloque 7 — Padding y truncado
"""Este bloque aplica padding y truncado a la secuencia de IDs.
La variable "longitud_objetivo" define la longitud deseada, incluidos los tokens especiales; en este caso, 12.
El tokenizador conserva los tokens especiales al truncar y usa su propio token de padding cuando hace falta.
La variable "secuencia_fija" almacena los IDs y "mascara_atencion" distingue los tokens reales del padding."""

longitud_objetivo = 12
codificacion_fija = tokenizer(texto, padding="max_length", truncation=True, max_length=longitud_objetivo)
secuencia_fija = codificacion_fija["input_ids"]
mascara_atencion = codificacion_fija["attention_mask"]

secuencia_fija, mascara_atencion # Muestra los IDs y la máscara de atención.
```

```python
# Bloque 8 — Resultados finales
"""Este bloque organiza los resultados del proceso de tokenización.
La variable "resultado" almacena en forma de diccionario todos los datos obtenidos del proceso de tokenización.
La estructura permite visualizar las características esenciales de la tokenización y sus representaciones numéricas."""

resultado = {
    "tokens_palabras": tokens_palabras,
    "vocabulario": vocabulario,
    "secuencia_indices": secuencia_indices,
    "tokens_subword": tokens_subword,
    "ids_subword": ids_subword,
    "secuencia_fija": secuencia_fija,
    "mascara_atencion": mascara_atencion
}

resultado # Muestra el diccionario con los resultados del proceso.
```
