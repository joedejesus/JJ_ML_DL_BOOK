```python
# Bloque 1 — Definición del tensor de entrada
"""Este bloque define un tensor que representa una secuencia de embeddings.
Este bloque importa "torch" para trabajar con tensores en PyTorch.

Aplicamos la función "torch.randn()" a la variable "secuencia_embedded" para generar un tensor con valores aleatorios que simulan embeddings de tokens.
El primer argumento "1" indica el tamaño del batch.
El segundo argumento "6" indica la longitud de la secuencia de tokens.
El tercer argumento "16" indica la dimensión de los embeddings.
La variable "secuencia_embedded" contiene vectores densos que simulan representaciones de tokens antes de entrar al encoder.
Esta estructura permite aplicar las operaciones fundamentales del encoder sobre una secuencia numérica."""

import torch

secuencia_embedded = torch.randn(1, 6, 16)

secuencia_embedded # Muestra el tensor que representa la secuencia de embeddings.
```

```python
# Bloque 2 — Definición de la capa de atención multi-cabeza
"""Este bloque define una capa de atención multi-cabeza.
Este bloque importa "torch.nn" para definir capas de redes neuronales en PyTorch.

Aplicamos la clase "nn.MultiheadAttention()" a la variable "atencion_multi_cabeza" para definir una capa de atención multi-cabeza.
El parámetro "embed_dim=" indica la dimensionalidad de los embeddings de entrada; en este caso, 16 dimensiones.
El parámetro "num_heads=" define el número de cabezas de atención; en este caso, 2 cabezas.
El parámetro "batch_first=" indica si la primera dimensión de la entrada corresponde al tamaño del batch; en este caso, se establece como "True".
La variable "atencion_multi_cabeza" almacena una instancia de la capa de atención multi-cabeza configurada con los parámetros especificados.

Esta operación consiste en relacionar cada token de la secuencia con todos los demás tokens, permitiendo capturar dependencias a 
largo plazo en los datos de entrada."""

import torch.nn as nn

atencion_multi_cabeza = nn.MultiheadAttention(embed_dim=16, num_heads=2, batch_first=True)

atencion_multi_cabeza # Muestra la configuración de la capa de atención "multi-cabeza" para verificar sus parámetros.
```

```python
# Bloque 3 — Aplicación de la atención multi-cabeza
"""Este bloque aplica la atención multi-cabeza sobre la secuencia de entrada.
Asignamos de forma múltiple el resultado de la llamada a la instancia de la clase a las variables "salida_atencion" y "pesos_atencion" mediante la 
expresión "atencion_multi_cabeza(secuencia_embedded, secuencia_embedded, secuencia_embedded)".

El argumento "secuencia_embedded" es el conjunto de tensores densos de entrada que queremos procesar mediante la atención multi-cabeza.
Se pasa por triplicado para que la atención multi-cabeza pueda calcular las relaciones entre las consultas, claves y valores de la secuencia de entrada.

La variable "salida_atencion" contiene la representación contextualizada generada por la atención.
La variable "pesos_atencion" almacena los pesos de atención promediados entre cabezas por defecto."""

salida_atencion, pesos_atencion = atencion_multi_cabeza(secuencia_embedded, secuencia_embedded, secuencia_embedded)

salida_atencion, pesos_atencion # Muestra la salida de la atención multi-cabeza y los pesos de atención correspondientes.
```

```python
# Bloque 4 — Conexión residual y normalización
"""Este bloque aplica una conexión residual y normalización de capa.
Aplicamos la clase "nn.LayerNorm()" a la variable "capa_norm" para definir una capa de normalización de capa.
El argumento "16" indica la dimensionalidad de los embeddings de entrada.
La variable "capa_norm" almacena una instancia de la capa de normalización de capa.

Asignamos la llamada a la instancia de la clase a la variable "salida_norm" mediante la expresión "capa_norm(secuencia_embedded + salida_atencion)".
El argumento "secuencia_embedded" representa la secuencia de entrada original.
El argumento "salida_atencion" representa la salida de la capa de atención multi-cabeza.

Los valores se suman antes de aplicar la normalización de capa para mantener la información de la secuencia original y la salida de atención.
La variable "salida_norm" contiene la secuencia tras combinar la salida de atención con la entrada original y normalizarla."""

capa_norm = nn.LayerNorm(16)

salida_norm = capa_norm(secuencia_embedded + salida_atencion)

salida_norm # Muestra la secuencia tras combinar la salida de atención con la entrada original y normalizarla.
```

```python
# Bloque 5 — Red feed-forward del encoder
"""Este bloque define la red feed-forward del encoder.
Aplicamos la clase "nn.Sequential()" a la variable "feed_forward" para definir la red feed-forward del encoder.

El primer argumento es la llamada a la clase "nn.Linear()" y define la primera capa lineal de la red feed-forward.
En este caso la dimensionalidad de la entrada es 16 y la de la salida de la capa lineal intermedia es 32.

El segundo argumento es la llamada a la clase "nn.ReLU()" e indica la función de activación que se aplicará tras la primera capa lineal.
La función de activación introduce no linealidad en la red, permitiendo que el modelo capture relaciones más complejas entre las características de entrada.

El tercer argumento es la llamada  a la clase "nn.Linear()" y define la segunda capa lineal de la red feed-forward.
En este caso la dimensionalidad de la entrada es 32 y la de la salida de la segunda capa lineal es 16, cerrando así la estructura de la red feed-forward del encoder.
La variable "feed_forward" almacena una instancia de la red feed-forward del encoder.

Asignamos la llamada a la instancia de la clase a la variable "salida_ff" mediante la expresión "feed_forward(salida_norm)".
El argumento "salida_norm" representa la secuencia de entrada normalizada tras la combinación con la salida de atención.
La variable "salida_ff" contiene la salida de la red feed-forward del encoder aplicada a la secuencia normalizada."""

feed_forward = nn.Sequential(
    nn.Linear(16, 32),
    nn.ReLU(),
    nn.Linear(32, 16)
)

salida_ff = feed_forward(salida_norm)

salida_ff # Muestra la salida de la red feed-forward del encoder aplicada a la secuencia normalizada.
```

```python
# Bloque 6 — Segunda conexión residual y normalización
"""Este bloque aplica una segunda conexión residual y normalización de capa.
Asignamos la llamada a la instancia de la clase a la variable "salida_encoder" mediante la expresión "capa_norm(salida_norm + salida_ff)".
El argumento "salida_norm" representa la salida de la primera conexión residual y normalización de capa.
El argumento "salida_ff" representa la salida de la red feed-forward del encoder.

Los valores se suman antes de aplicar la normalización de capa para mantener la información de la entrada original y la salida de la red feed-forward.
La variable "salida_encoder" contiene la representación final normalizada del encoder.
Esta salida incorpora atención, normalización y refinamiento mediante la red feed-forward.
El ejemplo reutiliza la misma capa de normalización en ambas conexiones y no añade codificación posicional ni entrenamiento."""

salida_encoder = capa_norm(salida_norm + salida_ff)

salida_encoder # Muestra la representación final generada por el encoder para la secuencia de entrada.
```

```python
# Bloque 7 — Resultados finales
"""Este bloque organiza los resultados del procesamiento del encoder.
La variable "resultado" almacena en forma de diccionario las salidas de atención, los pesos, las normalizaciones y la salida final del encoder.
La estructura permite visualizar cada componente del mecanismo del encoder."""

resultado = {
    "salida_atencion": salida_atencion,
    "pesos_atencion": pesos_atencion,
    "salida_norm": salida_norm,
    "salida_ff": salida_ff,
    "salida_encoder": salida_encoder
}

resultado # Muestra el diccionario con los resultados del procesamiento del encoder.
```
