```python
# Bloque 1 — Definición del tensor de entrada
"""Este bloque define un tensor que representa una secuencia de embeddings.
Este bloque importa "torch" para trabajar con tensores en PyTorch.

Aplicamos la función "torch.randn()" a la variable "secuencia_embedded" para generar un tensor con valores aleatorios que simulan embeddings de tokens.
El primer argumento "1" indica el tamaño del batch.
El segundo argumento "5" indica la longitud de la secuencia de tokens.
El tercer argumento "8" indica la dimensión de los embeddings.
La variable "secuencia_embedded" almacena este tensor generado aleatoriamente.
Esta estructura permite aplicar el mecanismo de atención sobre una secuencia numérica."""

import torch

secuencia_embedded = torch.randn(1, 5, 8)

secuencia_embedded # Muestra el tensor que representa la secuencia de embeddings.
```

```python
# Bloque 2 — Definición de las proyecciones Q, K y V
"""Este bloque define las capas lineales que generan las matrices de consultas, claves y valores.
Este bloque importa "torch.nn" para definir capas lineales para las proyecciones Q, K y V.

Aplicamos la clase "nn.Linear()" a la variable "capa_Q" para definir la proyección lineal de las consultas.
Los argumentos "8, 8" indican la dimensión de entrada y la dimensión de salida de la proyección lineal.
La variable "capa_Q" almacena una instancia de la capa lineal para las consultas.

Aplicamos la clase "nn.Linear()" a la variable "capa_K" para definir la proyección lineal de las claves.
Los argumentos "8, 8" indican la dimensión de entrada y la dimensión de salida de la proyección lineal.
La variable "capa_K" almacena una instancia de la capa lineal para las claves.

Aplicamos la clase "nn.Linear()" a la variable "capa_V" para definir la proyección lineal de los valores.
Los argumentos "8, 8" indican la dimensión de entrada y la dimensión de salida de la proyección lineal.
La variable "capa_V" almacena una instancia de la capa lineal para los valores.

Llamamos a la instancia de la capa lineal "capa_Q" mediante la expresión "capa_Q(secuencia_embedded)".
El argumento "secuencia_embedded" representa la secuencia de entrada embebida que se proyecta linealmente.
La variable "Q" almacena la proyección lineal de las consultas obtenida al aplicar la capa "capa_Q" a la secuencia embebida.

Llamamos a la instancia de la capa lineal "capa_K" mediante la expresión "capa_K(secuencia_embedded)".
El argumento "secuencia_embedded" representa la secuencia de entrada embebida que se proyecta linealmente.
La variable "K" almacena la proyección lineal de las claves obtenida al aplicar la capa "capa_K" a la secuencia embebida.

Llamamos a la instancia de la capa lineal "capa_V" mediante la expresión "capa_V(secuencia_embedded)".
El argumento "secuencia_embedded" representa la secuencia de entrada embebida que se proyecta linealmente.
La variable "V" almacena la proyección lineal de los valores obtenida al aplicar la capa "capa_V" a la secuencia embebida.

Estas matrices resultantes constituyen los elementos fundamentales del mecanismo de atención y se utilizan para calcular los pesos 
de atención que determinan la ponderación relativa de cada elemento en la secuencia de entrada. Las capas están inicializadas sin entrenamiento,
por lo que los pesos no representan relevancia semántica aprendida."""

import torch.nn as nn

capa_Q = nn.Linear(8, 8)
capa_K = nn.Linear(8, 8)
capa_V = nn.Linear(8, 8)

Q = capa_Q(secuencia_embedded)
K = capa_K(secuencia_embedded)
V = capa_V(secuencia_embedded)

Q, K, V  # Muestra las proyecciones lineales de las consultas, claves y valores.
```

```python
# Bloque 3 — Cálculo de las puntuaciones de atención
"""Este bloque calcula las puntuaciones de atención mediante el producto punto entre consultas y claves.
Aplicamos el método ".size()" a la variable "K" para obtener la dimensión de las claves.
El argumento "-1" indica que queremos obtener el tamaño de la última dimensión de la matriz "K".
La variable "dim" almacena la dimensión de las claves calculada a partir de la última dimensión de "K".

Aplicamos la función "torch.matmul()" a la variable "a" para calcular el producto punto entre consultas y claves.
El primer argumento "Q" representa la matriz de consultas.
El segundo argumento "K.transpose(-2, -1)" representa la matriz de claves con sus dos últimas dimensiones transpuestas.
La variable "a" almacena el resultado del producto punto entre consultas y claves.

Aplicamos la función "torch.sqrt()" a la variable "b" para calcular la raíz cuadrada de la dimensión de las claves.
El argumento "torch.tensor(dim, dtype=torch.float32)" crea un tensor a partir de "dim" con tipo de dato "torch.float32".
La variable "b" almacena la raíz cuadrada de la dimensión de las claves.

Dividimos la matriz "a" entre la raíz cuadrada "b" para obtener las puntuaciones de atención escaladas.
La variable "puntuaciones" almacena la matriz resultante de comparar cada token con todos los demás mediante el producto punto escalado.
La operación se escala dividiendo el producto punto entre consultas y claves por la raíz cuadrada de la dimensión de las claves."""

dim = K.size(-1)

a = torch.matmul(Q, K.transpose(-2, -1))

b = torch.sqrt(torch.tensor(dim, dtype=torch.float32))

puntuaciones = (a / b)

puntuaciones # Muestra las puntuaciones de atención escaladas para cada token respecto a los demás tokens.
```

```python
# Bloque 4 — Aplicación de softmax para obtener pesos de atención
"""Este bloque aplica la función softmax a las puntuaciones de atención.
Aplicamos la función "torch.softmax()" a la variable "puntuaciones" para normalizar las puntuaciones de atención 
y obtener probabilidades que sumen 1 a lo largo de la última dimensión.

El argumento "puntuaciones" representa la matriz de puntuaciones de atención que se desea normalizar.
El parámetro "dim=" indica en qué dimensión se realizará la normalización de las puntuaciones de atención. 
En este caso, se utiliza "dim=-1" para normalizar a lo largo de la última dimensión.

La variable "pesos_atencion" contiene los pesos normalizados que indican cómo se ponderan las claves para cada consulta.
Estos pesos determinan cómo se combinarán los valores para generar la representación contextualizada."""

pesos_atencion = torch.softmax(puntuaciones, dim=-1)

pesos_atencion # Muestra los pesos de atención normalizados para cada token respecto a los demás tokens.
```

```python
# Bloque 5 — Combinación de valores ponderados
"""Este bloque combina los valores utilizando los pesos de atención.
Aplicamos la función "torch.matmul()" a los pesos de atención y la matriz de valores "V" para combinar los valores según su relevancia.
El argumento "pesos_atencion" contiene los pesos normalizados que ponderan los valores para cada consulta.
El argumento "V" representa la matriz de valores que se desea combinar según los pesos de atención.
La variable "salida_atencion" contiene la representación contextualizada resultante del mecanismo de atención.
Esta salida incorpora información global de la secuencia en cada token."""

salida_atencion = torch.matmul(pesos_atencion, V)

salida_atencion # Muestra la representación contextualizada resultante del mecanismo de atención.
```

```python
# Bloque 6 — Resultados finales
"""Este bloque organiza los resultados del mecanismo de atención.
La variable "resultado" almacena en forma de diccionario las matrices Q, K, V, las puntuaciones, los pesos y la salida final.
La estructura permite visualizar cada componente del proceso de atención de manera organizada."""

resultado = {
    "Q": Q,
    "K": K,
    "V": V,
    "puntuaciones": puntuaciones,
    "pesos_atencion": pesos_atencion,
    "salida_atencion": salida_atencion
}

resultado # Muestra el diccionario que organiza todos los componentes del mecanismo de atención.
```
