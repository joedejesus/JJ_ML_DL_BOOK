```python
# Bloque 1 — Definición del texto de entrada
"""Este bloque define un texto de entrada que será procesado mediante un modelo transformer.
La variable "texto" contiene una secuencia lingüística que permitirá observar el funcionamiento del mecanismo de atención."""

texto = "Los transformers procesan secuencias mediante mecanismos de atención."

texto # Muestra el texto de entrada para su verificación.
```

```python
# Bloque 2 — Carga del tokenizador basado en transformers
"""Este bloque carga un tokenizador compatible con modelos transformers.
Este bloque importa "AutoTokenizer" de "transformers" para utilizar un tokenizador preentrenado de HuggingFace.

Aplicamos la clase "AutoTokenizer" acompañada del método ".from_pretrained()" a la variable "tokenizador" para cargar un tokenizador preentrenado.
El argumento "bert-base-uncased" especifica el modelo preentrenado que se desea utilizar.
La variable "tokenizador" almacena el objeto tokenizador preentrenado cargado.

Aplicamos el método ".tokenize()" al objeto tokenizador preentrenado para obtener la lista de tokens subword correspondientes al texto de entrada.
El argumento "texto" es el texto de entrada que se desea tokenizar.
La variable "tokens" contiene la secuencia tokenizada en unidades subword, sin los tokens especiales añadidos al codificar.
El tokenizador elegido está entrenado para inglés, no específicamente para el texto en español del ejemplo."""

from transformers import AutoTokenizer

tokenizador = AutoTokenizer.from_pretrained("bert-base-uncased")

tokens = tokenizador.tokenize(texto)

tokens # Muestra la lista de tokens subword generados por el tokenizador.
```

```python
# Bloque 3 — Conversión del texto a IDs
"""Este bloque convierte el texto de entrada en identificadores numéricos.
Aplicamos el método ".encode()" al tokenizador para convertir el texto en IDs numéricos.
El argumento "texto" representa el texto de entrada que queremos convertir en IDs.
El parámetro "add_special_tokens=" indica si se deben agregar tokens especiales al inicio y al final de la secuencia.

La variable "ids" almacena la secuencia numérica resultante.
La conversión permite que los modelos de lenguaje trabajen con representaciones numéricas de texto, facilitando el procesamiento y 
la comprensión por parte de los algoritmos de aprendizaje automático."""

ids = tokenizador.encode(texto, add_special_tokens=True)

ids # Muestra la secuencia de IDs generada por el tokenizador.
```

```python
# Bloque 4 — Carga del modelo transformer
"""Este bloque carga un modelo transformer preentrenado.
Este bloque importa "torch" para trabajar con tensores en PyTorch.
Este bloque importa "AutoModel" de "transformers" para utilizar un modelo transformer preentrenado.
     
Aplicamos la clase "AutoModel" acompañada del método ".from_pretrained()" a la variable "modelo" para cargar un modelo preentrenado.
El argumento "bert-base-uncased" especifica el modelo preentrenado que se desea utilizar.
El parámetro "attn_implementation=" permite obtener los mapas de atención en los bloques posteriores.
El valor "eager" selecciona una implementación de atención que permite devolver los mapas al solicitarlos con "output_attentions=True".
La variable "modelo" almacena una instancia del modelo preentrenado cargado.

Aplicamos la función "torch.tensor()" a los IDs de tokens obtenidos del tokenizador preentrenado.
El argumento "[ids]" contiene la lista de IDs dentro de otra lista para representar un lote de una secuencia.
La variable "ids_tensor" contiene la secuencia de IDs convertida en un tensor de PyTorch."""

import torch
from transformers import AutoModel

modelo = AutoModel.from_pretrained("bert-base-uncased", attn_implementation="eager")

ids_tensor = torch.tensor([ids])

ids_tensor # Muestra el tensor de IDs de tokens.
```

```python
# Bloque 5 — Forward pass del transformer
"""Este bloque realiza el forward pass del modelo transformer.
Asignamos la llamada a la instancia de la clase a la variable "salida" mediante la expresión "modelo(ids_tensor)".
El argumento "ids_tensor" es el tensor de IDs de tokens que queremos procesar mediante el transformer.
La variable "salida" contiene el objeto de salida del modelo; esta llamada no solicita los estados ocultos de todas las capas.

Aplicamos el atributo ".last_hidden_state" a la variable "salida" para obtener la última capa oculta del transformer.
La variable "embeddings_contextuales" almacena la última capa oculta, que contiene los embeddings contextuales."""

salida = modelo(ids_tensor)  

embeddings_contextuales = salida.last_hidden_state    

salida, embeddings_contextuales # Muestra la salida del modelo y los embeddings contextuales.
```

```python
# Bloque 6 — Cálculo de atención mediante el modelo
"""Este bloque obtiene los pesos de atención del modelo.
Asignamos la llamada a la instancia de la clase a la variable "salida_atencion" mediante la expresión "modelo(ids_tensor, output_attentions=True).attentions".
El argumento "ids_tensor" representa el tensor de IDs de tokens que queremos procesar mediante el transformer.
El parámetro "output_attentions" indica al modelo que debe devolver los pesos de atención además de la salida habitual.

Acompañamos la llamada al modelo con el atributo ".attentions" para obtener los pesos de atención.
La variable "salida_atencion" almacena los mapas de atención generados por las capas del transformer.
Estos mapas muestran los pesos de atención entre posiciones, incluidos los tokens especiales; por sí solos no demuestran la importancia causal de cada token."""

salida_atencion = modelo(ids_tensor, output_attentions=True).attentions   

salida_atencion # Muestra los mapas de atención generados por las capas del transformer.
```

```python
# Bloque 7 — Selección de la atención de la primera capa
"""Este bloque selecciona los pesos de atención de la primera capa del transformer.
Accedemos al primer elemento de la secuencia de mapas de atención mediante el operador de indexación "[0]".
La variable "atencion_capa" almacena los pesos de la primera capa, con dimensiones de lote, cabezas, consultas y claves.
Estos pesos muestran cómo se distribuye la atención entre posiciones en la primera capa."""

atencion_capa = salida_atencion[0]

atencion_capa # Muestra los pesos de atención de la primera capa del transformer.
```

```python
# Bloque 8 — Resultados finales
"""Este bloque organiza los resultados del procesamiento mediante el modelo transformer.
La variable "resultado" almacena en forma de diccionario los tokens, IDs y las dimensiones de los embeddings contextuales y los mapas de atención.
La estructura permite visualizar las características fundamentales de los transformers."""

resultado = {
    "tokens": tokens,
    "ids": ids,
    "embeddings_contextuales_shape": embeddings_contextuales.shape, # El atributo ".shape" devuelve las dimensiones del tensor de embeddings contextuales.
    "atencion_capa_shape": atencion_capa.shape                      # El atributo ".shape" devuelve las dimensiones de los pesos de atención de la primera capa.
}

resultado # Muestra el diccionario con los resultados del procesamiento mediante el modelo transformer.
```
