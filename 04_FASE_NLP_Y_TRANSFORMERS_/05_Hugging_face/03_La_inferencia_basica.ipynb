```python
# Bloque 1 — Carga del tokenizer
"""Este bloque carga un tokenizer preentrenado desde Hugging Face.
Este bloque importa "AutoTokenizer" de "transformers" para utilizar un tokenizador preentrenado de Hugging Face.

Aplicamos la clase "AutoTokenizer", acompañada del método ".from_pretrained()", a la variable "tokenizador" para cargar un tokenizador preentrenado.
El argumento "bert-base-uncased" especifica el modelo preentrenado que se desea utilizar.
La variable "tokenizador" almacena el objeto tokenizador preentrenado cargado.

Definimos un texto de entrada que será tokenizado por el tokenizador preentrenado.
La variable "texto" contiene una secuencia de texto que será tokenizada por el tokenizador preentrenado.

Aplicamos el método ".tokenize()" al objeto tokenizador preentrenado para obtener la lista de tokens subword correspondientes al texto de entrada.
El argumento "texto" es el texto de entrada que se desea tokenizar.
La variable "tokens" almacena la secuencia tokenizada en unidades subword, sin los tokens especiales que se añadirán al codificar.
El tokenizador elegido está entrenado para inglés, no específicamente para el texto en español del ejemplo."""

from transformers import AutoTokenizer

tokenizador = AutoTokenizer.from_pretrained("bert-base-uncased")
texto = "La inferencia básica permite obtener predicciones sin entrenamiento adicional."
tokens = tokenizador.tokenize(texto)

tokens # Muestra la secuencia tokenizada en unidades subword para el texto de entrada.
```

```python
# Bloque 2 — Conversión del texto a IDs
"""Este bloque convierte el texto de entrada en identificadores numéricos.
Aplicamos el método ".encode()" al tokenizador para convertir el texto de entrada en una secuencia de IDs numéricos.
El argumento "texto" representa el texto de entrada que se desea convertir en IDs numéricos.
El parámetro "add_special_tokens=" indica si se deben agregar los tokens especiales del modelo; en este caso, se añaden al inicio y al final de la secuencia.

La variable "ids" almacena la secuencia numérica resultante del texto de entrada.
La conversión permite que los modelos de lenguaje trabajen con representaciones numéricas de texto, facilitando el procesamiento y 
su uso por parte de los algoritmos de aprendizaje automático."""

ids = tokenizador.encode(texto, add_special_tokens=True)

ids # Muestra la secuencia de IDs numéricos correspondiente al texto de entrada.
```

```python
# Bloque 3 — Construcción del tensor de entrada
"""Este bloque convierte la secuencia de IDs en un tensor.
Este bloque importa "torch" para trabajar con tensores en PyTorch.

Aplicamos la función "torch.tensor()" a la lista "[ids]" para obtener un tensor que pueda ser procesado por el modelo.
El argumento "[ids]" contiene la secuencia de IDs dentro de otra lista para representar un lote de una secuencia.
La variable "ids_tensor" almacena el tensor que será utilizado como entrada del modelo durante la inferencia."""

import torch

ids_tensor = torch.tensor([ids])

ids_tensor # Muestra el tensor correspondiente a la secuencia de IDs.   
```

```python
# Bloque 4 — Carga del modelo preentrenado
"""Este bloque carga un modelo preentrenado desde Hugging Face.
Este bloque importa "AutoModel" de "transformers" para cargar modelos preentrenados.

Aplicamos la clase "AutoModel", acompañada del método ".from_pretrained()", a la variable "modelo" para cargar un modelo preentrenado.
El argumento "bert-base-uncased" especifica el modelo preentrenado que se desea utilizar.
La variable "modelo" almacena una instancia del modelo preentrenado de BERT cargado."""

from transformers import AutoModel

modelo = AutoModel.from_pretrained("bert-base-uncased")

modelo # Muestra las características del modelo cargado.
```

```python
# Bloque 5 — Forward pass del modelo
"""Este bloque ejecuta el forward pass del modelo utilizando el tensor de entrada.
Asignamos la llamada a la instancia del modelo a la variable "salida" para obtener la salida del modelo.
El argumento "ids_tensor" representa el tensor que contiene la secuencia de IDs que se desea procesar con el modelo.
La variable "salida" contiene el objeto de salida del modelo antes de extraer los embeddings contextuales; no se solicitan los estados ocultos de todas las capas.

Aplicamos el atributo ".last_hidden_state" a la variable "salida" para obtener los embeddings contextuales generados por el modelo para cada token.
La variable "embeddings_contextuales" almacena la última capa oculta, que contiene los embeddings contextuales producidos durante la inferencia.
Esta operación extrae las representaciones de la última capa oculta para cada token de la secuencia de entrada, incluidos los tokens especiales.
El modelo base produce representaciones, no una predicción de clasificación; esta llamada tampoco desactiva el cálculo de gradientes."""

salida = modelo(ids_tensor)
embeddings_contextuales = salida.last_hidden_state

embeddings_contextuales # Muestra los embeddings contextuales generados por el modelo para cada token.
```

```python
# Bloque 6 — Resultados finales
"""Este bloque organiza los resultados del proceso de inferencia básica.
La variable "resultado" almacena en forma de diccionario los tokens, IDs y la forma de los embeddings generados por el modelo."""

resultado = {
    "tokens": tokens,
    "ids": ids,
    "embeddings_contextuales_shape": embeddings_contextuales.shape # El atributo shape devuelve la forma del tensor de embeddings contextuales.
}

resultado # Muestra el diccionario con los tokens, los IDs y la forma de los embeddings contextuales.
```
