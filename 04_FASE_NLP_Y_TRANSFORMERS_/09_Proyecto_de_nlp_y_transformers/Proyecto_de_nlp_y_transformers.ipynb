```python
# 1.1 — Importación de librerías
"""Este bloque importa las librerías necesarias para el procesamiento de texto.
- transformers: modelos y tokenizadores.
- datasets: manejo de datasets.
- torch: tensores y entrenamiento.
- Path: permite localizar la carpeta del proyecto y sus recursos locales."""

from pathlib import Path

PROJECT_DIR = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "Proyecto_de_nlp_y_transformers.ipynb").exists()
    and (path / "local_bert").exists()
    and (path / "tickets_train.jsonl").exists()
    and (path / "tickets_test.jsonl").exists()
)
MODEL_DIR = PROJECT_DIR / "local_bert"
TRAIN_PATH = PROJECT_DIR / "tickets_train.jsonl"
TEST_PATH = PROJECT_DIR / "tickets_test.jsonl"
LABEL_NAMES = {
    0: "acceso_autenticacion",
    1: "facturacion_cobros",
    2: "error_tecnico",
}

from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments
from datasets import DatasetDict, load_dataset
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
```

```python
# 1.2 — Carga del dataset
"""Este bloque carga un dataset de tickets de soporte.
- load_dataset: carga los splits de entrenamiento y prueba desde archivos locales.
- train_test_split: separa una validación estratificada a partir del conjunto de entrenamiento.
- dataset: estructura con textos y etiquetas para entrenamiento, validación y prueba."""

dataset = load_dataset("json", data_files={"train": str(PROJECT_DIR / "tickets_train.jsonl"), "test": str(PROJECT_DIR / "tickets_test.jsonl")})
train_indices, validation_indices = train_test_split(
    list(range(len(dataset["train"]))), test_size=1/3, random_state=42,
    stratify=list(dataset["train"]["label"])
)
dataset = DatasetDict({
    "train": dataset["train"].select(train_indices),
    "validation": dataset["train"].select(validation_indices),
    "test": dataset["test"]
})
dataset
```

```python
# 1.3 — Tokenización
"""Este bloque tokeniza los textos del dataset.
- BertTokenizer.from_pretrained: carga el tokenizador local del proyecto.
- tokenizer: instancia del tokenizador.
- tokenize_function: función que tokeniza los textos en lotes, con relleno y truncamiento a 128 tokens.
- El vocabulario local es pequeño y puede producir tokens desconocidos en textos nuevos."""

tokenizer = BertTokenizer.from_pretrained(str(MODEL_DIR), local_files_only=True)

def tokenize_function(batch):
    return tokenizer(batch["text"], padding="max_length", truncation=True, max_length=128)

tokenized_dataset = dataset.map(tokenize_function, batched=True)
```

```python
# 1.4 — Preparación final del dataset
"""Este bloque prepara el dataset para el entrenamiento.
- set_format: configura la salida de las columnas como tensores de PyTorch.
- columns: selección de columnas relevantes."""

tokenized_dataset.set_format(
    type="torch",
    columns=["input_ids", "attention_mask", "label"]
)
```

```python
# 2.1 — Carga del modelo BERT
"""Este bloque carga un modelo BERT para clasificación.
- BertForSequenceClassification: modelo local del proyecto adaptado a clasificación.
- LABEL_NAMES: mapea ids numéricos a categorías legibles del dominio."""

model = BertForSequenceClassification.from_pretrained(str(MODEL_DIR), local_files_only=True)
model.config.id2label = LABEL_NAMES
model.config.label2id = {label: idx for idx, label in LABEL_NAMES.items()}
```

```python
# 3.1 — Configuración del entrenamiento
"""Este bloque define los parámetros de entrenamiento.
- TrainingArguments: configuración del proceso.
- output_dir: carpeta de resultados.
- eval_strategy: evaluación del conjunto de validación por época.
- disable_tqdm: reduce ruido visual en una ejecución educativa."""

training_args = TrainingArguments(
    output_dir=str(PROJECT_DIR / "bert_tickets"),
    eval_strategy="epoch",
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=2,
    logging_strategy="no",
    save_strategy="no",
    report_to=[],
    seed=42,
    dataloader_pin_memory=False,
    disable_tqdm=True
)
```

```python
# 3.2 — Definición del Trainer
"""Este bloque crea el objeto Trainer para gestionar el entrenamiento.
- Trainer: clase que entrena el modelo.
- model: modelo BERT.
- args: parámetros de entrenamiento.
- train_dataset: conjunto de entrenamiento.
- eval_dataset: conjunto de validación.
- compute_metrics: calcula la exactitud a partir de las clases predichas."""

def compute_metrics(eval_prediction):
    logits, labels = eval_prediction
    return {"accuracy": (logits.argmax(axis=-1) == labels).mean().item()}

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["validation"],
    compute_metrics=compute_metrics
)
```

```python
# 3.3 — Ejecución del entrenamiento
"""Este bloque ejecuta el entrenamiento del modelo.
- trainer.train(): inicia el proceso de entrenamiento."""

trainer.train()
```

```python
# 4.1 — Evaluación del modelo
"""Este bloque evalúa el modelo en el conjunto de prueba.
- trainer.evaluate(): calcula pérdida y exactitud sobre el conjunto de prueba reservado.
- pd.Series: resume el estado del experimento con una salida legible."""

eval_metrics = trainer.evaluate(eval_dataset=tokenized_dataset["test"])
pd.Series(
    {
        "eval_loss": float(eval_metrics["eval_loss"]),
        "eval_accuracy": float(eval_metrics["eval_accuracy"]),
        "train_size": len(tokenized_dataset["train"]),
        "validation_size": len(tokenized_dataset["validation"]),
        "test_size": len(tokenized_dataset["test"]),
    },
    name="valor"
).round(3)
```

```python
# 5.1 — Clasificación de un ticket nuevo
"""Este bloque realiza inferencia sobre un ticket nuevo.
- tokenizer: tokeniza el texto.
- model.eval(): activa el modo de evaluación.
- torch.no_grad(): evita el cálculo innecesario de gradientes.
- torch.softmax: convierte los logits en puntuaciones normalizadas, no necesariamente calibradas.
- inputs: los tensores se trasladan al mismo dispositivo que el modelo."""

text = "El sistema no permite iniciar sesión desde ayer."
inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=128)
inputs = {name: tensor.to(model.device) for name, tensor in inputs.items()}
model.eval()
with torch.no_grad():
    outputs = model(**inputs)
    probabilities = torch.softmax(outputs.logits, dim=1).squeeze(0)

prediction = int(torch.argmax(probabilities).item())
predicted_label = LABEL_NAMES[prediction]
prediction_confidence = float(probabilities[prediction].item())

pd.Series(
    {
        "texto": text,
        "categoria_predicha": predicted_label,
        "confianza": prediction_confidence,
    },
    name="resultado"
 )
```
