```python
# Bloque 1 — Definición del vocabulario
"""Este bloque define un vocabulario básico compuesto por tokens discretos.
Utilizamos un diccionario para asignar un identificador numérico a cada token en el vocabulario.
Cada clave del diccionario representa un token del vocabulario, y su valor asociado es el identificador numérico correspondiente.
La variable "vocabulario" almacena el vocabulario en forma de diccionario.
Esta estructura permite mapear elementos simbólicos a índices que serán utilizados por la capa de embeddings."""

vocabulario = {"hola": 0, "mundo": 1, "embeddings": 2, "vector": 3}

vocabulario # Muestra el vocabulario para verificar su contenido.
```

```python
# Bloque 2 — Construcción de una secuencia de entrada numérica
"""Este bloque construye una secuencia de entrada numérica utilizando los índices del vocabulario.
Utilizamos una lista para reunir los identificadores numéricos de los tokens seleccionados del vocabulario.
En cada caso utilizamos el operador de indexación "[]" para acceder al identificador numérico correspondiente a cada token en el vocabulario.
La variable "secuencia" contiene la lista de identificadores numéricos correspondientes a los tokens de la frase de entrada.
Esta secuencia será transformada en vectores densos mediante la capa de embeddings."""

secuencia = [vocabulario["hola"], vocabulario["embeddings"], vocabulario["mundo"]]

secuencia # Muestra la secuencia de entrada para verificar su contenido.    
```

```python
# Bloque 3 — Definición de la capa de embeddings
"""Este bloque define una capa de embeddings utilizando la clase "Embedding" de PyTorch.
Este bloque importa "torch" para trabajar con tensores y "torch.nn" para definir la capa de embeddings.

Aplicamos la función "len()" al vocabulario para obtener su longitud.
La variable "num_embeddings" almacena la longitud del vocabulario en forma de número entero.

Asignamos la dimensionalidad de los vectores densos a la variable "embedding_dim" en forma de número entero; en este caso, 8.
La variable "embedding_dim" contiene la dimensionalidad de los vectores densos específicos para cada palabra.
La dimensionalidad permite representar cada palabra en un espacio continuo; las relaciones semánticas y sintácticas deben aprenderse durante el entrenamiento.

Aplicamos la clase "nn.Embedding()" a la variable "capa_embeddings" para crear la capa de embeddings que transformará índices en vectores continuos.
El parámetro "num_embeddings=" indica el tamaño del vocabulario; en este caso, la longitud del diccionario "vocabulario".
El parámetro "embedding_dim=" indica la dimensionalidad de los vectores densos; en este caso, 8.
La variable "capa_embeddings" almacena una instancia de la clase "nn.Embedding()" que se utilizará para convertir índices de palabras en vectores densos.
Los vectores de esta capa están inicializados aleatoriamente y aún no han aprendido información semántica."""

import torch
import torch.nn as nn

num_embeddings = len(vocabulario)

embedding_dim = 8

capa_embeddings = nn.Embedding(num_embeddings=num_embeddings, embedding_dim=embedding_dim)

capa_embeddings # Muestra la información de la capa de embeddings, la longitud del vocabulario y la dimensionalidad de los vectores densos.
```

```python
# Bloque 4 — Conversión de la secuencia a tensores
"""Este bloque convierte la secuencia de índices en un tensor de PyTorch.
Aplicamos la función "torch.tensor()" a la variable "secuencia" para obtener un tensor de PyTorch.
La variable "secuencia_tensor" almacena la representación numérica en formato tensorial de la secuencia.
Esta conversión es necesaria para procesar la secuencia mediante la capa de embeddings."""

secuencia_tensor = torch.tensor(secuencia)

secuencia_tensor # Muestra la representación tensorial de la secuencia.
```

```python
# Bloque 5 — Obtención de los embeddings
"""Este bloque obtiene los embeddings correspondientes a cada índice de la secuencia.
Asignamos la llamada a la instancia de la clase a la variable "vectores" mediante la expresión "capa_embeddings(secuencia_tensor)".
El argumento "secuencia_tensor" es el tensor resultante de convertir la secuencia de índices a un tensor de PyTorch.
La variable "vectores" almacena los tres vectores densos generados por la capa de embeddings, es decir, uno por cada índice de la secuencia.
Cada vector denso representa un token del vocabulario en un espacio continuo de 8 dimensiones."""

vectores = capa_embeddings(secuencia_tensor)

vectores # Muestra los embeddings correspondientes a cada índice de la secuencia, es decir, 3 embeddings de dimensión 8 cada uno.
```

```python
# Bloque 6 — Cálculo de similitud entre embeddings
"""Este bloque calcula la similitud entre dos embeddings utilizando el producto punto.
Aplicamos la función "torch.dot()" a los vectores en las posiciones 0 y 1 de la secuencia para obtener su producto punto.
Los argumentos "vectores[0]" y "vectores[1]" representan los embeddings en las posiciones 0 y 1 de la secuencia, respectivamente.
La variable "similitud" almacena el valor resultante de comparar dos vectores densos.
El producto punto depende de la magnitud y dirección de los vectores; en este ejemplo no indica una similitud semántica aprendida."""

similitud = torch.dot(vectores[0], vectores[1])

similitud # Muestra el producto punto entre los embeddings en las posiciones 0 y 1 de la secuencia.
```

```python
# Bloque 7 — Normalización de embeddings
"""Este bloque normaliza los embeddings para obtener vectores unitarios.
Aplicamos la función "F.normalize()" a la variable "vectores" para obtener los embeddings normalizados.
El parámetro "p=" indica la norma que se utilizará para la normalización, en este caso 2 (norma L2).
El parámetro "dim=" indica la dimensión a lo largo de la cual se realiza la normalización; en este caso, 1 (la dimensión de los embeddings).
La variable "vectores_normalizados" almacena los embeddings normalizados; los vectores no nulos con norma mayor que epsilon quedan unitarios.
Esta operación permite comparar vectores en términos de dirección más que de magnitud."""

import torch.nn.functional as F

vectores_normalizados = F.normalize(vectores, p=2, dim=1)

vectores_normalizados # Muestra los embeddings normalizados correspondientes a cada índice de la secuencia.
```

```python
# Bloque 8 — Resultados finales
"""Este bloque organiza los resultados del proceso de generación y análisis de embeddings.
La variable "resultado" almacena en forma de diccionario la secuencia original, los vectores densos, la similitud calculada y los vectores normalizados.
La estructura permite visualizar las características esenciales de los embeddings y su comportamiento geométrico."""

resultado = {
    "secuencia": secuencia,
    "vectores_embeddings": vectores,
    "similitud": similitud.item(), # El método "item()" convierte el valor de similitud de un tensor a un número Python.
    "vectores_normalizados": vectores_normalizados
}

resultado # Muestra el diccionario con la secuencia original, los vectores densos, la similitud calculada y los vectores normalizados.
```
