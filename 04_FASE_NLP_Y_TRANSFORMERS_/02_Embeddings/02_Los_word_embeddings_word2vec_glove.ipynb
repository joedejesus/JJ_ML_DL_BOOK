```python
# Bloque 1 — Definición del corpus
"""Este bloque define un corpus simple que servirá como base para entrenar o ilustrar embeddings.
La variable "corpus" contiene un conjunto de varias frases que representan secuencias lingüísticas.
El contenido se utilizará para entrenar Word2Vec; los vectores GloVe del ejemplo se cargan por separado y no se entrenan con este corpus."""

corpus = [
    "los modelos de lenguaje utilizan embeddings para representar palabras",
    "los embeddings capturan relaciones semánticas entre palabras",
    "word2vec aprende vectores prediciendo el contexto de una palabra",
    "glove aprende vectores a partir de coocurrencias globales"
]

corpus # Muestra el contenido del corpus, que es un conjunto de frases que representan secuencias lingüísticas.
```

```python
# Bloque 2 — Preparación del corpus para Word2Vec
"""Este bloque tokeniza el corpus en listas de palabras.
Utilizamos una comprensión de lista "[]" para convertir cada frase del corpus en una lista de tokens separados por espacios en blanco.
La subexpresión "for frase in corpus" recorre cada frase en el corpus original.
La subexpresión "frase.split()" divide cada frase en una lista de palabras (tokens).
Esta comprensión de lista permite tokenizar todo el corpus de manera concisa y eficiente.

La variable "corpus_tokenizado" almacena el corpus segmentado en tokens.
Esta estructura es compatible con el entrenamiento de Word2Vec."""

corpus_tokenizado = [frase.split() for frase in corpus]

corpus_tokenizado # Muestra el corpus tokenizado para verificar su contenido.
```

```python
# Bloque 3 — Entrenamiento de Word2Vec
"""Este bloque entrena un modelo Word2Vec utilizando gensim.
Este bloque importa "Word2Vec" de "gensim.models" para entrenar el modelo de embeddings.

Aplicamos la clase "Word2Vec()" a la variable "modelo_w2v" para entrenar el modelo de embeddings a partir del corpus tokenizado.
El parámetro "sentences=" indica el corpus tokenizado que se utilizará para entrenar el modelo; en este caso, "corpus_tokenizado".
El parámetro "vector_size=" define la dimensionalidad de los embeddings; en este caso, 16.
El parámetro "window=" controla el tamaño máximo del contexto; en este caso, 3.
El parámetro "min_count=" establece el mínimo de apariciones para incluir una palabra en el vocabulario; en este caso, 1.
La variable "modelo_w2v" almacena una instancia del modelo Word2Vec entrenado con el corpus tokenizado proporcionado.
El corpus es muy pequeño, por lo que las similitudes obtenidas no son una medida fiable de relaciones semánticas."""

from gensim.models import Word2Vec

modelo_w2v = Word2Vec(sentences=corpus_tokenizado, vector_size=16, window=3, min_count=1)

modelo_w2v # Muestra una representación del modelo Word2Vec entrenado.
```

```python
# Bloque 4 — Obtención de embeddings Word2Vec
"""Este bloque obtiene el embedding de una palabra específica utilizando el modelo entrenado.
Asignamos a la variable "vector_w2v" el resultado de la expresión "modelo_w2v.wv["palabras"]".

El atributo ".wv" y la indexación "[]" permiten acceder al vector de la palabra "palabras" almacenado en el modelo Word2Vec.
La clave "palabras" indica la palabra específica de la cual se desea obtener el embedding.
La variable "vector_w2v" almacena el vector denso correspondiente a la palabra seleccionada.
Este vector representa la palabra en el espacio continuo aprendido por Word2Vec."""

vector_w2v = modelo_w2v.wv["palabras"]

vector_w2v # Muestra el embedding de la palabra específica obtenida del modelo Word2Vec.
```

```python
# Bloque 5 — Cálculo de similitud Word2Vec
"""Este bloque calcula la similitud entre dos palabras utilizando sus embeddings Word2Vec.
Aplicamos el método ".similarity()" al modelo para calcular la similitud entre dos palabras.
Utilizamos el atributo ".wv" para acceder a los vectores de palabras almacenados en el modelo antes de aplicar el método ".similarity()".
Los argumentos "palabras" y "embeddings" representan las dos palabras cuya similitud se desea calcular.

La variable "similitud_w2v" almacena el valor resultante.
La operación calcula la similitud coseno entre los dos vectores aprendidos por Word2Vec."""

similitud_w2v = modelo_w2v.wv.similarity("palabras", "embeddings")

similitud_w2v # Muestra la similitud calculada entre las dos palabras.
```

```python
# Bloque 6 — Carga de embeddings GloVe preentrenados
"""Este bloque carga embeddings GloVe preentrenados utilizando torchtext.
Este bloque importa "GloVe" de "torchtext.vocab" para cargar embeddings preentrenados.

Aplicamos la clase "GloVe()" a la variable "glove" para cargar los embeddings preentrenados.
El parámetro "name=" especifica el conjunto de embeddings GloVe a utilizar; en este caso, "6B" (Wikipedia 2014 y Gigaword 5).
El parámetro "dim=" indica la dimensión de los vectores de embeddings; en este caso, 50.
La variable "glove" almacena los vectores GloVe preentrenados con las especificaciones indicadas.
Si los archivos no están disponibles en caché, la carga requiere descargarlos."""

from torchtext.vocab import GloVe

glove = GloVe(name="6B", dim=50)

glove # Muestra una representación de los vectores GloVe preentrenados.
```

```python
# Bloque 7 — Obtención de embeddings GloVe
"""Este bloque obtiene el embedding GloVe de una palabra específica.
Asignamos a la variable "vector_glove" el resultado de la expresión "glove["language"]".

Utilizamos corchetes "[]" en vez de paréntesis "()" para acceder al vector de la palabra específica dentro de GloVe.
La clave "language" indica la palabra específica de la cual se desea obtener el embedding.
La variable "vector_glove" almacena el vector denso correspondiente a la palabra seleccionada.
Este vector refleja patrones estadísticos globales del corpus utilizado para entrenar GloVe."""

vector_glove = glove["language"]

vector_glove # Muestra el embedding GloVe de la palabra "language".
```

```python
# Bloque 8 — Cálculo de similitud GloVe
"""Este bloque calcula el producto punto entre dos embeddings GloVe.
Este bloque importa "torch" para trabajar con tensores en PyTorch.

Aplicamos la función "torch.dot()" a los embeddings de GloVe para calcular su producto punto.
El argumento "vector_glove" representa el embedding de la palabra de interés, que se obtuvo previamente mediante "glove['language']".
El argumento "glove['model']" representa el embedding de la palabra con la que se desea comparar el embedding de interés.

La variable "similitud_glove" almacena el valor resultante del producto punto entre los embeddings de GloVe.
Este valor depende de la magnitud de los vectores y no es directamente comparable con la similitud coseno de Word2Vec."""

import torch

similitud_glove = torch.dot(vector_glove, glove["model"])

similitud_glove # Muestra el producto punto entre los embeddings de GloVe.
```

```python
# Bloque 9 — Resultados finales
"""Este bloque organiza los resultados del proceso de generación y análisis de embeddings Word2Vec y GloVe.
La variable "resultado" almacena en forma de diccionario los vectores y las medidas calculadas.
Las medidas usan operaciones distintas, corpus diferentes y dimensiones distintas, por lo que sus valores no son directamente comparables."""

resultado = {
    "vector_word2vec": vector_w2v,
    "similitud_word2vec": similitud_w2v,
    "vector_glove": vector_glove,
    "similitud_glove": similitud_glove.item()
}

resultado # Muestra el diccionario con los resultados finales de los embeddings Word2Vec y GloVe.
```
