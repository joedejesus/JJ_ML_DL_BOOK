```python
# Bloque 1 — Carga del modelo y del tokenizer
"""Este bloque carga un modelo preentrenado y su tokenizer asociado.
Este bloque importa "AutoModelForSequenceClassification" y "AutoTokenizer" de "transformers" para cargar un modelo preentrenado 
y su tokenizador correspondiente.

Aplicamos la clase "AutoModelForSequenceClassification", acompañada del método ".from_pretrained()", a la variable "modelo" para cargar un modelo preentrenado.
El argumento "bert-base-uncased" especifica el modelo preentrenado que se desea utilizar.
El parámetro "num_labels=" especifica el número de etiquetas de clasificación que el modelo debe predecir.
La variable "modelo" almacena una instancia del modelo para clasificación de secuencias. Su cabeza de clasificación se inicializa de nuevo y no se entrena en este ejemplo.

Aplicamos la clase "AutoTokenizer", acompañada del método ".from_pretrained()", a la variable "tokenizador" para cargar un tokenizador preentrenado.
El argumento "bert-base-uncased" especifica el modelo preentrenado que se desea utilizar.
La variable "tokenizador" almacena una instancia del tokenizador preentrenado lista para ser utilizada en tareas de clasificación de secuencias.
Estos son los componentes que serán guardados y cargados de nuevo para inferencia.
El modelo y el tokenizador están preentrenados para inglés, mientras que el texto de prueba está en español."""

from transformers import AutoModelForSequenceClassification, AutoTokenizer

modelo = AutoModelForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=2)
tokenizador = AutoTokenizer.from_pretrained("bert-base-uncased")

modelo, tokenizador # Muestra el modelo y el tokenizador cargados.
```

```python
# Bloque 2 — Preparación del directorio de exportación
"""Este bloque define el directorio donde se almacenarán los artefactos exportados.
Este bloque importa el módulo "os" para manejar operaciones relacionadas con el sistema de archivos, como la creación de directorios.

Asignamos un nombre al directorio donde se almacenarán los artefactos exportados.
La variable "ruta_exportacion" contiene el nombre de la ruta de salida para el modelo y el tokenizer.

Aplicamos la función "os.makedirs" a la variable "ruta_exportacion" para crear el directorio de exportación si no existe.
El parámetro "exist_ok=" indica con el valor "True" que no se generará un error si el directorio ya existe.
Esta operación permite asegurar que el directorio de exportación exista antes de intentar guardar los artefactos, 
evitando errores relacionados con rutas inexistentes."""

import os

ruta_exportacion = "modelo_exportado"

os.makedirs(ruta_exportacion, exist_ok=True)

ruta_exportacion # Muestra la ruta de exportación creada para almacenar los artefactos exportados.
```

```python
# Bloque 3 — Exportación del modelo
"""Este bloque exporta el modelo guardando sus pesos y su configuración.
Aplicamos la función "os.path.join()" a la variable "ruta_modelo" para construir la ruta completa del directorio donde se exportará el modelo.

El argumento "ruta_exportacion" indica el directorio base donde se almacenará el modelo exportado.
El argumento "modelo" es el nombre de la subcarpeta donde se guardará el modelo.
La variable "ruta_modelo" almacena la ubicación final del modelo exportado.

Aplicamos el método ".save_pretrained()" al modelo "modelo" indicando la ruta donde se desea guardar.
El argumento "ruta_modelo" indica la ruta definida en el paso anterior donde se desea guardar el modelo.
La variable "modelo" contiene el modelo que se desea guardar; su cabeza de clasificación no se ha entrenado en este ejemplo.
Este guardado no convierte por sí mismo el modelo a un formato de inferencia optimizado como ONNX."""

ruta_modelo = os.path.join(ruta_exportacion, "modelo")

modelo.save_pretrained(ruta_modelo)

ruta_modelo # Muestra la ruta donde se ha guardado el modelo.
```

```python
# Bloque 4 — Exportación del tokenizer
"""Este bloque exporta el tokenizer asociado al modelo.
Aplicamos la función "os.path.join()" a la variable "ruta_tokenizador" para construir la ruta completa del directorio donde se exportará el tokenizer.

El argumento "ruta_exportacion" indica el directorio base donde se almacenará el tokenizer exportado.
El argumento "tokenizador" es el nombre de la subcarpeta donde se guardará el tokenizer.
La variable "ruta_tokenizador" almacena la ubicación final del tokenizer exportado.

Aplicamos el método ".save_pretrained()" al tokenizer "tokenizador" indicando la ruta donde se desea guardar.
El argumento "ruta_tokenizador" indica la ruta definida en el paso anterior donde se desea guardar el tokenizer.
La variable "tokenizador" contiene el objeto del tokenizer que se desea exportar."""

ruta_tokenizador = os.path.join(ruta_exportacion, "tokenizador")

tokenizador.save_pretrained(ruta_tokenizador)

ruta_tokenizador # Muestra la ruta donde se ha guardado el tokenizer.
```

```python
# Bloque 5 — Validación de la exportación
"""Este bloque valida que el modelo y el tokenizer exportados pueden cargarse correctamente.
Este bloque importa "AutoModelForSequenceClassification" y "AutoTokenizer" de "transformers" para cargar un modelo preentrenado 
y su tokenizador correspondiente.

Aplicamos la clase "AutoModelForSequenceClassification", acompañada del método ".from_pretrained()", a la variable "modelo_cargado" para 
cargar nuestro modelo preentrenado desde la ruta especificada.
El argumento "ruta_modelo" indica la ruta donde se encuentra el modelo preentrenado que se desea cargar.
La variable "modelo_cargado" almacena una instancia de nuestro modelo preentrenado lista para ser utilizada en tareas de clasificación de secuencias.

Aplicamos la clase "AutoTokenizer", acompañada del método ".from_pretrained()", a la variable "tokenizador_cargado" para 
cargar nuestro tokenizador preentrenado desde la ruta especificada.
El argumento "ruta_tokenizador" indica la ruta donde se encuentra el tokenizer preentrenado que se desea cargar.
La variable "tokenizador_cargado" almacena una instancia de nuestro tokenizador preentrenado lista para ser utilizada en tareas de clasificación de secuencias.
Estos son los componentes cargados desde los archivos guardados; la carga por sí sola no optimiza la inferencia."""

from transformers import AutoModelForSequenceClassification, AutoTokenizer

modelo_cargado = AutoModelForSequenceClassification.from_pretrained(ruta_modelo)
tokenizador_cargado = AutoTokenizer.from_pretrained(ruta_tokenizador)


modelo_cargado, tokenizador_cargado # Muestra el modelo y el tokenizador cargados desde los archivos exportados.
```

```python
# Bloque 6 — Inferencia con el modelo cargado
"""Este bloque ejecuta inferencia utilizando el modelo cargado desde la exportación.
Definimos un texto de prueba para la inferencia del modelo.
La variable "texto_prueba" contiene el texto que se utilizará como entrada para el modelo.

Asignamos la llamada a la instancia de la clase "tokenizador_cargado()" a la variable "entrada_tokenizada" mediante la expresión 
"tokenizador_cargado(texto_prueba, return_tensors="pt")".
El argumento "texto_prueba" indica el texto que se desea tokenizar.
El parámetro "return_tensors=" indica con el valor "pt" que se desea obtener los tensores en formato PyTorch.
La variable "entrada_tokenizada" contiene los tensores tokenizados que se utilizarán como entrada para el modelo.

Asignamos el atributo ".logits" de la salida de "modelo_cargado()" a la variable "logits_exportado".
El argumento "**entrada_tokenizada" indica los tensores tokenizados que se pasan al modelo.
Los asteriscos dobles "**" se utilizan para desempaquetar el diccionario "entrada_tokenizada" y pasar sus elementos como argumentos al modelo.
La variable "logits_exportado" contiene las salidas del modelo para la prueba de inferencia.
Esta llamada no desactiva explícitamente el cálculo de gradientes."""

texto_prueba = "El servicio fue excelente."

entrada_tokenizada = tokenizador_cargado(texto_prueba, return_tensors="pt")

logits_exportado = modelo_cargado(**entrada_tokenizada).logits

logits_exportado # Muestra las salidas del modelo para la prueba de inferencia.
```

```python
# Bloque 7 — Resultados finales
"""Este bloque organiza los resultados del proceso de exportación y validación.
La variable "resultado" almacena en forma de diccionario las rutas de exportación y logits generados por el modelo cargado."""

resultado = {
    "ruta_modelo_exportado": ruta_modelo,
    "ruta_tokenizador_exportado": ruta_tokenizador,
    "logits_modelo_exportado": logits_exportado
}

resultado # Muestra los resultados del proceso de exportación y validación.
```
