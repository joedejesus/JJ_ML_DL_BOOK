```python
# Bloque 1 — Definición del texto y tokenización
"""Este bloque define un texto de entrada y aplica tokenización básica.
La variable "texto" contiene una cadena lingüística continua.
Aplicamos el método ".split()" a la variable "texto" para dividir el texto en fragmentos separados por espacios en blanco, sin separar la puntuación.
La variable "tokens" almacena una lista ordenada de unidades discretas que conforman la frase.
Esta secuencia preserva el orden temporal del lenguaje."""

texto = "Las secuencias de texto representan cadenas lingüísticas ordenadas."
tokens = texto.split()

tokens # Muestra la lista de tokens obtenida del texto para verificar la tokenización.
```

```python
# Bloque 2 — Construcción del vocabulario
"""Este bloque construye un vocabulario asignando un identificador numérico a cada token.
Utilizamos una comprensión de diccionario "{}" para asignar un índice único a cada token en la lista de tokens.
La subexpresión "token: idx" asigna a cada token el índice de su posición en la lista de tokens.
La subexpresión "for idx, token in enumerate(tokens)" recorre cada token junto con su índice en la lista de tokens.
La clase "enumerate()" enumera los elementos y proporciona un índice para cada elemento en la secuencia.

La variable "vocabulario" almacena un mapeo entre tokens y sus índices.
Si un token se repite, su índice anterior se sobrescribe y pueden quedar índices no consecutivos; en este ejemplo todos son distintos."""

vocabulario = {token: idx for idx, token in enumerate(tokens)}

vocabulario # Muestra el vocabulario con los tokens y sus índices correspondientes.
```

```python
# Bloque 3 — Conversión de tokens a índices
"""Este bloque convierte cada token del texto original en su índice correspondiente en el vocabulario.
Utilizamos una comprensión de lista "[]" para convertir cada token en su índice correspondiente en el vocabulario.
La subexpresión "vocabulario[token]" obtiene el índice correspondiente a cada token en el vocabulario definido previamente.
La subexpresión "for token in tokens" recorre cada token en la lista de tokens del texto original.
Esta comprensión de lista permite transformar el texto tokenizado en una secuencia de índices de manera eficiente.

La variable "secuencia_indices" almacena la secuencia de números enteros correspondiente al texto tokenizado.
Esta representación es compatible con modelos que operan sobre tensores."""

secuencia_indices = [vocabulario[token] for token in tokens]

secuencia_indices # Muestra la secuencia de índices correspondiente a los tokens del texto.
```

```python
# Bloque 4 — Conversión a tensor
"""Este bloque convierte la secuencia numérica en un tensor de PyTorch.
Este bloque importa "torch" para trabajar con tensores en PyTorch.
   
Aplicamos la función "torch.tensor()" a la variable "secuencia_indices" para obtener un tensor de PyTorch.
Utilizamos el método ".unsqueeze()" para agregar una dimensión adicional al tensor, convirtiéndolo en un tensor de tamaño (1, longitud_secuencia).
El argumento "0" indica que la nueva dimensión se agregará en la posición 0, es decir, al inicio del tensor.

La variable "secuencia_tensor" almacena la representación numérica de la lista en formato tensorial.
Esta conversión es necesaria para utilizar la secuencia en arquitecturas de deep learning."""

import torch      

secuencia_tensor = torch.tensor(secuencia_indices).unsqueeze(0)

secuencia_tensor # Muestra el tensor resultante obtenido a partir de la lista de índices "secuencia_indices".
```

```python
# Bloque 5 — Capa de embeddings para representar la secuencia
"""Este bloque define una capa de embeddings que transforma índices en vectores densos.
Este bloque importa "torch.nn" para definir modelos de redes neuronales recurrentes.
La variable "embedding_dim" se establece en 8, lo que significa que cada índice se representará en un vector denso de 8 dimensiones.

Aplicamos la clase "nn.Embedding()" a la variable "capa_embeddings" para crear la capa de embeddings que transformará índices en vectores densos.
El parámetro "num_embeddings=" indica el tamaño del vocabulario; en este caso, la longitud del diccionario obtenida por "len(vocabulario)".
El parámetro "embedding_dim=" indica la dimensionalidad de los vectores densos; en este caso, el valor de "embedding_dim", es decir, 8.
La variable "capa_embeddings" almacena una instancia de la clase "nn.Embedding()" que se utilizará para convertir índices de palabras en vectores densos.

Asignamos la llamada a la instancia de la clase a la variable "embeddings" mediante la expresión "capa_embeddings(secuencia_tensor)".
El argumento "secuencia_tensor" representa la secuencia de índices de palabras que se desea convertir en vectores densos.
La variable "embeddings" contiene la secuencia convertida en vectores de dimensión fija, inicializados sin entrenamiento."""

import torch.nn as nn

embedding_dim = 8

capa_embeddings = nn.Embedding(num_embeddings=len(vocabulario), embedding_dim=embedding_dim)

embeddings = capa_embeddings(secuencia_tensor)

embeddings # Muestra la secuencia convertida en vectores densos de dimensión fija.

```

```python
# Bloque 6 — Procesamiento de la secuencia mediante una RNN
"""Este bloque procesa la secuencia embebida mediante una RNN simple.
La variable "hidden_size" define la dimensionalidad del estado oculto; en este caso, 12.

Aplicamos la clase "nn.RNN()" a la variable "rnn" para definir una red neuronal recurrente simple.
El parámetro "input_size=" indica la dimensionalidad de los embeddings de entrada; en este caso, 8 dimensiones.
El parámetro "hidden_size=" define la dimensionalidad del estado oculto; en este caso, 12 dimensiones.
El parámetro "batch_first=" indica si la primera dimensión de la entrada y la salida corresponde al tamaño del batch; en este caso, se establece como "True".
La variable "rnn" almacena una instancia del modelo recurrente que procesará la secuencia de entrada.

Aplicamos la función "torch.zeros()" a la variable "estado_inicial" para inicializar el estado oculto con ceros en todas sus dimensiones.
El primer argumento "1" indica el número de capas y direcciones de la RNN en este caso.
El segundo argumento "1" indica el tamaño del lote.
El argumento "hidden_size" indica la dimensionalidad del estado oculto, en este caso 12.
La variable "estado_inicial" almacena el estado oculto inicial de la RNN, que en este caso es un tensor de ceros en todas sus dimensiones.

Asignamos de forma múltiple el resultado de la llamada a la instancia de la clase a las variables "salida_rnn" y "estado_final" mediante la expresión "rnn(embeddings, estado_inicial)".
El argumento "embeddings" es el conjunto de tensores densos de entrada que queremos procesar mediante la RNN.
El argumento "estado_inicial" es el estado oculto de partida definido previamente para la RNN.

La variable "salida_rnn" contiene las salidas generadas en cada paso en forma de tensor.
La variable "estado_final" contiene el estado oculto final tras procesar toda la secuencia."""

hidden_size = 12

rnn = nn.RNN(input_size=embedding_dim, hidden_size=hidden_size, batch_first=True)

estado_inicial = torch.zeros(1, 1, hidden_size)

salida_rnn, estado_final = rnn(embeddings, estado_inicial)

salida_rnn, estado_final  # Muestra los resultados de la RNN: las salidas de cada paso y el estado final.
```

```python
# Bloque 7 — Longitud de la secuencia y padding
"""Este bloque calcula la longitud de la secuencia y aplica padding para obtener una longitud fija.
La variable "longitud" almacena la longitud original de la secuencia de índices mediante la función "len()".
La variable "longitud_objetivo" define la longitud deseada; en este caso, 12 elementos.

Aplicamos slicing sobre la variable "secuencia_indices" para limitar su longitud a "longitud_objetivo".
El rango [:longitud_objetivo] indica que se tomarán los elementos desde el índice cero hasta el índice "longitud_objetivo" (excluyendo este último).
La variable "secuencia_padded" almacena la secuencia ajustada con la longitud deseada; en este caso, 12 elementos.

Utilizamos el operador de asignación "+=" para añadir ceros hasta alcanzar la longitud deseada.
La expresión "[0] * (longitud_objetivo - len(secuencia_padded))" genera una lista de ceros cuya longitud es la diferencia entre la longitud objetivo 
y la longitud actual de la secuencia.
La variable "secuencia_padded" almacena la secuencia final con longitud uniforme.
El índice 0 también corresponde a un token del vocabulario y la secuencia rellenada se crea después del procesamiento, por lo que la RNN no la utiliza."""

longitud = len(secuencia_indices)

longitud_objetivo = 12

secuencia_padded = secuencia_indices[:longitud_objetivo]

secuencia_padded += [0] * (longitud_objetivo - len(secuencia_padded))

secuencia_padded # Muestra la secuencia ajustada con padding para verificar la longitud uniforme.
```

```python
# Bloque 8 — Resultados finales
"""Este bloque organiza los resultados del proceso de construcción y procesamiento de secuencias de texto.
La variable "resultado" almacena en forma de diccionario los tokens, vocabulario, secuencias numéricas, embeddings y salidas de la RNN.
La estructura permite visualizar cómo se representan y manipulan secuencias en modelos de deep learning."""

resultado = {
    "tokens": tokens,
    "vocabulario": vocabulario,
    "secuencia_indices": secuencia_indices,
    "secuencia_tensor": secuencia_tensor,
    "embeddings": embeddings,
    "salida_rnn": salida_rnn,
    "secuencia_padded": secuencia_padded
}

resultado # Muestra el diccionario con los resultados finales para verificar la estructura y los valores almacenados.
```
