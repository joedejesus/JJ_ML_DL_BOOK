```python
# Bloque 1 — Definición del vocabulario y secuencia
"""Este bloque define un vocabulario básico y una secuencia de entrada.
Utilizamos un diccionario para asignar un identificador numérico a cada token en el vocabulario.
Cada clave del diccionario representa un token del vocabulario, y su valor asociado es el identificador numérico correspondiente.
La variable "vocabulario" almacena el vocabulario en forma de diccionario.

Utilizamos una lista para reunir los identificadores numéricos de los tokens seleccionados del vocabulario.
En cada caso utilizamos el operador de indexación "[]" para acceder al identificador numérico correspondiente a cada token en el vocabulario.
La variable "secuencia" representa una frase convertida en índices que serán procesados por un modelo recurrente."""

vocabulario = {"los": 0, "modelos": 1, "recurrentes": 2, "procesan": 3, "texto": 4}
secuencia = [vocabulario["los"], vocabulario["modelos"], vocabulario["recurrentes"], vocabulario["procesan"], vocabulario["texto"]]

secuencia # Muestra la secuencia de índices correspondiente a la frase de entrada.
```

```python
# Bloque 2 — Conversión de la secuencia a tensor
"""Este bloque convierte la secuencia de índices en un tensor de PyTorch.
Este bloque importa "torch" para trabajar con tensores en PyTorch.

Aplicamos la función "torch.tensor()" a la variable "secuencia" para obtener un tensor de PyTorch.
Utilizamos el método ".unsqueeze()" para agregar una dimensión adicional al tensor, convirtiéndolo en un tensor de tamaño (1, longitud_secuencia).
El argumento "0" indica que la nueva dimensión se agregará en la posición 0, es decir, al inicio del tensor.

La variable "secuencia_tensor" almacena la representación numérica de la lista en formato tensorial.
Esta conversión es necesaria para procesar la secuencia mediante un modelo recurrente."""

import torch

secuencia_tensor = torch.tensor(secuencia).unsqueeze(0)

secuencia_tensor # Muestra el tensor resultante para verificar su forma y contenido.
```

```python
# Bloque 3 — Definición de una RNN simple
"""Este bloque define una RNN simple utilizando torch.nn.
Este bloque importa "torch.nn" para definir modelos de redes neuronales recurrentes.

La variable "input_size" define la dimensionalidad de los embeddings de entrada; en este caso, 8.
La variable "hidden_size" define la dimensionalidad del estado oculto; en este caso, 16.

Aplicamos la clase "nn.RNN()" a la variable "rnn" para definir una red neuronal recurrente simple.
El parámetro "input_size=" indica la dimensionalidad de los embeddings de entrada; en este caso, 8 dimensiones.
El parámetro "hidden_size=" define la dimensionalidad del estado oculto; en este caso, 16 dimensiones.
El parámetro "batch_first=" indica si la primera dimensión de la entrada y la salida corresponde al tamaño del batch; en este caso, se establece como "True".
La variable "rnn" almacena una instancia del modelo recurrente que procesará la secuencia de entrada."""

import torch.nn as nn

input_size = 8
hidden_size = 16

rnn = nn.RNN(input_size=input_size, hidden_size=hidden_size, batch_first=True)

rnn # Muestra las características de la arquitectura del modelo recurrente.
```

```python
# Bloque 4 — Capa de embeddings para la secuencia
"""Este bloque define una capa de embeddings que transforma índices en vectores densos.
Aplicamos la clase "nn.Embedding()" a la variable "capa_embeddings" para crear la capa de embeddings que transformará índices en vectores densos.
El parámetro "num_embeddings=" indica el tamaño del vocabulario; en este caso, la longitud del diccionario obtenida por "len(vocabulario)".
El parámetro "embedding_dim=" indica la dimensionalidad de los vectores densos; en este caso, el valor de "input_size", es decir, 8.
La variable "capa_embeddings" almacena una instancia de la clase "nn.Embedding()" que se utilizará para convertir índices de palabras en vectores densos.

Asignamos la llamada a la instancia de la clase a la variable "embeddings" mediante la expresión "capa_embeddings(secuencia_tensor)".
El argumento "secuencia_tensor" representa la secuencia de índices de palabras que se desea convertir en vectores densos.
La variable "embeddings" contiene los vectores densos de 8 dimensiones, inicializados sin entrenamiento."""

capa_embeddings = nn.Embedding(num_embeddings=len(vocabulario), embedding_dim=input_size)

embeddings = capa_embeddings(secuencia_tensor)

embeddings # Muestra los embeddings generados para la secuencia de índices de palabras.
```

```python
# Bloque 5 — Procesamiento de la secuencia mediante la RNN
"""Este bloque procesa la secuencia embebida mediante la RNN.
Aplicamos la función "torch.zeros()" a la variable "estado_inicial" para crear un tensor de ceros con la forma adecuada para el estado oculto inicial de la RNN.
El primer argumento "1" indica el número de capas y direcciones de la RNN en este caso.
El segundo argumento "1" indica el tamaño del lote.
El argumento "hidden_size" indica la dimensionalidad del estado oculto, en este caso 16.
La variable "estado_inicial" almacena el estado oculto inicial de la RNN, que en este caso es un tensor de ceros en todas sus dimensiones.

Asignamos de forma múltiple el resultado de la llamada a la instancia de la clase a las variables "salida_rnn" y "estado_final" mediante la expresión "rnn(embeddings, estado_inicial)".
El argumento "embeddings" es el conjunto de tensores densos de entrada que queremos procesar mediante la RNN.
El argumento "estado_inicial" es el estado oculto de partida definido previamente para la RNN.

La variable "salida_rnn" contiene las salidas generadas en cada paso en forma de tensor.
La variable "estado_final" contiene el estado oculto final tras procesar toda la secuencia."""

estado_inicial = torch.zeros(1, 1, hidden_size)

salida_rnn, estado_final = rnn(embeddings, estado_inicial)

salida_rnn, estado_final # Muestra los resultados de la RNN: las salidas de cada paso y el estado final.
```

```python
# Bloque 6 — Definición de una LSTM
"""Este bloque define una red LSTM utilizando torch.nn.
Aplicamos la clase "nn.LSTM()" a la variable "lstm" para definir una red neuronal recurrente con memoria mejorada.
El parámetro "input_size=" indica la dimensionalidad de los embeddings de entrada; en este caso, 8 dimensiones.
El parámetro "hidden_size=" define la dimensionalidad del estado oculto; en este caso, 16 dimensiones.
El parámetro "batch_first=" indica si la primera dimensión de la entrada y la salida corresponde al tamaño del batch; en este caso, se establece como "True".

La variable "lstm" almacena la arquitectura recurrente, diseñada para facilitar el aprendizaje de dependencias a largo plazo.
La red LSTM incorpora mecanismos de puertas que controlan el flujo de información."""

lstm = nn.LSTM(input_size=input_size, hidden_size=hidden_size, batch_first=True)

lstm # Muestra las características de la arquitectura LSTM definida.
```

```python
# Bloque 7 — Procesamiento de la secuencia mediante la LSTM
"""Este bloque procesa la secuencia embebida mediante la LSTM.
Aplicamos la función "torch.zeros()" a las variables "estado_oculto" y "estado_celda" respectivamente para inicializar el estado oculto 
y el estado de celda con ceros en todas sus dimensiones.

El primer argumento "1" en cada caso indica el número de capas y direcciones de la LSTM en este ejemplo.
El segundo argumento "1" en cada caso indica el tamaño del lote.
El argumento "hidden_size" en cada caso indica la dimensionalidad de los estados; en este caso, 16.
Las variables "estado_oculto" y "estado_celda" almacenan el estado oculto inicial y el estado de celda inicial de la LSTM, 
que en este caso son tensores de ceros en todas sus dimensiones.

Asignamos de forma múltiple el resultado de la llamada a la instancia de la clase a las variables "salida_lstm" y "estados_finales" mediante 
la expresión "lstm(embeddings, (estado_oculto, estado_celda))".
El argumento "embeddings" es el conjunto de tensores densos de entrada que queremos procesar mediante la LSTM.
El argumento "estado_oculto" es el estado oculto de partida definido previamente para la LSTM.
El argumento "estado_celda" es el estado de celda de partida definido previamente para la LSTM.
En este caso se pasan los estados en forma de tupla "()", donde el primer elemento es el estado oculto y el segundo elemento es el estado de celda.

La variable "salida_lstm" almacena las representaciones generadas en cada paso.
La variable "estados_finales" almacena los estados oculto y de celda resultantes."""

estado_oculto = torch.zeros(1, 1, hidden_size)
estado_celda = torch.zeros(1, 1, hidden_size)

salida_lstm, estados_finales = lstm(embeddings, (estado_oculto, estado_celda))

salida_lstm, estados_finales # Muestra los resultados de la LSTM: las salidas en cada paso y los estados finales.
```

```python
# Bloque 8 — Definición de una GRU
"""Este bloque define una GRU utilizando torch.nn.
Aplicamos la clase "nn.GRU()" a la variable "gru" para definir una red neuronal recurrente basada en la GRU.
El parámetro "input_size=" indica la dimensionalidad de los embeddings de entrada; en este caso, 8 dimensiones.
El parámetro "hidden_size=" define la dimensionalidad del estado oculto; en este caso, 16 dimensiones.
El parámetro "batch_first=" indica si la primera dimensión de la entrada y la salida corresponde al tamaño del batch; en este caso, se establece como "True".

La variable "gru" almacena la arquitectura recurrente alternativa basada en la GRU.
La GRU simplifica la estructura de la LSTM manteniendo mecanismos de actualización y reinicio."""

gru = nn.GRU(input_size=input_size, hidden_size=hidden_size, batch_first=True)

gru # Muestra la arquitectura de la GRU definida.
```

```python
# Bloque 9 — Procesamiento de la secuencia mediante la GRU
"""Este bloque procesa la secuencia embebida mediante la GRU.
Aplicamos la función "torch.zeros()" a la variable "estado_inicial_gru" para inicializar el estado oculto con ceros en todas sus dimensiones.
El primer argumento "1" indica el número de capas y direcciones de la GRU en este caso.
El segundo argumento "1" indica el tamaño del lote.
El argumento "hidden_size" indica la dimensionalidad del estado oculto, en este caso 16.
La variable "estado_inicial_gru" define el estado oculto inicial de la GRU.

Asignamos de forma múltiple el resultado de la llamada a la instancia de la clase a las variables "salida_gru" y "estado_final_gru" mediante la expresión "gru(embeddings, estado_inicial_gru)".
El argumento "embeddings" es el conjunto de tensores densos de entrada que queremos procesar mediante la GRU.
El argumento "estado_inicial_gru" es el estado oculto de partida definido previamente para la GRU.

La variable "salida_gru" almacena las representaciones generadas en cada paso.
La variable "estado_final_gru" almacena el estado oculto resultante."""

estado_inicial_gru = torch.zeros(1, 1, hidden_size)

salida_gru, estado_final_gru = gru(embeddings, estado_inicial_gru)

salida_gru, estado_final_gru # Muestra los resultados de la GRU: las salidas de cada paso y el estado final.
```

```python
# Bloque 10 — Resultados finales
"""Este bloque organiza los resultados del procesamiento mediante RNN, LSTM y GRU.
La variable "resultado" almacena en forma de diccionario las salidas y estados finales de cada arquitectura recurrente.
La estructura permite comparar las formas y salidas de los modelos inicializados, que aún no se han entrenado."""

resultado = {
    "salida_rnn": salida_rnn,
    "estado_final_rnn": estado_final,
    "salida_lstm": salida_lstm,
    "estados_finales_lstm": estados_finales,
    "salida_gru": salida_gru,
    "estado_final_gru": estado_final_gru
}

resultado # Muestra el diccionario con los resultados de las arquitecturas recurrentes.
```
