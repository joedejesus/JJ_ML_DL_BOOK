```python
# Bloque 1 — Creación de tensores de entrada
"""Este bloque crea un tensor que representa valores de entrada para aplicar funciones de activación.
Este bloque importa "torch" para la creación y manipulación de tensores.
La variable "x" contiene valores positivos, negativos y cercanos a cero.
El tensor permite observar cómo cada función de activación transforma los datos."""

import torch

x = torch.tensor([-3.0, -1.0, 0.0, 1.0, 3.0])

x # Muestra el tensor de entrada con valores positivos, negativos y cercanos a cero.
```

```python
# Bloque 2 — Aplicación de la función ReLU
"""Este bloque aplica la función de activación ReLU.
Aplicamos la función "torch.relu()" al tensor de entrada "x" y almacenamos el resultado en la variable "relu_salida".
La función "torch.relu()" establece a cero los valores negativos y mantiene los positivos sin cambios.
La variable "relu_salida" almacena el resultado de aplicar ReLU al tensor de entrada."""

relu_salida = torch.relu(x)

relu_salida # Muestra el resultado de aplicar ReLU al tensor de entrada.
```

```python
# Bloque 3 — Aplicación de la función Sigmoid
"""Este bloque aplica la función de activación sigmoide.
Aplicamos la función "torch.sigmoid()" al tensor de entrada "x" y almacenamos el resultado en la variable "sigmoid_salida".
La función "torch.sigmoid()" transforma cada valor en un rango entre 0 y 1 mediante una curva logística.
La variable "sigmoid_salida" almacena los valores transformados por la sigmoide."""

sigmoid_salida = torch.sigmoid(x)

sigmoid_salida # Muestra el resultado de aplicar la función sigmoide al tensor de entrada.
```

```python
# Bloque 4 — Aplicación de la función Tanh
"""Este bloque aplica la función de activación tanh.
Aplicamos la función "torch.tanh()" al tensor de entrada "x" y almacenamos el resultado en la variable "tanh_salida".
La función "torch.tanh()" transforma los valores al rango entre -1 y 1 mediante una función hiperbólica.
La variable "tanh_salida" almacena los valores resultantes de aplicar tanh al tensor de entrada."""

tanh_salida = torch.tanh(x)

tanh_salida # Muestra el resultado de aplicar la función tanh al tensor de entrada.
```

```python
# Bloque 5 — Comparación de funciones de activación
"""Este bloque organiza los resultados de las funciones de activación.
La variable "comparacion" almacena en forma de diccionario las salidas de ReLU, sigmoide y tanh en formato de lista.
La estructura permite comparar cómo cada función modifica los mismos valores de entrada."""

comparacion = {
    "entrada": x.tolist(),
    "relu": relu_salida.tolist(),
    "sigmoid": sigmoid_salida.tolist(),
    "tanh": tanh_salida.tolist()
}

comparacion # Muestra el diccionario que compara las salidas de las funciones de activación.
```

```python
# Bloque 6 — Aplicación de funciones de activación dentro de una red
"""Este bloque define una red neuronal simple para demostrar el uso de funciones de activación en capas internas.
Este bloque importa "torch.nn" para definir módulos y capas de redes neuronales.

Definimos una clase personalizada llamada "RedActivaciones()" que representa un modelo de red neuronal simple.
La clase "RedActivaciones()" hereda de "nn.Module", que es la clase base utilizada para construir redes neuronales en PyTorch.

Definimos la función "def __init__()", que inicializa la estructura interna del modelo.
El parámetro "self" de la función hace referencia a la instancia actual de la clase "RedActivaciones()".

Utilizamos "super().__init__()" para invocar el constructor de la clase base "nn.Module".     
Esto asegura que la clase base se inicialice correctamente cada vez que se crea una instancia del modelo.

Aplicamos la clase "nn.Linear()" a la variable de instancia "self.capa1" para definir la primera capa lineal.
Los argumentos "1, 3" de "nn.Linear()" indican que la capa lineal recibe 1 entrada y produce 3 salidas.
Aplicamos la clase "nn.Linear()" a la variable de instancia "self.capa2" para definir la segunda capa lineal.
Los argumentos "3, 1" de "nn.Linear()" indican que la capa lineal recibe 3 entradas y produce 1 salida.

Definimos la función "def forward()" que especifica cómo fluye la entrada a través de las capas del modelo.
El método ".forward()" especifica cómo se propaga la información a través de la red.
Los parámetros "self" y "x" representan la instancia del modelo y los datos de entrada, respectivamente.

Asignamos el resultado de llamar a "self.capa1()" a la variable "x", aplicando la primera capa lineal a la entrada.
Aplicamos la función de activación "torch.relu()" a la variable "x" para introducir no linealidad en la red.
Asignamos el resultado de llamar a "self.capa2()" a la variable "x", aplicando la segunda capa lineal a la salida de la activación.
Utilizamos la instrucción "return" para devolver la salida del modelo aplicando las capas lineales a la entrada.

En todos los casos se pasa "x" como parámetro a las funciones correspondientes para que la información fluya a través de la red.
Además, la variable "x" se actualiza con la salida de cada operación, asegurando que la información fluya correctamente a través de la red.
La variable "red" almacena una instancia de la clase "RedActivaciones()", representando el modelo de red neuronal definido.

Aplicamos el método ".unsqueeze()" al tensor de entrada "x" para expandir su dimensión y que sea compatible con la primera capa lineal.
El argumento "1" indica que se está agregando una nueva dimensión en la posición 1 del tensor.
La variable "entrada_expandida" almacena el tensor de entrada "x" después de haber sido expandido con el método ".unsqueeze()".

Asignamos el resultado de llamar a la instancia de la clase a la variable "salida_red" mediante la expresión "red(entrada_expandida)".
El argumento de la función "red()" es el tensor de entrada expandido "entrada_expandida".
La variable "salida_red" almacena la salida generada por la red aún no entrenada al procesar el tensor de entrada expandido."""

import torch.nn as nn

class RedActivaciones(nn.Module):
    def __init__(self):
        super().__init__()
        self.capa1 = nn.Linear(1, 3)
        self.capa2 = nn.Linear(3, 1)

    def forward(self, x):
        x = self.capa1(x)
        x = torch.relu(x)
        x = self.capa2(x)
        return x

red = RedActivaciones()

entrada_expandida = x.unsqueeze(1)
salida_red = red(entrada_expandida)

salida_red # Muestra la salida generada por la red al procesar el tensor de entrada expandido.
```

```python
# Bloque 7 — Resultados finales
"""Este bloque organiza los resultados principales del uso de funciones de activación.
La variable "resultado" almacena en forma de diccionario las transformaciones individuales y la salida de la red neuronal.
La estructura permite visualizar el papel de las funciones de activación en el procesamiento de datos."""

resultado = {
    "comparacion_funciones": comparacion,
    "salida_red": salida_red.tolist()
}

resultado # Muestra el diccionario que contiene las transformaciones individuales y la salida de la red neuronal.
```
