```python
# Bloque 1 — Creación de tensores básicos
"""Este bloque crea tensores fundamentales en PyTorch.
Este bloque importa "torch" para crear y manipular tensores en PyTorch.
La variable "t1" almacena un tensor unidimensional.
La variable "t2" almacena un tensor bidimensional.
La creación de tensores constituye la base de todas las operaciones en PyTorch."""

import torch

t1 = torch.tensor([1.0, 2.0, 3.0])
t2 = torch.tensor([[1.0, 2.0], [3.0, 4.0]])

t1, t2 # Muestra los tensores "t1" y "t2".
```

```python
# Bloque 2 — Inspección de dimensiones
"""Este bloque inspecciona las dimensiones de los tensores.
Aplicamos el método ".dim()" al tensor "t1" para obtener su número de dimensiones.
La variable "dim_t1" almacena el número de dimensiones del tensor "t1".
Accedemos al atributo ".shape" del tensor "t2" para obtener su forma.
La variable "shape_t2" almacena la forma del tensor "t2".
La inspección permite comprender la estructura interna de cada tensor."""

dim_t1 = t1.dim()
shape_t2 = t2.shape

dim_t1, shape_t2 # Muestra el número de dimensiones de "t1" y la forma de "t2".
```

```python
# Bloque 3 — Operaciones elementales
"""Este bloque aplica operaciones elementales sobre tensores.
La variable "suma" almacena la suma del tensor "t1" con otro tensor literal unidimensional.
La variable "producto" almacena el producto elemento a elemento del tensor "t2" con otro tensor literal bidimensional.
Nos aseguramos de que las dimensiones de los tensores sean compatibles para realizar las operaciones correspondientes.
Las operaciones permiten manipular datos dentro del flujo computacional."""

suma = t1 + torch.tensor([4.0, 5.0, 6.0])
producto = t2 * torch.tensor([[2.0, 2.0], [2.0, 2.0]])

suma, producto # Muestra la suma y el producto de los tensores.
```

```python
# Bloque 4 — Cambio de tipo y dispositivo
"""Este bloque transforma el tipo de dato del tensor y lo mueve a un dispositivo específico.
Aplicamos el método ".to()" al tensor "t1" para cambiar el tipo de dato.
El argumento corresponde al tipo de dato que queremos obtener, en este caso un flotante de 32 bits (float32).
La variable "t1_float" almacena el tensor "t1" convertido a tipo float32.

Aplicamos el método ".to()" al tensor "t1" para mover el tensor a la CPU.
El argumento "cpu" indica que queremos mover el tensor a la CPU.
La variable "t1_cpu" almacena el tensor "t1" en CPU.
Con la configuración predeterminada, "t1" ya es float32 y está en la CPU; estas operaciones permiten gestionar su tipo y ubicación."""

t1_float = t1.to(torch.float32)
t1_cpu = t1.to("cpu")

t1_float, t1_cpu # Muestra el tensor convertido a tipo float32 y el tensor en CPU.
```

```python
# Bloque 5 — Activación del cálculo de gradientes
"""Este bloque activa el cálculo de gradientes en un tensor.
Añadimos el parámetro "requires_grad=True" al crear el tensor "x" para habilitar el cálculo de gradientes.
La variable "x" almacena un tensor con gradientes habilitados.
La activación del gradiente permite registrar operaciones para retropropagación."""

x = torch.tensor([2.0, 3.0], requires_grad=True)

x # Muestra el tensor "x" con gradientes habilitados.
```

```python
# Bloque 6 — Construcción del grafo computacional
"""Este bloque construye un grafo computacional mediante operaciones sobre x.
La operación "y = x * x + 2 * x" construye un grafo computacional donde "y" depende de "x".
La variable "y" almacena el resultado de una transformación diferenciable.
El grafo computacional permite calcular gradientes mediante retropropagación."""

y = x * x + 2 * x

y # Muestra el valor de y.
```

```python
# Bloque 7 — Cálculo de gradientes
"""Este bloque calcula los gradientes mediante retropropagación.
Aplicamos el método ".backward()" sobre la suma de "y" para calcular su gradiente respecto a "x".
La suma de "y" se obtiene con el método ".sum()".

Accedemos al atributo "grad" de "x", donde se almacenan los gradientes calculados mediante retropropagación.
La variable "gradientes" almacena los valores de gradiente obtenidos.
El cálculo de gradientes constituye el núcleo del entrenamiento en PyTorch."""


y.sum().backward()
gradientes = x.grad

gradientes # Muestra los valores de gradiente obtenidos.
```

```python
# Bloque 8 — Estructura final de resultados
"""Este bloque organiza los elementos principales del trabajo con tensores.
La variable "resultado" almacena tensores, dimensiones, operaciones y gradientes en forma de diccionario.
La estructura permite visualizar los componentes esenciales del manejo de tensores en PyTorch."""

resultado = {
    "tensores": {
        "t1": t1.tolist(),
        "t2": t2.tolist()
    },
    "dimensiones": {
        "dim_t1": dim_t1,
        "shape_t2": list(shape_t2) # Convertimos la tupla en una lista para mayor claridad.
    },
    "operaciones": {
        "suma": suma.tolist(),
        "producto": producto.tolist()
    },
    "gradientes": gradientes.tolist()
}

resultado # Muestra la estructura final de resultados organizada en forma de diccionario.
```
