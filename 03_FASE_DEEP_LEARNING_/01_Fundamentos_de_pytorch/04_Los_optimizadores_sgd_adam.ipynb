```python
# Bloque 1 — Creación de parámetros y función de pérdida
"""Este bloque crea un tensor de parámetros y un valor objetivo para una función de pérdida simple.
Este bloque importa "torch" para trabajar con tensores y optimizadores en PyTorch.

Añadimos el parámetro "requires_grad=True" al crear el tensor "w" para habilitar el cálculo de gradientes.
La variable "w" almacena un parámetro entrenable con gradientes habilitados en forma de tensor.

La variable "objetivo" define el valor que el parámetro debe aproximar en forma de tensor.
La función de pérdida permite medir la diferencia entre el parámetro y el objetivo."""

import torch

w = torch.tensor([5.0], requires_grad=True)
objetivo = torch.tensor([0.0])

w, objetivo # Muestra los valores actuales de w y objetivo.
```

```python
# Bloque 2 — Definición del optimizador SGD
"""Este bloque define un optimizador SGD.
Este bloque importa "torch.optim" para definir y utilizar optimizadores en PyTorch.

Aplicamos la clase "SGD" de "optim" a la variable "optimizador_sgd" para definir un optimizador que actualizará el parámetro "w".
El argumento "[w]" indica los parámetros que el optimizador debe actualizar.
El parámetro "lr=" define la tasa de aprendizaje del optimizador, en este caso 0.1.

La variable "optimizador_sgd" almacena el optimizador con una tasa de aprendizaje fija.
La definición del optimizador permite que PyTorch actualice los parámetros del modelo durante el entrenamiento."""

import torch.optim as optim

optimizador_sgd = optim.SGD([w], lr=0.1)

optimizador_sgd # Muestra la configuración del optimizador SGD.
```

```python
# Bloque 3 — Paso de entrenamiento con SGD
"""Este bloque ejecuta un paso de entrenamiento utilizando SGD. 
El método "zero_grad()" restablece los gradientes acumulados de los parámetros antes de calcular los nuevos gradientes.
La variable "optimizador_sgd" almacena el optimizador que se utiliza para actualizar los parámetros durante el entrenamiento.

Los métodos "pow()" y "sum()" se utilizan para calcular la función de pérdida respecto a la diferencia de "w" y "objetivo".
(w - objetivo) es la diferencia entre el parámetro actual "w" y el valor objetivo "objetivo".
La variable "perdida" almacena el valor de la función de pérdida antes de calcular los gradientes.

El método "backward()" calcula los gradientes de la función de pérdida con respecto a los parámetros.
Los gradientes calculados después de aplicar "backward()" se almacenan en "w.grad".

El método "step()" actualiza los parámetros utilizando los gradientes calculados.
La variable "optimizador_sgd" contiene el optimizador que se utiliza para actualizar los parámetros.

Los métodos "clone()" y "detach()" se utilizan para crear una copia del tensor sin los gradientes asociados.
La variable "w_sgd" almacena la copia del parámetro "w" después de actualizarlo con el optimizador.
El optimizador actualiza el parámetro mediante un paso de gradiente descendente."""

optimizador_sgd.zero_grad()

perdida = (w - objetivo).pow(2).sum()
perdida.backward()

optimizador_sgd.step()
w_sgd = w.clone().detach()

optimizador_sgd, perdida, w_sgd # Muestra el optimizador, la pérdida y el valor actualizado de w_sgd.
```

```python
# Bloque 4 — Reinicio del parámetro
"""Este bloque reinicia el parámetro para entrenarlo con Adam.
La variable "w" almacena nuevamente el valor inicial con gradientes habilitados en forma de tensor.
El reinicio permite comparar el comportamiento de ambos optimizadores."""

w = torch.tensor([5.0], requires_grad=True)

w # Muestra el valor reiniciado del parámetro.
```

```python
# Bloque 5 — Definición del optimizador Adam
"""Este bloque define un optimizador Adam.
Aplicamos la clase "Adam" de "optim" a la variable "optimizador_adam" para definir un optimizador que actualizará el parámetro "w".
El argumento "[w]" indica los parámetros que el optimizador debe actualizar.
El parámetro "lr=" define la tasa de aprendizaje del optimizador, en este caso 0.1.

La variable "optimizador_adam" almacena el optimizador con una tasa de aprendizaje base de 0.1.
Adam utiliza momentos y ajustes adaptativos para actualizar los parámetros."""

optimizador_adam = optim.Adam([w], lr=0.1)

optimizador_adam # Muestra la instancia del optimizador Adam.
```

```python
# Bloque 6 — Paso de entrenamiento con Adam
"""Este bloque ejecuta un paso de entrenamiento utilizando Adam.
El método "zero_grad()" restablece los gradientes acumulados de los parámetros antes de calcular los nuevos gradientes.
La variable "optimizador_adam" almacena el optimizador que se utiliza para actualizar los parámetros durante el entrenamiento.

Los métodos "pow()" y "sum()" se utilizan para calcular la función de pérdida respecto a la diferencia de "w" y "objetivo".
(w - objetivo) es la diferencia entre el parámetro actual "w" y el valor objetivo "objetivo".
La variable "perdida" almacena el valor de la función de pérdida antes de calcular los gradientes.

El método "backward()" calcula los gradientes de la función de pérdida con respecto a los parámetros.
Los gradientes calculados después de aplicar "backward()" se almacenan en "w.grad".

El método "step()" actualiza los parámetros utilizando los gradientes calculados.
La variable "optimizador_adam" contiene el optimizador que se utiliza para actualizar los parámetros.

Los métodos "clone()" y "detach()" se utilizan para crear una copia del tensor sin los gradientes asociados.
La variable "w_adam" almacena la copia del parámetro "w" después de actualizarlo con el optimizador.
El optimizador actualiza el parámetro utilizando momentos y ajustes adaptativos."""

optimizador_adam.zero_grad()

perdida = (w - objetivo).pow(2).sum()
perdida.backward()

optimizador_adam.step()
w_adam = w.clone().detach()

optimizador_adam, perdida, w_adam # Muestra el optimizador, la pérdida y el valor actualizado de w_adam.
```

```python
# Bloque 7 — Estructura final de resultados
"""Este bloque organiza los resultados obtenidos con ambos optimizadores.
La variable "resultado" almacena los valores del parámetro tras un paso de SGD y Adam en forma de diccionario.
La estructura permite comparar el efecto de cada optimizador sobre la actualización del parámetro."""

resultado = {
    "valor_con_sgd": w_sgd.tolist(),
    "valor_con_adam": w_adam.tolist(),
}

resultado # Muestra el diccionario con los valores actualizados del parámetro para cada optimizador.
```
