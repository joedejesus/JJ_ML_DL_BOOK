```python
# Bloque 1 — Creación de tensores
"""Este bloque crea tensores básicos en PyTorch.
Este bloque importa "torch" para crear y manipular tensores en PyTorch.
La variable "t1" almacena un tensor unidimensional.
La variable "t2" almacena un tensor bidimensional.
La creación de tensores constituye el fundamento de todas las operaciones en PyTorch."""

import torch

t1 = torch.tensor([1.0, 2.0, 3.0])
t2 = torch.tensor([[1.0, 2.0], [3.0, 4.0]])

t1, t2 # Muestra los tensores creados.
```

```python
# Bloque 2 — Operaciones sobre tensores
"""Este bloque aplica operaciones elementales sobre tensores.
La variable "suma" almacena la suma del tensor "t1" con otro tensor literal unidimensional.
La variable "producto" almacena el producto elemento a elemento del tensor "t2" con otro tensor literal bidimensional.
Nos aseguramos de que las dimensiones de los tensores sean compatibles para realizar las operaciones correspondientes.
Las operaciones permiten manipular datos. En este caso no se registran para calcular gradientes."""

suma = t1 + torch.tensor([4.0, 5.0, 6.0])
producto = t2 * torch.tensor([[2.0, 2.0], [2.0, 2.0]])

suma, producto # Muestra los resultados de las operaciones.
```

```python
# Bloque 3 — Activación del cálculo de gradientes
"""Este bloque activa el cálculo de gradientes en un tensor.
Añadimos el parámetro "requires_grad=True" al crear el tensor "x" para habilitar el cálculo de gradientes.
La variable "x" almacena un tensor con gradientes habilitados.
La activación del gradiente permite que PyTorch registre operaciones para retropropagación."""

x = torch.tensor([2.0, 3.0], requires_grad=True)

x # Muestra el tensor "x" con gradientes habilitados.
```

```python
# Bloque 4 — Construcción del grafo computacional
"""Este bloque construye un grafo computacional mediante operaciones sobre x.
La operación "y = x * x + 2 * x" construye un grafo computacional donde "y" depende de "x".
La variable "y" almacena el resultado de una transformación diferenciable.
El grafo computacional permite calcular gradientes mediante retropropagación."""

y = x * x + 2 * x

y # Muestra el tensor "y" que depende de "x" en el grafo computacional.
```

```python
# Bloque 5 — Cálculo de gradientes
"""Este bloque calcula los gradientes mediante retropropagación.
Aplicamos el método ".backward()" sobre la suma de "y" para calcular su gradiente respecto a "x".
La suma de "y" se obtiene con el método ".sum()".

Accedemos al atributo "grad" de "x", donde se almacenan los gradientes calculados mediante retropropagación.
La variable "gradientes" almacena los valores de gradiente obtenidos.
El cálculo de gradientes constituye el núcleo del entrenamiento en PyTorch."""


y.sum().backward()
gradientes = x.grad

gradientes # Muestra los gradientes calculados.
    
```

```python
# Bloque 6 — Definición de un módulo
"""Este bloque define un módulo básico en PyTorch.
Este bloque importa "torch.nn" para definir módulos y capas de redes neuronales.

Definimos una clase personalizada llamada "Modelo()" que representa nuestro modelo de red neuronal.
La clase "Modelo()" hereda de "nn.Module", que es la clase base utilizada para construir modelos en PyTorch.

Definimos la función "def __init__()", que inicializa la estructura interna del modelo.
El parámetro "self" de la función hace referencia a la instancia actual del modelo, en este caso de la clase "Modelo".

Utilizamos "super().__init__()" para invocar el constructor de la clase base "nn.Module".
Esto asegura que la clase base se inicialice correctamente cada vez que se crea una instancia del modelo.

Aplicamos la clase "nn.Linear()" a la variable de instancia "self.capa" para definir una capa lineal.
Los argumentos "2, 1" de "nn.Linear()" indican que la capa lineal recibe 2 entradas y produce 1 salida.

Definimos la función "def forward()" que especifica cómo fluye la entrada a través de las capas del modelo.
El método ".forward()" especifica cómo se propaga la información a través de la red.
Los parámetros "self" y "entrada" representan la instancia del modelo y los datos de entrada, respectivamente.

Utilizamos la instrucción "return" para devolver la salida del modelo aplicando la capa lineal a la entrada.
La definición del módulo permite que el modelo procese los datos de entrada y genere la salida correspondiente."""

import torch.nn as nn

class Modelo(nn.Module):

    def __init__(self):
        super().__init__()
        self.capa = nn.Linear(2, 1)

    def forward(self, entrada):
        return self.capa(entrada)

# Este es el procedimiento estándar para definir un módulo en PyTorch.
```

```python
# Bloque 7 — Instanciación y ejecución del modelo
"""Este bloque instancia el modelo y ejecuta una predicción.
La variable "modelo" almacena una instancia del módulo definido, en este caso de la clase "Modelo".
La variable "entrada" almacena los datos de entrada en forma de tensor que se pasarán al modelo para generar una predicción.
La variable "salida" almacena el resultado de llamar al modelo con los datos de entrada como argumento.
La ejecución del modelo permite aplicar el flujo de datos a través de sus capas."""

modelo = Modelo()
entrada = torch.tensor([1.0, 2.0])

salida = modelo(entrada)

salida # Muestra la salida del modelo para la entrada dada.
       
```

```python
# Bloque 8 — Estructura final de resultados
"""Este bloque organiza los elementos principales de los fundamentos de PyTorch.
La variable "resultado" almacena tensores, operaciones, gradientes y predicciones del modelo en un diccionario estructurado.
La estructura permite visualizar los componentes esenciales del flujo de trabajo en PyTorch."""

resultado = {
    "tensores": {
        "t1": t1.tolist(),
        "t2": t2.tolist()
    },
    "operaciones": {
        "suma": suma.tolist(),
        "producto": producto.tolist()
    },
    "gradientes": gradientes.tolist(),
    "prediccion_modelo": salida.tolist()
}

resultado # Muestra la estructura final de resultados con tensores, operaciones, gradientes y predicciones del modelo.
```
