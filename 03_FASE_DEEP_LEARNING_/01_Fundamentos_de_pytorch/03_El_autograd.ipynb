<h1 style="margin:0 0 14px 0; font-size:34px; line-height:1.2; font-weight:700; color:#ffffff;">El Autograd</h1>
<p style="margin:0; color:#cbd5e1; font-size:18px;">Autograd es el mecanismo con el que PyTorch calcula derivadas de forma automática durante el entrenamiento. Gracias a él, no hace falta derivar manualmente cada operación del modelo para ajustar sus parámetros.</p>

    <div style="color:#f8fafc;">La retropropagación práctica en deep learning depende de poder calcular gradientes de manera fiable y automática.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Qué vas a aprender aquí</div>
    <div style="color:#f8fafc;">A activar el seguimiento de gradientes, construir operaciones diferenciables y recuperar derivadas sobre tensores.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Regla práctica que debes recordar</div>
    <div style="color:#f8fafc;">Cuando algo falle en entrenamiento, revisar qué tensores requieren gradiente y cómo se construyó el grafo suele ahorrar mucho tiempo.</div>

    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Contexto mínimo</div>
    <div style="color:#f8fafc;">Autograd registra operaciones sobre tensores y después aplica la regla de la cadena para obtener gradientes.</div>

```python
# Bloque 1 — Creación de tensores con gradiente
"""Este bloque crea un tensor con gradientes habilitados.
Este bloque importa "torch" para trabajar con tensores y autograd en PyTorch.
Añadimos el parámetro "requires_grad=True" al crear el tensor "x" para habilitar el cálculo de gradientes.
La variable "x" almacena un tensor cuyas operaciones se registrarán para autograd.
La activación del gradiente permite construir el grafo computacional dinámico."""

import torch

x = torch.tensor([2.0, 3.0], requires_grad=True)

x # Muestra el tensor "x" con gradientes habilitados.
```

```python
# Bloque 2 — Construcción del grafo computacional
"""Este bloque construye un grafo computacional mediante operaciones sobre x.
La operación "y = x * x + 2 * x" construye un grafo computacional donde "y" depende de "x".
La variable "y" almacena el resultado de una transformación diferenciable.
El grafo computacional registra cada operación para permitir retropropagación."""

y = x * x + 2 * x

y # Muestra el tensor "y" que resulta de las operaciones sobre "x".
```

```python
# Bloque 3 — Cálculo de gradientes
"""Este bloque calcula los gradientes mediante retropropagación.
Aplicamos el método ".backward()" sobre la suma de "y" para calcular su gradiente respecto a "x".
La suma de "y" se obtiene con el método ".sum()".

Accedemos al atributo "grad" de "x", donde se almacenan los gradientes calculados mediante retropropagación.
La variable "gradientes" almacena los valores de gradiente obtenidos.
El cálculo de gradientes constituye el núcleo del sistema autograd."""

y.sum().backward()
gradientes = x.grad

gradientes # Muestra los gradientes calculados respecto a "x".
```

```python
# Bloque 4 — Reinicio y cálculo del gradiente
"""Este bloque muestra cómo se reinicia el gradiente antes de una nueva retropropagación.
Asignamos el valor "None" a "x.grad" para eliminar los gradientes calculados anteriormente y evitar su acumulación.

Asignamos el valor "x * 3" a "y" para construir un nuevo grafo de cómputo a partir de "x".
La variable "y" almacena el resultado de la operación sobre "x" y forma parte del nuevo grafo de cómputo.

Aplicamos el método ".backward()" sobre la suma de "y" para calcular su gradiente respecto a "x".
La suma de "y" se obtiene con el método ".sum()".

Accedemos al atributo "grad" de "x", donde se almacenan los gradientes calculados mediante retropropagación.
La variable "acumulado" almacena los nuevos gradientes, sin sumar los anteriores porque se reiniciaron.
Este reinicio permite comenzar un nuevo ciclo de entrenamiento sin conservar gradientes anteriores."""

x.grad = None
y = x * 3

y.sum().backward()
acumulado = x.grad

acumulado # Muestra el gradiente calculado después de reiniciar y realizar la nueva retropropagación.
```

```python
# Bloque 5 — Desactivación del autograd
"""Este bloque desactiva el sistema autograd temporalmente.
Aplicamos la clase ".no_grad()" de "torch" para desactivar temporalmente el autograd.
La palabra clave "with" se utiliza para crear un contexto en el que el autograd está desactivado.

Definimos una variable "z" que almacena el resultado de una operación realizada sin registrar gradientes.
La variable "z" almacena un tensor calculado sin registrar operaciones a partir del tensor "x".
La desactivación permite ejecutar cálculos en el entorno "with torch.no_grad()" que no deben formar parte del grafo computacional."""

with torch.no_grad():
    z = x * 5

z # Muestra el tensor calculado sin registrar gradientes.
```

```python
# Bloque 6 — Verificación del estado del gradiente
"""Este bloque verifica si un tensor participa en autograd.
Accedemos al atributo "requires_grad" del tensor "x" para verificar si tiene habilitado el cálculo de gradientes.
La variable "requiere_grad" almacena el estado del tensor "x".
La verificación permite comprobar si se registran operaciones para calcular gradientes del tensor.
En este caso "x" tiene habilitados los gradientes, por lo que "requiere_grad" será True."""

requiere_grad = x.requires_grad

requiere_grad # Muestra si el tensor "x" tiene habilitados los gradientes.
```

```python
# Bloque 7 — Estructura final de resultados
"""Este bloque organiza los elementos principales del sistema autograd.
La variable "resultado" almacena gradientes y el estado de seguimiento en forma de diccionario.
La clave "gradientes_acumulados" contiene los gradientes posteriores al reinicio, no la suma con los anteriores.
La estructura permite visualizar los componentes esenciales del funcionamiento de autograd."""

resultado = {
    "gradientes_iniciales": gradientes.tolist(),
    "gradientes_acumulados": acumulado.tolist(),
    "tensor_sin_gradiente": z.tolist(),
    "requiere_grad": requiere_grad
}

resultado # Muestra la estructura final de resultados en forma de diccionario.
```
