```python
# Bloque 1 — Tensores de predicciones y etiquetas
"""Este bloque crea tensores que representan predicciones y etiquetas reales.
Este bloque importa "torch" para trabajar con tensores en PyTorch.
La variable "predicciones" contiene valores continuos definidos manualmente en forma de tensor.
La variable "etiquetas" contiene los valores reales en forma de tensor asociados a cada muestra.
Estos tensores permiten calcular métricas básicas de evaluación."""

import torch

predicciones = torch.tensor([0.2, 0.7, 0.9, 0.1])
etiquetas = torch.tensor([0.0, 1.0, 1.0, 0.0])

predicciones, etiquetas # Muestra los tensores de predicciones y etiquetas reales.
```

```python
# Bloque 2 — Cálculo de accuracy
"""Este bloque calcula la métrica accuracy.
Aplicamos el método ".float()" a la expresión "(predicciones > 0.5)" para convertir el resultado booleano en un tensor de tipo float.
La variable "pred_binarias" almacena valores binarios: 1 si la predicción es mayor que 0.5 y 0 en caso contrario.

Aplicamos el método ".float()" a la expresión "(pred_binarias == etiquetas)" para convertir la comparación booleana en un tensor de tipo float.
La variable "aciertos" almacena un tensor de tipo float donde cada elemento indica si la predicción correspondiente es correcta (1.0) o incorrecta (0.0).

Aplicamos el método ".mean()" a la variable "aciertos" para calcular la proporción de predicciones correctas.
La variable "accuracy" almacena la proporción de predicciones correctas."""

pred_binarias = (predicciones > 0.5).float()
aciertos = (pred_binarias == etiquetas).float()
accuracy = aciertos.mean()

accuracy # Muestra la proporción de predicciones correctas.
```

```python
# Bloque 3 — Cálculo del error cuadrático medio (MSE)
"""Este bloque calcula el error cuadrático medio.
Aplicamos el método ".mean()" a la expresión "(predicciones - etiquetas) ** 2" para calcular el promedio de los cuadrados de las diferencias
entre las predicciones y las etiquetas.

La variable "mse" almacena el promedio de los cuadrados de las diferencias entre las predicciones y las etiquetas.
El MSE cuantifica la magnitud del error en tareas de regresión o en salidas continuas."""

mse = ((predicciones - etiquetas) ** 2).mean()

mse # Muestra el error cuadrático medio.
```

```python
# Bloque 4 — Cálculo del error absoluto medio (MAE)
"""Este bloque calcula el error absoluto medio.
Aplicamos los métodos ".abs()" y ".mean()" a la expresión "(predicciones - etiquetas)" para obtener los valores absolutos de las diferencias entre
predicciones y etiquetas y calcular su promedio.

La variable "mae" almacena el promedio de los valores absolutos de las diferencias entre predicciones y etiquetas.
El MAE mide la desviación promedio sin penalizar con más fuerza los errores grandes."""

mae = (predicciones - etiquetas).abs().mean()

mae # Muestra el error absoluto medio.
```

```python
# Bloque 5 — Organización de métricas
"""Este bloque organiza las métricas básicas calculadas.
La variable "resultado" almacena, en forma de diccionario, accuracy, MSE y MAE en formato numérico.
En todos los casos se usa el método ".item()" para convertir los tensores de un solo elemento en valores escalares.
La estructura permite visualizar de forma conjunta las métricas fundamentales de evaluación."""

resultado = {
    "accuracy": float(accuracy.item()), 
    "mse": float(mse.item()),
    "mae": float(mae.item())
}

resultado # Muestra el diccionario con las métricas organizadas.    
```
