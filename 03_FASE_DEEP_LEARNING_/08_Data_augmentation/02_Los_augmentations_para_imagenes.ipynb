```python
# Bloque 1 — Creación de una imagen de ejemplo
"""Este bloque crea una imagen de ejemplo para aplicar augmentations.
Este bloque importa "Image" de "PIL" para crear imágenes y "torchvision.transforms" para aplicar data augmentation.

Aplicamos la función "Image.new()" a la variable "imagen" para crear una imagen uniforme con un color específico.
El argumento "RGB" indica que la imagen tendrá tres canales de color: rojo, verde y azul.
El argumento "(128, 128)" especifica el tamaño de la imagen en píxeles; en este caso, 128 píxeles de ancho y 128 píxeles de alto.
El parámetro "color=" especifica el color de fondo de la imagen; en este caso, un color RGB con valores (150, 100, 200).

La variable "imagen" contiene la imagen original.
El uso de una imagen uniforme permite observar algunos cambios, aunque no muestra el efecto de todas las transformaciones sobre detalles visuales."""

from PIL import Image
import torchvision.transforms as T

imagen = Image.new("RGB", (128, 128), color=(150, 100, 200))

imagen # Muestra la imagen original.
```

```python
# Bloque 2 — Definición de augmentations básicos
"""Este bloque define un conjunto de augmentations básicos para imágenes.
Aplicamos la clase "T.Compose()" a la variable "augmentations" para definir una secuencia de transformaciones de data augmentation.

La clase "T.RandomRotation()" permite rotar la imagen aleatoriamente dentro de un rango de grados especificado.
El parámetro "degrees=" especifica el rango de grados dentro del cual se puede rotar la imagen aleatoriamente.

La clase "T.RandomHorizontalFlip()" permite invertir la imagen horizontalmente con una probabilidad determinada.
El parámetro "p=" especifica la probabilidad con la que se realiza la inversión horizontal.

La clase "T.ColorJitter()" permite ajustar el brillo, contraste, saturación y matiz de la imagen.
El parámetro "contrast=" especifica el rango de ajuste del contraste de la imagen; en este caso solo se ajusta el contraste.

La clase "T.RandomResizedCrop()" permite recortar aleatoriamente una región de la imagen y redimensionarla a un tamaño específico.
El parámetro "size=" especifica el tamaño de la imagen resultante.
El parámetro "scale=" especifica el rango de proporciones del área original para el recorte aleatorio.

Las transformaciones se aplican en secuencia y pueden generar una variación de la imagen original.
La variable "augmentations" almacena la secuencia de transformaciones."""

augmentations = T.Compose([
    T.RandomRotation(degrees=25),
    T.RandomHorizontalFlip(p=0.5),
    T.ColorJitter(contrast=0.4),
    T.RandomResizedCrop(size=128, scale=(0.7, 1.0))
])

augmentations # Muestra la secuencia de transformaciones definidas.
```

```python
# Bloque 3 — Aplicación de los augmentations
"""Este bloque aplica los augmentations definidos sobre la imagen original.
Asignamos la llamada a la instancia de la clase a la variable "imagen_aug" mediante la expresión "augmentations(imagen)".
El argumento "imagen" es la imagen original sobre la cual se aplicarán las transformaciones.
La variable "imagen_aug" contiene una versión modificada de la imagen.
Cada ejecución puede producir una variación distinta debido a la aleatoriedad de las transformaciones."""

imagen_aug = augmentations(imagen)

imagen_aug # Muestra la imagen transformada.
```

```python
# Bloque 4 — Conversión a tensores
"""Este bloque convierte la imagen original y la imagen transformada en tensores.
Aplicamos la clase "T.ToTensor()" a la variable "to_tensor" para crear un objeto conversor de imágenes a tensores.
La variable "to_tensor" contiene el objeto conversor, que se utiliza para transformar las imágenes en tensores.

Asignamos la llamada a la instancia de la clase a la variable "tensor_original" mediante la expresión "to_tensor(imagen)".
El argumento "imagen" es la imagen original que se convertirá en tensor.

Asignamos la llamada a la instancia de la clase a la variable "tensor_aug" mediante la expresión "to_tensor(imagen_aug)".
El argumento "imagen_aug" es la imagen transformada que se convertirá en tensor.
Las variables "tensor_original" y "tensor_aug" almacenan los tensores resultantes de las imágenes original y transformada, respectivamente."""

to_tensor = T.ToTensor() 

tensor_original = to_tensor(imagen)
tensor_aug = to_tensor(imagen_aug)

tensor_original, tensor_aug # Muestra los tensores resultantes de la imagen original y la imagen transformada.
```

```python
# Bloque 5 — Resultados finales
"""Este bloque organiza los resultados del proceso de augmentations para imágenes.
La variable "resultado" almacena en forma de diccionario los tensores de la imagen original y de la imagen transformada, convertidos en listas.
La estructura permite visualizar el resultado de las transformaciones; su validez para entrenamiento depende del problema."""

resultado = {
    "tensor_original": tensor_original.tolist(),
    "tensor_aug": tensor_aug.tolist()
}

resultado # Muestra el diccionario con los tensores de la imagen original y la imagen transformada convertidos en listas.
```
