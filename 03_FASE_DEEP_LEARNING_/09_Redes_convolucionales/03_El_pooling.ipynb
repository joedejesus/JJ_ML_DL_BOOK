```python
# Bloque 1 — Creación de un mapa de características sintético
"""Este bloque crea un tensor que simula un mapa de características generado por una convolución.
Este bloque importa "torch" para crear y manipular tensores en PyTorch.
Aplicamos la función "torch.rand()" a la variable "mapa" para generar valores aleatorios que simulan un mapa de características.
La variable "mapa" contiene un tensor de cuatro dimensiones (lote, canales, altura, anchura), con valores aleatorios entre 0 y 1.
Este tensor permite demostrar cómo el pooling reduce la dimensionalidad espacial."""

import torch

mapa = torch.rand(1, 1, 4, 4)

mapa # Muestra el mapa de características sintético generado aleatoriamente.
```

```python
# Bloque 2 — Definición de una operación de max pooling
"""Este bloque define una operación de max pooling utilizando nn.MaxPool2d.
Este bloque importa "torch.nn" para definir capas de redes neuronales convolucionales en PyTorch.

Aplicamos la clase "nn.MaxPool2d()" a la variable "pool_max" para definir un objeto de max pooling.
El parámetro "kernel_size=" indica el tamaño de la ventana de pooling (2x2 en este caso).

La variable "pool_max" contiene una instancia de la clase "nn.MaxPool2d()", que representa el objeto de max pooling definido.
La ventana de tamaño 2x2, con paso 2 por defecto, divide el mapa en regiones locales y selecciona el valor máximo de cada región.
La estructura permite observar cómo el pooling conserva el máximo de cada región."""

import torch.nn as nn

pool_max = nn.MaxPool2d(kernel_size=2)

pool_max # Muestra la configuración del objeto de max pooling definido.
```

```python
# Bloque 3 — Aplicación del max pooling
"""Este bloque aplica la operación de max pooling sobre el mapa de características.
Asignamos la llamada a la instancia de la clase a la variable "salida_max" mediante la expresión "pool_max(mapa)".
El argumento "mapa" es el mapa de características de entrada sobre el cual se aplicará el max pooling.
La variable "salida_max" contiene el mapa reducido resultante.
Cada valor representa el máximo dentro de una región local del mapa original."""

salida_max = pool_max(mapa)

salida_max # Muestra el mapa reducido resultante después de aplicar el max pooling.
```

```python
# Bloque 4 — Definición de una operación de average pooling
"""Este bloque define una operación de average pooling utilizando nn.AvgPool2d.
Aplicamos la clase "nn.AvgPool2d()" a la variable "pool_avg" para definir un objeto de average pooling.
El parámetro "kernel_size=" indica el tamaño de la ventana de pooling (2x2 en este caso).

La variable "pool_avg" contiene una instancia de la clase "nn.AvgPool2d()", que representa el objeto de average pooling definido.
La ventana de tamaño 2x2, con paso 2 por defecto, divide el mapa en regiones locales y calcula el promedio de cada región.
La estructura permite comparar cómo diferentes funciones de agregación afectan la representación."""

pool_avg = nn.AvgPool2d(kernel_size=2)

pool_avg # Muestra la configuración del objeto de average pooling.
```

```python
# Bloque 5 — Aplicación del average pooling
"""Este bloque aplica la operación de average pooling sobre el mapa de características.
Asignamos la llamada a la instancia de la clase a la variable "salida_avg" mediante la expresión "pool_avg(mapa)".
El argumento "mapa" es el mapa de características de entrada sobre el cual se aplicará el average pooling.
La variable "salida_avg" contiene el mapa reducido resultante.
Cada valor representa el promedio dentro de una región local del mapa original."""

salida_avg = pool_avg(mapa)

salida_avg # Muestra el mapa reducido resultante del average pooling.
```

```python
# Bloque 6 — Resultados finales
"""Este bloque organiza los resultados del proceso de pooling.
La variable "resultado" almacena en forma de diccionario el mapa original y los resultados de max pooling y average pooling, convertidos en listas.
La estructura permite visualizar cómo ambas operaciones reducen la dimensionalidad espacial mediante distintas agregaciones."""

resultado = {
    "mapa_original": mapa.tolist(),
    "max_pooling": salida_max.tolist(),
    "avg_pooling": salida_avg.tolist()
}

resultado # Muestra el diccionario con los resultados del pooling.  
```
