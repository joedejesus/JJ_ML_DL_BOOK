```python
# Bloque 1 — Transformación básica sobre tensores 
"""Este bloque define una transformación elemental aplicada a tensores.
Este bloque importa "torch" para trabajar con tensores.

Definimos la función "def transformar_multiplicar(x)" para aplicar una transformación de multiplicación a los tensores.
El parámetro "x" representa el tensor que será transformado.
La función recibe un tensor y devuelve una versión escalada mediante la instrucción "return x * 2".

La variable "tensor_original" almacena un tensor con los valores iniciales.
La variable "tensor_transformado" contiene el resultado de llamar a la función "transformar_multiplicar()" con el tensor original como argumento.
Las transformaciones permiten modificar los datos antes de que sean utilizados por un modelo."""

import torch

def transformar_multiplicar(x):
    return x * 2

tensor_original = torch.tensor([1.0, 2.0, 3.0])
tensor_transformado = transformar_multiplicar(tensor_original)

tensor_transformado # Muestra el tensor transformado.
```

```python
# Bloque 2 — Transformación mediante clase personalizada
"""Este bloque define una transformación encapsulada en una clase personalizada.
Definimos una clase personalizada llamada "TransformacionSuma" que aplica una suma constante a cada tensor recibido.

Definimos la función "def __init__()", que inicializa la estructura interna de la clase.
El parámetro "self" de la función hace referencia a la instancia actual de la clase.
El parámetro "valor" representa la constante que se sumará a cada tensor recibido.
Dentro de la función "__init__()" definimos la variable de instancia "self.valor" y le asignamos el valor del parámetro "valor".

Definimos la función "def __call__()", que permite que la instancia de la clase sea llamada como una función.
El parámetro "self" de la función hace referencia a la instancia actual de la clase.
El parámetro "x" representa el tensor al que se le aplicará la transformación.
Dentro de la función "__call__()" retornamos el resultado de sumar la variable de instancia "self.valor" al tensor "x".

La variable "transformacion_suma" es una instancia de la clase "TransformacionSuma()" que contiene el valor de la constante 
que se sumará a cada tensor recibido; en este caso el valor de la constante es 5.

La variable "tensor_transformado_suma" contiene el resultado de llamar a la instancia de la clase "TransformacionSuma()" aplicada 
al tensor original pasado como argumento.
Las transformaciones basadas en clases permiten parametrizar y reutilizar operaciones."""

class TransformacionSuma:
    def __init__(self, valor):
        self.valor = valor

    def __call__(self, x):
        return x + self.valor

transformacion_suma = TransformacionSuma(5)
tensor_transformado_suma = transformacion_suma(tensor_original)

tensor_transformado_suma # Muestra el tensor transformado mediante la suma constante.
```

```python
# Bloque 3 — Composición de transformaciones
"""Este bloque compone múltiples transformaciones en una secuencia.
La variable "transformaciones" contiene la lista de transformaciones que se aplicarán al tensor original, las cuales se corresponden con las
transformaciones definidas en los bloques anteriores.

Definimos la función "def aplicar_transformaciones(x, transformaciones)" para aplicar una secuencia de transformaciones a un tensor.
El parámetro "x" representa el tensor que será transformado.
El parámetro "transformaciones" es una lista de funciones o instancias de clases que representan las transformaciones que se aplicarán al tensor "x".

Dentro de la función, se itera sobre cada transformación de la lista "transformaciones" y, en cada iteración del 
bucle, "t" se aplica al tensor "x" mediante la expresión "x = t(x)".
Además, se retorna el valor final del tensor transformado con la expresión "return x".
La función recibe un tensor y una lista de transformaciones, y aplica cada transformación "t" al tensor "x" en el orden especificado.

La variable "tensor_compuesto" contiene el resultado de llamar a la función "aplicar_transformaciones()".
El argumento "tensor_original" representa el tensor que será transformado.
El argumento "transformaciones" es la lista de transformaciones que se aplicarán al tensor "tensor_original".
La composición permite encadenar operaciones de preprocesamiento de forma controlada."""

transformaciones = [transformar_multiplicar, transformacion_suma]

def aplicar_transformaciones(x, transformaciones):
    for t in transformaciones:
        x = t(x) # "t" representa cada transformación de la lista y se llama con el tensor "x".
    return x

tensor_compuesto = aplicar_transformaciones(tensor_original, transformaciones)

tensor_compuesto # Muestra el tensor resultante después de aplicar todas las transformaciones.
```

```python
# Bloque 4 — Transformaciones dentro de un dataset personalizado
"""Este bloque integra transformaciones dentro de un dataset personalizado.
Este bloque importa "Dataset" de "torch.utils.data" para crear un dataset personalizado.

Definimos una clase personalizada llamada "DatasetConTransformacion()" que representa un dataset con transformaciones aplicadas a los datos.
La clase "DatasetConTransformacion()" hereda de "Dataset", que es la clase base utilizada para crear datasets en PyTorch.

Definimos la función "def __init__()", que inicializa la estructura interna del dataset.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
El parámetro "transformacion" de la función representa la transformación que se aplicará a los datos del dataset.

Dentro de la función "__init__()" definimos dos tensores que representan los datos y las etiquetas del dataset.
En ambos casos asignamos los tensores a las variables de instancia "self.datos" y "self.etiquetas", respectivamente.
También asignamos el argumento "transformacion" a la variable de instancia "self.transformacion" para que pueda ser 
utilizado posteriormente al acceder a los elementos del dataset.

Definimos la función "def __len__()", que devuelve la cantidad de elementos del dataset.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
Utilizamos "return len(self.datos)" para devolver la longitud de los datos del dataset.

Definimos la función "def __getitem__()", que permite acceder a un elemento específico del dataset mediante un índice.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
El parámetro "idx" representa el índice del elemento que se desea obtener.
Utilizamos "return self.transformacion(self.datos[idx]), self.etiquetas[idx]" para retornar los datos transformados 
y la etiqueta correspondiente al índice proporcionado.

La variable "dataset_transformado" es una instancia de la clase "DatasetConTransformacion()" que recibe una 
transformación como argumento y la aplica a los datos cada vez que se accede a un elemento del dataset.
La transformación es "transformar_multiplicar", que multiplica cada tensor por 2.
El dataset con transformación permite modificar los datos automáticamente al acceder a cada muestra del 
propio dataset con su índice correspondiente."""

from torch.utils.data import Dataset

class DatasetConTransformacion(Dataset):
    def __init__(self, transformacion):
        self.datos = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
        self.etiquetas = torch.tensor([0, 1])
        self.transformacion = transformacion

    def __len__(self):
        return len(self.datos)

    def __getitem__(self, idx):
        return self.transformacion(self.datos[idx]), self.etiquetas[idx]

dataset_transformado = DatasetConTransformacion(transformar_multiplicar)

dataset_transformado[0] # Muestra el primer elemento del dataset transformado, que incluye los datos transformados y la etiqueta correspondiente.
```

```python
# Bloque 5 — Dataloader con datos transformados
"""Este bloque crea un dataloader para el dataset transformado.
Este bloque importa "DataLoader" de "torch.utils.data" para crear un dataloader a partir del dataset transformado.

Aplicamos la clase "DataLoader()" a la variable "dataloader" para crear un dataloader a partir del dataset transformado.
El argumento "dataset_transformado" corresponde a la instancia del dataset transformado que se desea cargar en el dataloader.
El parámetro "batch_size=" especifica el tamaño máximo de los lotes que se obtendrán del dataloader.
El parámetro "shuffle=" indica si debe mezclarse el orden de las muestras antes de crear los lotes.
La variable "dataloader" almacena un objeto iterable con datos transformados.

Creamos un contenedor de lista llamado "lotes_transformados" para almacenar los lotes obtenidos del dataloader.
La variable "lotes_transformados" almacena los lotes generados durante la iteración.

Iteramos sobre el dataloader mediante un bucle "for" para obtener los lotes de datos y etiquetas.
La variable "datos" corresponde al tensor que contiene las características del lote en esa iteración.
La variable "etiquetas" corresponde al tensor que contiene las etiquetas asociadas al lote en esa iteración.
    
Aplicamos el método ".append()" a la lista "lotes_transformados" para añadir cada lote obtenido del dataloader en cada iteración del bucle.
La tupla añadida contiene los tensores "datos" y "etiquetas" correspondientes al lote en cada iteración.

La variable "lotes_transformados" contiene todos los lotes obtenidos del dataloader durante la iteración.
El dataloader permite verificar que las transformaciones se aplican correctamente."""

from torch.utils.data import DataLoader

dataloader = DataLoader(dataset_transformado, batch_size=1, shuffle=False)

lotes_transformados = []

for datos, etiquetas in dataloader:
    lotes_transformados.append((datos, etiquetas))

lotes_transformados # Muestra todos los lotes transformados obtenidos del dataloader en forma de lista de tuplas (datos, etiquetas).
```

```python
# Bloque 6 — Transformación condicional
"""Este bloque define una transformación condicional.
Definimos la función "def transformar_condicional(x)" para aplicar una transformación condicional al tensor "x".
El parámetro "x" representa el tensor que será transformado.

Dentro de la función, se aplica una operación distinta según el valor del tensor "x".
Se utiliza la expresión condicional "(x * 10) if x.sum() > 5 else (x - 1)" para determinar qué operación aplicar.

En el caso de que la suma de los elementos del tensor sea mayor que 5, se multiplica el tensor por 10; de lo 
contrario, se le resta 1.

La variable "tensor_condicional_1" contiene el resultado de llamar a la función "transformar_condicional" con un tensor literal como argumento.
La variable "tensor_condicional_2" contiene el resultado de llamar a la función "transformar_condicional" con un tensor literal como argumento.
Las transformaciones condicionales permiten adaptar el preprocesamiento a cada muestra."""

def transformar_condicional(x):
    return (x * 10) if x.sum() > 5 else (x - 1)

tensor_condicional_1 = transformar_condicional(torch.tensor([1.0, 1.0]))
tensor_condicional_2 = transformar_condicional(torch.tensor([4.0, 4.0]))


tensor_condicional_1, tensor_condicional_2 # Muestra los tensores transformados condicionalmente.
```

```python
# Bloque 7 — Estructura final de resultados
"""Este bloque organiza los elementos principales del trabajo con transformaciones.
La variable "resultado" almacena, en forma de diccionario, tensores transformados, transformaciones compuestas y lotes del dataloader.
La estructura permite visualizar el efecto de las transformaciones sobre los datos."""

resultado = {
    "tensor_original": tensor_original.tolist(),
    "transformacion_multiplicar": tensor_transformado.tolist(),
    "transformacion_suma": tensor_transformado_suma.tolist(),
    "transformacion_compuesta": tensor_compuesto.tolist(),
    "lotes_transformados": [(d.tolist(), e.tolist()) for d, e in lotes_transformados],
    "transformacion_condicional": {
        "caso_1": tensor_condicional_1.tolist(),
        "caso_2": tensor_condicional_2.tolist()
    }
}

resultado # Muestra la estructura final de resultados con todos los tensores transformados y los lotes del dataloader.
```
