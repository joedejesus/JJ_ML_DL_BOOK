```python
# Bloque 1 — Transformación de escalado
"""Este bloque define una transformación básica de escalado.
Este bloque importa "torch" para trabajar con tensores.

Definimos la función "def escalar(x, factor)" para aplicar una transformación de escalado a los tensores.
El parámetro "x" representa el tensor que será transformado.
El parámetro "factor" representa el valor por el cual se multiplicará el tensor.
La función recibe un tensor y un factor de escalado, y devuelve una versión escalada mediante la instrucción "return x * factor".

La variable "tensor_original" almacena un tensor con los valores iniciales.
La variable "tensor_escalado" contiene el resultado de llamar a la función "escalar()" con los argumentos correspondientes.
El argumento "tensor_original" representa el tensor que se desea escalar.
El argumento "2" representa el factor de escalado aplicado al tensor original.
El escalado permite ajustar la magnitud de los datos antes del entrenamiento."""

import torch

def escalar(x, factor):
    return x * factor

tensor_original = torch.tensor([1.0, 2.0, 3.0])
tensor_escalado = escalar(tensor_original, 2)

tensor_escalado # Muestra el tensor escalado.
```

```python
# Bloque 2 — Transformación de suma constante
"""Este bloque define una transformación que suma un valor constante al tensor.
Definimos la función "def sumar(x, valor)" para aplicar una transformación de suma constante a los tensores.
El parámetro "x" representa el tensor que será transformado.
El parámetro "valor" representa el valor constante que se sumará al tensor.
La función recibe un tensor y un valor constante, y devuelve una versión sumada mediante la instrucción "return x + valor".

La variable "tensor_sumado" contiene el resultado de llamar a la función "sumar()" con los argumentos correspondientes.
El argumento "tensor_original" representa el tensor al que se desea sumar un valor constante.
El argumento "5" representa el valor constante que se sumará al tensor original.
La suma constante permite modificar la distribución de los valores de entrada."""

def sumar(x, valor):
    return x + valor

tensor_sumado = sumar(tensor_original, 5)

tensor_sumado # Muestra el tensor sumado.
```

```python
# Bloque 3 — Transformación de normalización
"""Este bloque define una transformación de normalización.
Definimos la función "def normalizar(x)" para aplicar una transformación de normalización a los tensores.
El parámetro "x" representa el tensor que será transformado.

Dentro de la función se calculan la media y la desviación estándar del tensor.
La variable "media" almacena la media del tensor mediante el método ".mean()" aplicado al tensor "x".
La variable "desviacion" almacena la desviación estándar muestral del tensor mediante el método ".std()" aplicado al tensor "x".
La función retorna una versión normalizada del tensor mediante la instrucción "return (x - media) / desviacion".

La variable "tensor_normalizado" contiene el resultado de llamar a la función "normalizar()" con el argumento correspondiente.
El argumento "tensor_original" representa el tensor que se desea normalizar.
La función rechaza tensores con menos de dos elementos o con una desviación nula o no finita."""

def normalizar(x):
    if x.numel() < 2:
        raise ValueError("La normalización requiere al menos dos elementos.")
    media = x.mean()
    desviacion = x.std()
    if not torch.isfinite(desviacion) or desviacion == 0:
        raise ValueError("La desviación estándar debe ser finita y distinta de cero.")
    return (x - media) / desviacion

tensor_normalizado = normalizar(tensor_original)

tensor_normalizado # Muestra el tensor normalizado.
```

```python
# Bloque 4 — Transformación de conversión de tipo
"""Este bloque convierte un tensor a un tipo de dato distinto.
Definimos la función "def convertir_tipo(x)" para aplicar una transformación de conversión de tipo a los tensores.
El parámetro "x" representa el tensor que será transformado.

Dentro de la función se realiza la conversión de tipo del tensor.
Aplicamos el método ".to()" al tensor "x" para convertir su tipo de dato a float32.
El parámetro "torch.float32" indica el tipo de dato al que se desea convertir el tensor.
La función retorna una versión del tensor convertida a float32 mediante la instrucción "return x.to(torch.float32)".

La variable "tensor_convertido" contiene el resultado de llamar a la función "convertir_tipo()" con el argumento correspondiente.
El argumento "tensor_original" representa el tensor que se desea convertir.
Con la configuración predeterminada, el tensor original ya es float32. La conversión permite adaptar el tipo de dato."""

def convertir_tipo(x):
    return x.to(torch.float32)

tensor_convertido = convertir_tipo(tensor_original)

tensor_convertido # Muestra el tensor convertido a float32.
```

```python
# Bloque 5 — Composición de transformaciones básicas
"""Este bloque compone varias transformaciones básicas en una secuencia.
La variable "transformaciones" contiene la lista de transformaciones que se aplicarán al tensor original.
Estas transformaciones son dos funciones anónimas "lambda" y la función "normalizar" definida previamente.

La primera función anónima (lambda) escala el tensor multiplicándolo por 3 mediante la función previa "escalar(x, 3)".
La segunda función anónima (lambda) suma -1 al tensor mediante la función previa "sumar(x, -1)".
En ambos casos se utiliza "lambda x" para fijar el argumento adicional y obtener una transformación que recibe el tensor.
La función "normalizar" se aplica al tensor para normalizar sus valores y fue definida en el bloque anterior.

Definimos la función "def aplicar_transformaciones(x, transformaciones)" para aplicar una secuencia de transformaciones a un tensor.
El parámetro "x" representa el tensor que será transformado.
El parámetro "transformaciones" es una lista de funciones o instancias de clases que representan las transformaciones que se aplicarán al tensor "x".

Dentro de la función, se itera sobre cada transformación de la lista "transformaciones" y, en cada iteración del
bucle, "t" se aplica al tensor "x" mediante la expresión "x = t(x)".
Además, se retorna el valor final del tensor transformado con la expresión "return x".
La función recibe un tensor y una lista de transformaciones, y aplica cada transformación "t" al tensor "x" en el orden especificado.

La variable "tensor_compuesto" contiene el resultado de llamar a la función "aplicar_transformaciones()".
El argumento "tensor_original" representa el tensor que será transformado.
El argumento "transformaciones" es la lista de transformaciones que se aplicarán al tensor "tensor_original".
La composición permite construir pipelines de preprocesamiento simples y reproducibles."""

transformaciones = [
    lambda x: escalar(x, 3),
    lambda x: sumar(x, -1),
    normalizar
]

def aplicar_transformaciones(x, transformaciones):
    for t in transformaciones:
        x = t(x) # "t" representa cada transformación de la lista y se llama con el tensor "x".
    return x

tensor_compuesto = aplicar_transformaciones(tensor_original, transformaciones)

tensor_compuesto # Muestra el tensor compuesto después de aplicar todas las transformaciones.
         
```

```python
# Bloque 6 — Transformaciones dentro de un dataset
"""Este bloque integra transformaciones básicas dentro de un dataset personalizado.
Este bloque importa "Dataset" de "torch.utils.data" para crear un dataset personalizado.

Definimos una clase personalizada llamada "DatasetBasico()" que representa un dataset con transformaciones aplicadas a los datos.
La clase "DatasetBasico()" hereda de "Dataset", que es la clase base utilizada para crear datasets en PyTorch.

Definimos la función "def __init__()", que inicializa la estructura interna del dataset.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
El parámetro "transformacion" de la función representa la transformación que se aplicará a los datos del dataset.

Dentro de la función "__init__()" definimos dos tensores que representan los datos y las etiquetas del dataset.
En ambos casos asignamos los tensores a las variables de instancia "self.datos" y "self.etiquetas", respectivamente.
También asignamos el argumento "transformacion" a la variable de instancia "self.transformacion" para que pueda ser 
utilizado posteriormente al acceder a los elementos del dataset.

Definimos la función "def __len__()", que devuelve la cantidad de elementos del dataset.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
Utilizamos "return len(self.datos)" para devolver la longitud de los datos del dataset.

Definimos la función "def __getitem__()", que permite acceder a un elemento específico del dataset mediante un índice.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
El parámetro "idx" representa el índice del elemento que se desea obtener.
Utilizamos "return self.transformacion(self.datos[idx]), self.etiquetas[idx]" para retornar los datos transformados 
y la etiqueta correspondiente al índice proporcionado.

La variable "dataset_transformado" es una instancia de la clase "DatasetBasico()" que recibe una 
transformación como argumento y la aplica a los datos cada vez que se accede a un elemento del dataset.
En este caso, la transformación es una función lambda que utiliza la función "escalar()" para multiplicar cada tensor por 2.
El dataset transformado permite modificar los datos automáticamente al acceder a cada muestra del propio dataset con su índice correspondiente."""

from torch.utils.data import Dataset

class DatasetBasico(Dataset):
    def __init__(self, transformacion):
        self.datos = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
        self.etiquetas = torch.tensor([0, 1])
        self.transformacion = transformacion

    def __len__(self):
        return len(self.datos)

    def __getitem__(self, idx):
        return self.transformacion(self.datos[idx]), self.etiquetas[idx]

dataset_transformado = DatasetBasico(lambda x: escalar(x, 2))

dataset_transformado[0] # Muestra el primer elemento del dataset transformado, que incluye los datos transformados y la etiqueta correspondiente.
```

```python
# Bloque 7 — Dataloader con transformaciones básicas
"""Este bloque crea un dataloader para el dataset transformado.
Este bloque importa "DataLoader" de "torch.utils.data" para crear un dataloader a partir del dataset transformado.

Aplicamos la clase "DataLoader()" a la variable "dataloader" para crear un dataloader a partir del dataset transformado.
El argumento "dataset_transformado" corresponde a la instancia del dataset transformado que se desea cargar en el dataloader.
El parámetro "batch_size=" especifica el tamaño máximo de los lotes que se obtendrán del dataloader.
El parámetro "shuffle=" indica si debe mezclarse el orden de las muestras antes de crear los lotes.
La variable "dataloader" almacena un objeto iterable con datos transformados.

Creamos un contenedor de lista llamado "lotes_transformados" para almacenar los lotes obtenidos del dataloader.
La variable "lotes_transformados" almacena los lotes generados durante la iteración.

Iteramos sobre el dataloader mediante un bucle "for" para obtener los lotes de datos y etiquetas.
La variable "datos" corresponde al tensor que contiene las características del lote en esa iteración.
La variable "etiquetas" corresponde al tensor que contiene las etiquetas asociadas al lote en esa iteración.
    
Aplicamos el método ".append()" a la lista "lotes_transformados" para añadir cada lote obtenido del dataloader en cada iteración del bucle.
La tupla añadida contiene los tensores "datos" y "etiquetas" correspondientes al lote en cada iteración.

La variable "lotes_transformados" contiene todos los lotes obtenidos del dataloader durante la iteración.
El dataloader permite verificar que las transformaciones se aplican correctamente."""

from torch.utils.data import DataLoader

dataloader = DataLoader(dataset_transformado, batch_size=1, shuffle=False)

lotes_transformados = []

for datos, etiquetas in dataloader:
    lotes_transformados.append((datos, etiquetas))

lotes_transformados # Muestra todos los lotes transformados obtenidos del dataloader en forma de lista de tuplas (datos, etiquetas).
```

```python
# Bloque 8 — Estructura final de resultados
"""Este bloque organiza los elementos principales de las transformaciones básicas.
La variable "resultado" almacena en forma de diccionario los tensores transformados, transformaciones compuestas y lotes del dataloader.
La estructura permite visualizar el efecto de las transformaciones sobre los datos."""

resultado = {
    "tensor_original": tensor_original.tolist(),
    "tensor_escalado": tensor_escalado.tolist(),
    "tensor_sumado": tensor_sumado.tolist(),
    "tensor_normalizado": tensor_normalizado.tolist(),
    "tensor_convertido": tensor_convertido.tolist(),
    "tensor_compuesto": tensor_compuesto.tolist(),
    "lotes_transformados": [(d.tolist(), e.tolist()) for d, e in lotes_transformados]
}

resultado # Muestra la estructura final de resultados.
```
