```python
# Bloque 1 — Transformaciones básicas
"""Este bloque define transformaciones básicas que operan sobre tensores.
Este bloque importa "torch" para trabajar con tensores.

Definimos la función "def multiplicar(x)" para aplicar una transformación de multiplicación constante a los tensores.
El parámetro "x" representa el tensor que será transformado.
La función recibe un tensor y devuelve una versión multiplicada mediante la instrucción "return x * 2".

Definimos la función "def sumar(x)" para aplicar una transformación de suma constante a los tensores.
El parámetro "x" representa el tensor que será transformado.
La función recibe un tensor y devuelve una versión sumada mediante la instrucción "return x + 3".

La variable "tensor_original" contiene el tensor inicial antes de aplicar cualquier transformación.
Estas transformaciones sirven como componentes individuales para la composición."""

import torch

def multiplicar(x):
    return x * 2   

def sumar(x):
    return x + 3       

tensor_original = torch.tensor([1.0, 2.0, 3.0])

tensor_original # Muestra el tensor original.
```

```python
# Bloque 2 — Composición manual de transformaciones
"""Este bloque compone transformaciones aplicándolas secuencialmente.
La variable "transformaciones" contiene la lista de transformaciones que se aplicarán al tensor original.
Estas transformaciones son dos funciones definidas previamente: "multiplicar()" y "sumar()".

Definimos la función "def componer(x, transformaciones)" para aplicar una secuencia de transformaciones a un tensor.
El parámetro "x" representa el tensor que será transformado.
El parámetro "transformaciones" es una lista de funciones que representan las transformaciones que se aplicarán al tensor "x".

Dentro de la función, se itera sobre cada transformación de la lista "transformaciones" y, en cada iteración del bucle, "t" se aplica 
al tensor "x" mediante la expresión "x = t(x)".
Además, se retorna el valor final del tensor transformado con la expresión "return x".
La función recibe un tensor y una lista de transformaciones, y aplica cada transformación "t" al tensor "x" en el orden especificado.

La variable "tensor_compuesto" contiene el resultado de llamar a la función "componer()".
El argumento "tensor_original" representa el tensor que será transformado.
El argumento "transformaciones" es la lista de transformaciones que se aplicarán al tensor "tensor_original".  
La composición permite encadenar operaciones de preprocesamiento de forma controlada."""

transformaciones = [multiplicar, sumar]

def componer(x, transformaciones):
    for t in transformaciones:
        x = t(x)
    return x


tensor_compuesto = componer(tensor_original, transformaciones)

tensor_compuesto # Muestra el tensor después de aplicar las transformaciones compuestas.
```

```python
# Bloque 3 — Transformación encapsulada en clase
"""Este bloque define una transformación encapsulada en una clase.
Definimos una clase personalizada llamada "TransformacionPotencia" que aplica una potencia constante a cada tensor recibido.

Definimos la función "def __init__()", que inicializa la estructura interna de la clase.
El parámetro "self" de la función hace referencia a la instancia actual de la clase.
El parámetro "p" representa la potencia a la que se elevará cada tensor recibido.
Dentro de la función "__init__()" definimos la variable de instancia "self.p" y le asignamos el valor del parámetro "p".

Definimos la función "def __call__()", que permite que la instancia de la clase sea llamada como una función.
El parámetro "self" de la función hace referencia a la instancia actual de la clase.
El parámetro "x" representa el tensor al que se le aplicará la transformación.
Dentro de la función "__call__()" retornamos el resultado de elevar el tensor "x" a la potencia de la variable de instancia "self.p".

La variable "transformacion_potencia" es una instancia de la clase "TransformacionPotencia()" que contiene el valor de la potencia 
a la que se elevará cada tensor recibido; en este caso el valor de la potencia es 2.

La variable "tensor_potencia" contiene el resultado de llamar a la instancia de la clase "TransformacionPotencia()" aplicada 
al tensor original pasado como argumento.
Las transformaciones basadas en clases permiten parametrización y reutilización."""

class TransformacionPotencia:
    def __init__(self, p):
        self.p = p

    def __call__(self, x):
        return x ** self.p

transformacion_potencia = TransformacionPotencia(2)
tensor_potencia = transformacion_potencia(tensor_original)

tensor_potencia # Muestra el tensor elevado a la potencia definida.
```

```python
# Bloque 4 — Composición con transformaciones mixtas
"""Este bloque compone transformaciones definidas como funciones y clases.
La variable "transformaciones_mixtas" contiene la lista de transformaciones que se aplicarán al tensor original.
Estas transformaciones son dos funciones definidas previamente: "multiplicar()" y "sumar()", además de la instancia "transformacion_potencia".

La función "componer()" recibe un tensor y una lista de transformaciones, y aplica cada transformación "t" al tensor "x" en el orden especificado.
La composición admite cualquier objeto llamable que reciba y devuelva tensores.

La variable "tensor_mixto" contiene el resultado de llamar a la función "componer()".
El argumento "tensor_original" representa el tensor que será transformado.
El argumento "transformaciones_mixtas" es la lista de transformaciones que se aplicarán al tensor "tensor_original".  
El resultado muestra la flexibilidad del pipeline de transformaciones."""

transformaciones_mixtas = [
    multiplicar,
    transformacion_potencia,
    sumar
]

tensor_mixto = componer(tensor_original, transformaciones_mixtas)

tensor_mixto # Muestra el tensor resultante de aplicar las transformaciones mixtas.
```

```python
# Bloque 5 — Integración de composición en un dataset
"""Este bloque integra la composición de transformaciones dentro de un dataset personalizado.
Este bloque importa "Dataset" de "torch.utils.data" para crear un dataset personalizado.

Definimos una clase personalizada llamada "DatasetTransformado()" que representa un dataset con transformaciones aplicadas a los datos.
La clase "DatasetTransformado()" hereda de "Dataset", que es la clase base utilizada para crear datasets en PyTorch.

Definimos la función "def __init__()", que inicializa la estructura interna del dataset.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
El parámetro "transformaciones" de la función representa la secuencia de transformaciones que se aplicará a los datos del dataset.

Dentro de la función "__init__()" definimos dos tensores que representan los datos y las etiquetas del dataset.
En ambos casos asignamos los tensores a las variables de instancia "self.datos" y "self.etiquetas", respectivamente.
También asignamos el argumento "transformaciones" a la variable de instancia "self.transformaciones" para que pueda ser 
utilizado posteriormente al acceder a los elementos del dataset.

Definimos la función "def __len__()", que devuelve la cantidad de elementos del dataset.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
Utilizamos "return len(self.datos)" para devolver la longitud de los datos del dataset.

Definimos la función "def __getitem__()", que permite acceder a un elemento específico del dataset mediante un índice.
El parámetro "self" de la función hace referencia a la instancia actual del dataset.
El parámetro "idx" representa el índice del elemento que se desea obtener.

Además, asignamos a "x" el valor del dato correspondiente al índice "idx" con la expresión "x = self.datos[idx]" antes de aplicar las transformaciones.
De esta forma, se asegura que las transformaciones se apliquen en el orden correcto a cada dato del dataset.

Dentro de la función, se itera sobre la variable "self.transformaciones" y, en cada iteración del
bucle, "t" se aplica al tensor "x" mediante la expresión "x = t(x)".
Además, se retornan los datos transformados y la etiqueta correspondiente al índice proporcionado con la expresión "return x, self.etiquetas[idx]" 
respectivamente.

La variable "dataset" es una instancia de la clase "DatasetTransformado()" que recibe una lista de dos transformaciones y las aplica a los 
datos cada vez que se accede a un elemento del dataset.
Las transformaciones son las funciones "multiplicar()" y "sumar()", que se aplican a cada tensor en el orden especificado.
El dataset transformado permite modificar automáticamente cada muestra al acceder a ella con el propio índice usando la expresión "dataset[idx]"."""

from torch.utils.data import Dataset

class DatasetTransformado(Dataset):
    def __init__(self, transformaciones):
        self.datos = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
        self.etiquetas = torch.tensor([0, 1])
        self.transformaciones = transformaciones

    def __len__(self):
        return len(self.datos)

    def __getitem__(self, idx):
        x = self.datos[idx]
        for t in self.transformaciones:
            x = t(x)
        return x, self.etiquetas[idx]

dataset = DatasetTransformado([multiplicar, sumar])

dataset[0] # Aplica las transformaciones "multiplicar()" y "sumar()" al primer elemento del dataset y devuelve el resultado junto con su etiqueta.
```

```python
# Bloque 6 — Dataloader con transformaciones compuestas
"""Este bloque crea un dataloader para el dataset transformado.
Este bloque importa "DataLoader" de "torch.utils.data" para crear un dataloader a partir del dataset.

Aplicamos la clase "DataLoader()" a la variable "dataloader" para crear un dataloader a partir del dataset.
El argumento "dataset" corresponde a la instancia del dataset transformado que se desea cargar en el dataloader.
El parámetro "batch_size=" especifica el tamaño máximo de los lotes que se obtendrán del dataloader.
El parámetro "shuffle=" indica si debe mezclarse el orden de las muestras antes de crear los lotes.
La variable "dataloader" almacena un objeto iterable con datos transformados.

Creamos un contenedor de lista llamado "lotes_transformados" para almacenar los lotes obtenidos del dataloader.
La variable "lotes_transformados" almacena los lotes generados durante la iteración.

Iteramos sobre el dataloader mediante un bucle "for" para obtener los lotes de datos y etiquetas.
La variable "datos" corresponde al tensor que contiene las características del lote en esa iteración.
La variable "etiquetas" corresponde al tensor que contiene las etiquetas asociadas al lote en esa iteración.
    
Aplicamos el método ".append()" a la lista "lotes_transformados" para añadir cada lote obtenido del dataloader en cada iteración del bucle.
La tupla añadida contiene los tensores "datos" y "etiquetas" correspondientes al lote en cada iteración.

La variable "lotes_transformados" contiene todos los lotes obtenidos del dataloader durante la iteración.
La iteración permite verificar que las transformaciones compuestas se aplican correctamente."""

from torch.utils.data import DataLoader

dataloader = DataLoader(dataset, batch_size=1, shuffle=False)

lotes_transformados = []

for datos, etiquetas in dataloader:
    lotes_transformados.append((datos, etiquetas))


lotes_transformados # Muestra todos los lotes transformados obtenidos del dataloader en forma de lista de tuplas (datos, etiquetas).
```

```python
# Bloque 7 — Estructura final de resultados
"""Este bloque organiza los resultados principales de la composición de transformaciones.
La variable "resultado" almacena en forma de diccionario los tensores transformados y lotes del dataloader.
La estructura permite visualizar el efecto de las transformaciones encadenadas."""

resultado = {
    "tensor_original": tensor_original.tolist(),
    "tensor_compuesto": tensor_compuesto.tolist(),
    "tensor_potencia": tensor_potencia.tolist(),
    "tensor_mixto": tensor_mixto.tolist(),
    "lotes_transformados": [(d.tolist(), e.tolist()) for d, e in lotes_transformados]
}

resultado # Muestra la estructura final de resultados en forma de diccionario.
```
