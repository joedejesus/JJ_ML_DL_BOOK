<h3 style="margin:0; font-size:24px; line-height:1.3; font-weight:700; color:#ffffff;">Bloque 1. Preparación del dataset visual</h3>

```python
# 1.1 — Importación de librerías
"""Este bloque importa las librerías necesarias para visión artificial.
- torch: operaciones tensoriales.
- torchvision: datasets y transformaciones.
- transforms: preprocesamiento de imágenes.
- DataLoader y Subset: carga por lotes y selección de ejemplos.
- Path: busca la carpeta del proyecto desde el directorio de trabajo actual y sus antecesores."""

from hashlib import sha256
from pathlib import Path
from PIL import Image

PROJECT_DIR = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "Proyecto_de_deep_learning.ipynb").exists()
    and (path / "data" / "train").exists()
    and (path / "data" / "test").exists()
)
DATA_DIR = PROJECT_DIR / "data"

import torch
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms

torch.manual_seed(42)
```

```python
# 1.2 — Transformaciones de imágenes
"""Este bloque define transformaciones para las imágenes.
- Resize: ajusta tamaño.
- ToTensor: convierte a tensores y escala los píxeles al intervalo [0, 1].
- Normalize: transforma los valores de los tres canales al intervalo [-1, 1]."""

transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
])
```

```python
# 1.3 — Carga del dataset
"""Este bloque carga las imágenes desde carpetas locales.
- datasets.ImageFolder: estructura de carpetas por clase.
- DataLoader: crea lotes de datos.
- batch_size=32: tamaño máximo de cada lote; el conjunto de entrenamiento se mezcla.
- Se excluyen del entrenamiento las imágenes cuyos píxeles RGB coinciden con alguna imagen de prueba.
- Las carpetas originales no se modifican."""

train_data = datasets.ImageFolder(str(PROJECT_DIR / "data" / "train"), transform=transform)
test_data = datasets.ImageFolder(str(PROJECT_DIR / "data" / "test"), transform=transform)

if train_data.class_to_idx != test_data.class_to_idx:
    raise ValueError("Las clases de entrenamiento y prueba no coinciden.")

def image_signature(path):
    with Image.open(path) as image:
        rgb = image.convert("RGB")
        return rgb.size, sha256(rgb.tobytes()).digest()

test_signatures = {image_signature(path) for path, _ in test_data.samples}
train_indices = [i for i, (path, _) in enumerate(train_data.samples) if image_signature(path) not in test_signatures]
if {train_data.targets[i] for i in train_indices} != set(train_data.class_to_idx.values()):
    raise ValueError("La exclusión de duplicados dejó alguna clase sin ejemplos de entrenamiento.")
print(f"Entrenamiento: {len(train_indices)} imágenes; duplicados excluidos: {len(train_data) - len(train_indices)}")

train_loader = DataLoader(Subset(train_data, train_indices), batch_size=32, shuffle=True)
test_loader = DataLoader(test_data, batch_size=32, shuffle=False)
```

```python
# 2.1 — Definición de la arquitectura CNN
"""Este bloque define una red neuronal convolucional.
- nn.Conv2d: capas de convolución.
- nn.ReLU: función de activación.
- nn.MaxPool2d: reducción espacial.
- nn.Flatten: aplana tensores.
- nn.Linear: capas densas."""

import torch.nn as nn

class DefectCNN(nn.Module):
    def __init__(self):
        super(DefectCNN, self).__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.fc = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32 * 32 * 32, 64),
            nn.ReLU(),
            nn.Linear(64, 2)
        )

    def forward(self, x):
        x = self.conv(x)
        x = self.fc(x)
        return x

model = DefectCNN()

```

```python
# 3.1 — Configuración y ejecución del entrenamiento
"""Este bloque define la función de pérdida, el optimizador y ejecuta un entrenamiento corto.
- nn.CrossEntropyLoss: pérdida para clasificación.
- torch.optim.Adam: optimizador Adam.
- loss_history: registra la media de las pérdidas de los lotes por época, sin ponderarlas por su tamaño."""

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
loss_history = []

for epoch in range(5):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += float(loss.item())
    epoch_loss = running_loss / max(len(train_loader), 1)
    loss_history.append(epoch_loss)
    print(f"Epoch {epoch + 1}/5 - loss: {epoch_loss:.4f}")
```

```python
# 4.1 — Cálculo de accuracy
"""Este bloque evalúa el modelo en el conjunto de prueba, cuyas imágenes también aparecen en entrenamiento.
- model.eval(): activa modo evaluación.
- torch.no_grad(): desactiva gradientes.
- torch.argmax: predicción de clase.
- pd.crosstab: resume aciertos y errores por clase de forma legible."""

import pandas as pd

correct = 0
total = 0
true_labels = []
predicted_labels = []

model.eval()
with torch.no_grad():
    for images, labels in test_loader:
        outputs = model(images)
        predictions = torch.argmax(outputs, dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)
        true_labels.extend(labels.tolist())
        predicted_labels.extend(predictions.tolist())

accuracy = correct / max(total, 1)
label_lookup = dict(enumerate(train_data.classes))
confusion_matrix = pd.crosstab(
    pd.Series([label_lookup[label] for label in true_labels], name="real"),
    pd.Series([label_lookup[label] for label in predicted_labels], name="pred"),
    dropna=False,
).reindex(index=train_data.classes, columns=train_data.classes, fill_value=0)

print(f"Accuracy del modelo: {accuracy:.3f}")
confusion_matrix
```
