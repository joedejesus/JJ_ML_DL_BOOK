```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena columnas adecuadas para entrenar un modelo supervisado.
El dataset permite calcular la importancia de las características analizando su contribución en el modelo."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700],
    "compra": [0, 1, 1, 0, 1, 1, 0, 0]
})


df # Muestra el DataFrame completo.
```

```python
# Bloque 2 — Separación de características y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para entrenar el modelo y calcular la importancia de las características."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y la variable objetivo.
```

```python
# Bloque 3 — Entrenamiento del modelo
"""Este bloque entrena un modelo de clasificación.
Este bloque importa "DecisionTreeClassifier" de "sklearn.tree" para crear un modelo de clasificación basado en árboles.

Aplicamos la clase "DecisionTreeClassifier()" a la variable "modelo" para crear un modelo de clasificación basado en árboles.
El parámetro "random_state=" se utiliza para garantizar la reproducibilidad del modelo.

Aplicamos el método ".fit()" al modelo utilizando las características "X" y la variable objetivo "y".
La variable "modelo" almacena un clasificador basado en árboles ajustado a los datos de entrenamiento."""

from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(random_state=0)
modelo.fit(X, y)

modelo # Muestra el modelo entrenado.
```

```python
# Bloque 4 — Cálculo de la importancia de características
"""Este bloque calcula la importancia de las características del modelo.
Accedemos al atributo "feature_importances_" del modelo para obtener la importancia de cada característica.
La variable "importancias" almacena los valores de importancia asignados por el clasificador.
El atributo "feature_importances_" cuantifica la reducción normalizada de impureza atribuida a cada característica."""

importancias = modelo.feature_importances_

importancias # Muestra la importancia de cada característica.
```

```python
# Bloque 5 — Estructura de resultados
"""Este bloque organiza las importancias de características en una estructura.
La variable "resultado" almacena cada característica junto con su importancia correspondiente en forma de diccionario.
En cada caso utilizamos el operador de indexación "[]" para asociar cada importancia con su columna correspondiente.
La estructura permite interpretar el peso relativo de cada variable en el modelo."""

resultado = {
    "edad": importancias[0],
    "ingresos": importancias[1]      
}
  
resultado # Muestra la estructura con las importancias de cada característica.
```
