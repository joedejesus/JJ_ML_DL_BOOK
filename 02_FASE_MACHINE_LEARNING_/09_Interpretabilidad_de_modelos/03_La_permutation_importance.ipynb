```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena un dataset con columnas adecuadas para entrenar un modelo supervisado.
El dataset permite aplicar permutation importance evaluando el cambio de rendimiento al permutar características."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700],
    "compra": [0, 1, 1, 0, 1, 1, 0, 0]
})

df # Muestra el DataFrame completo.
```

```python
# Bloque 2 — Separación de características y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para entrenar el modelo y aplicar permutation importance."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y  # Muestra las características y la variable objetivo.
```

```python
# Bloque 3 — Entrenamiento del modelo
"""Este bloque entrena un modelo de clasificación.
Este bloque importa "DecisionTreeClassifier" de "sklearn.tree" para crear un clasificador basado en árboles.

Aplicamos la clase "DecisionTreeClassifier()" a la variable "modelo" para crear un clasificador basado en árboles.
El parámetro "random_state=" asegura la reproducibilidad del modelo.

Aplicamos el método ".fit()" al modelo utilizando las características "X" y la variable objetivo "y".
La variable "modelo" almacena un clasificador basado en árboles ajustado con los datos de entrenamiento."""

from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(random_state=0)
modelo.fit(X, y)

modelo # Muestra el clasificador basado en árboles ajustado con los datos de entrenamiento.
```

```python
# Bloque 4 — Aplicación de permutation importance
"""Este bloque aplica permutation importance sobre el modelo entrenado.
Este bloque importa "permutation_importance" de "sklearn.inspection" para evaluar la importancia de las 
características mediante permutaciones.

Aplicamos la función "permutation_importance()" al modelo, los datos de entrada "X" y las etiquetas "y", especificando 
los siguientes parámetros:
El parámetro "n_repeats=" especifica el número de permutaciones a realizar.
El parámetro "random_state=" asegura la reproducibilidad de los resultados.

La variable "resultado_perm" almacena la diferencia de rendimiento al permutar cada característica. En este caso se usan los datos de entrenamiento.
La aplicación de "permutation_importance" permite identificar qué características tienen un mayor impacto en 
el rendimiento del modelo."""

from sklearn.inspection import permutation_importance

resultado_perm = permutation_importance(modelo, X, y, n_repeats=10, random_state=0)

resultado_perm # Muestra los resultados de la importancia de las características mediante permutaciones.
```

```python
# Bloque 5 — Organización de resultados
"""Este bloque organiza los resultados de permutation importance en una estructura.
La variable "importancias" almacena la media de la diferencia de rendimiento por característica en forma de diccionario.
En cada caso accedemos al atributo "importances_mean" de "resultado_perm" y usamos "[]" para obtener la media 
de las diferencias de rendimiento correspondientes a cada característica.
La estructura permite interpretar la dependencia del modelo respecto a cada variable."""

importancias = {
    "edad": resultado_perm.importances_mean[0],
    "ingresos": resultado_perm.importances_mean[1]
}

importancias # Muestra las importancias de las características en forma de diccionario.
```

```python
# Bloque 6 — Estructura final
"""Este bloque construye la estructura final de salida del análisis.
La variable "resultado" almacena en un diccionario tanto las importancias medias como el detalle completo de las permutaciones.

La clave "importancias_medias" contiene la media de la diferencia de rendimiento asociada a cada característica.
La clave "importancias_completas" contiene todas las importancias calculadas en cada repetición de la permutación.

Se utiliza "resultado_perm.importances" y luego se aplica el método ".tolist()" para convertir el resultado en una 
lista de Python.
Esta estructura permite analizar tanto la importancia media de cada variable como la variabilidad de su impacto en el modelo."""

resultado = {
    "importancias_medias": importancias,
    "importancias_completas": resultado_perm.importances.tolist()
}

resultado  # Muestra la estructura final con el resumen y el detalle completo de las importancias.
```
