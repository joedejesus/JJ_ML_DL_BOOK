```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados en formato de DataFrame.
La variable "df" almacena columnas adecuadas para entrenar un modelo supervisado.
El dataset permite aplicar SHAP para obtener explicaciones locales del modelo, que pueden agregarse para un análisis global."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700],
    "compra": [0, 1, 1, 0, 1, 1, 0, 0]
})

df # Muestra el dataset completo.
```

```python
# Bloque 2 — Separación de características y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para entrenar el modelo y generar valores SHAP."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y  # Muestra las características y la variable objetivo.
```

```python
# Bloque 3 — Entrenamiento del modelo
"""Este bloque entrena un modelo de clasificación.
Este bloque importa "DecisionTreeClassifier" de "sklearn.tree" para crear un clasificador basado en árboles.

Aplicamos la clase "DecisionTreeClassifier()" a la variable "modelo" para crear un clasificador basado en árboles.
El parámetro "random_state=" se utiliza para garantizar la reproducibilidad del modelo.

Aplicamos el método ".fit()" al modelo utilizando las características "X" y la variable objetivo "y".
La variable "modelo" almacena un clasificador basado en árboles ajustado con los datos de entrenamiento."""

from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(random_state=0)
modelo.fit(X, y)

modelo # Muestra el modelo entrenado.
```

```python
# Bloque 4 — Creación del explicador SHAP
"""Este bloque crea un explicador SHAP basado en el modelo entrenado.
Este bloque importa "shap" para crear un explicador SHAP.
Aplicamos la clase "shap.TreeExplainer()" a la variable "explicador" con el modelo entrenado como argumento para crear un explicador SHAP.
La variable "explicador" almacena un objeto que calcula valores SHAP mediante TreeExplainer.
El explicador permite obtener explicaciones locales que pueden agregarse para analizar el comportamiento global del modelo."""

import shap

explicador = shap.TreeExplainer(modelo)

explicador # Muestra el explicador SHAP creado.
```

```python
# Bloque 5 — Cálculo de valores SHAP
"""Este bloque calcula los valores SHAP para todas las instancias del dataset.
Aplicamos el método ".shap_values()" al explicador con el dataset "X" como argumento para calcular los valores SHAP de todas las instancias.
La variable "valores_shap" almacena los valores que cuantifican la contribución de cada característica.
Los valores SHAP permiten interpretar cómo cada variable modifica la predicción respecto a una referencia."""

valores_shap = explicador.shap_values(X)

valores_shap # Muestra los valores SHAP calculados.
```

```python
# Bloque 6 — Estructura de resultados
"""Este bloque organiza los valores SHAP en una estructura.
La variable "resultado" almacena los valores SHAP y las características originales en forma de diccionario.

Para la primera clave "valores_shap" se convierten los valores SHAP a una lista con el método ".tolist()".
Esta conversión requiere un array; en versiones de SHAP que devuelven una lista para varias salidas, el método falla.
Para la segunda clave "caracteristicas" se almacenan las características originales "X" en forma de lista de 
diccionarios con el método ".to_dict()" aplicado a la variable "X".
El parámetro "orient='records'" indica que cada fila del DataFrame original se convierte en un diccionario independiente dentro de la lista.
La estructura permite analizar explicaciones locales y agregarlas para un análisis global del modelo."""

resultado = {
    "valores_shap": valores_shap.tolist(),
    "caracteristicas": X.to_dict(orient="records")
}

resultado # Muestra la estructura organizada con los valores SHAP y las características originales.
```
