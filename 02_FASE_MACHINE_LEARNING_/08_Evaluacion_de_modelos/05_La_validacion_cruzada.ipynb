```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena un dataset con las columnas adecuadas para entrenar un modelo supervisado.
El dataset permite aplicar validación cruzada evaluando el rendimiento en múltiples particiones."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700],
    "compra": [0, 1, 1, 0, 1, 1, 0, 0]
})

df # Muestra el dataset completo.
```

```python
# Bloque 2 — Separación de características y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para entrenar y evaluar el modelo mediante validación cruzada."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y la variable objetivo.
```

```python
# Bloque 3 — Definición del modelo
"""Este bloque define un modelo de clasificación.
Este bloque importa "DecisionTreeClassifier" de "sklearn.tree" para crear un clasificador basado en árboles.

Aplicamos la clase "DecisionTreeClassifier()" a la variable "modelo" para crear un modelo de clasificación basado en árboles.
El parámetro "random_state=" se utiliza para garantizar la reproducibilidad del modelo.
La variable "modelo" almacena un clasificador basado en árboles.
El modelo se utilizará en cada partición de la validación cruzada."""   

from sklearn.tree import DecisionTreeClassifier   

modelo = DecisionTreeClassifier(random_state=0)      
modelo # Muestra la representación del modelo de clasificación basado en árboles.
```

```python
# Bloque 4 — Aplicación de la validación cruzada
"""Este bloque aplica validación cruzada sobre el dataset completo.
Este bloque importa "cross_val_score" de "sklearn.model_selection" para realizar validación cruzada del modelo.

Aplicamos la función "cross_val_score()" al modelo, los datos de entrada "X" y las etiquetas "y", especificando 
el número de particiones con el parámetro "cv=3".
El dataset tiene ocho muestras, cuatro por clase, y se divide en tres particiones estratificadas.

La variable "cv_scores" almacena la exactitud obtenida en cada partición.
La validación cruzada permite evaluar la estabilidad del modelo al entrenarlo y evaluarlo en diferentes particiones 
del conjunto de datos."""

from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(modelo, X, y, cv=3)

cv_scores # Muestra las puntuaciones obtenidas en cada partición de la validación cruzada.
```

```python
# Bloque 5 — Cálculo de estadísticas de la validación cruzada
"""Este bloque calcula estadísticas descriptivas de los resultados de la validación cruzada.
La variable "media" almacena el promedio de las puntuaciones mediante el método ".mean()" aplicado a "cv_scores".
La variable "desviacion" almacena la variabilidad entre particiones mediante el método ".std()" aplicado a "cv_scores".
La evaluación permite interpretar estabilidad y rendimiento del modelo."""

media = cv_scores.mean()  
desviacion = cv_scores.std()  

media, desviacion  # Muestra el promedio y la desviación estándar de las puntuaciones de la validación cruzada.
```

```python
# Bloque 6 — Estructura de resultados
"""Este bloque organiza los resultados de la validación cruzada en una estructura.
La variable "resultado" almacena las puntuaciones individuales, la media y la desviación estándar en forma de diccionario.
Para el caso de valores múltiples utilizamos el método ".tolist()" para convertir las puntuaciones en una lista.
La estructura permite interpretar la capacidad de generalización del modelo."""

resultado = {
    "scores": cv_scores.tolist(),
    "media": media,
    "desviacion": desviacion
}

resultado  # Muestra la estructura con los resultados de la validación cruzada.
```
