```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena un dataset con columnas adecuadas para entrenar un modelo supervisado.
El dataset sirve como base para aplicar métricas de evaluación comparando predicciones con valores reales."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700],
    "compra": [0, 1, 1, 0, 1, 1, 0, 0]
})

df # Muestra el dataset creado para verificar su contenido.
```

```python
# Bloque 2 — Separación de características y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para entrenar y evaluar el modelo."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y la variable objetivo separadas.
```

```python
# Bloque 3 — División en entrenamiento y test
"""Este bloque divide los datos en entrenamiento y test.
Este bloque importa "train_test_split" de "sklearn.model_selection" para dividir los datos en conjuntos de entrenamiento y prueba.

La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los argumentos "X" e "y" representan las características y las etiquetas de los datos, respectivamente.
Los parámetros "test_size=" y "random_state=" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.

Las variables "X_train" y "y_train" contienen los datos de entrenamiento y se usan para ajustar el modelo.
Las variables "X_test" y "y_test" contienen los datos de prueba y se usan para evaluar su rendimiento.
La división permite medir la capacidad de generalización del modelo."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

X_train, X_test, y_train, y_test # Muestra la división de los datos en entrenamiento y prueba.
```

```python
# Bloque 4 — Entrenamiento del modelo
"""Este bloque entrena un modelo de clasificación.
Este bloque importa "DecisionTreeClassifier" de "sklearn.tree" para crear un clasificador basado en árboles.

Aplicamos la clase "DecisionTreeClassifier()" a la variable "modelo" para crear un clasificador basado en árboles.
El parámetro "random_state=" controla la aleatoriedad del clasificador.

Aplicamos el método ".fit()" al modelo utilizando los datos de entrenamiento "X_train" y "y_train" para ajustar el modelo.
La variable "modelo" almacena un clasificador basado en árboles ajustado a los datos de entrenamiento."""

from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(random_state=0)
modelo.fit(X_train, y_train)

modelo # Muestra el clasificador basado en árboles ajustado a los datos de entrenamiento.
```

```python
# Bloque 5 — Generación de predicciones
"""Este bloque genera predicciones sobre el conjunto de test.
Aplicamos el método ".predict()" al modelo utilizando los datos de test "X_test" para generar las predicciones.
La variable "pred" almacena las etiquetas predichas por el modelo.
La predicción permite comparar resultados estimados con valores reales."""

pred = modelo.predict(X_test)

pred # Muestra las etiquetas predichas por el modelo.
```

```python
# Bloque 6 — Cálculo de métricas básicas
"""Este bloque calcula métricas de evaluación del modelo.
Este bloque importa las funciones "accuracy_score", "precision_score", "recall_score" y "f1_score" de "sklearn.metrics" 
para calcular métricas de evaluación del modelo.
La variable "accuracy" almacena la proporción de aciertos.
La variable "precision" almacena la precisión del modelo en la clase positiva.
La variable "recall" almacena la capacidad del modelo para detectar la clase positiva.
La variable "f1" almacena la media armónica entre precisión y recall.
En todos los casos se comparan las predicciones del modelo "pred" con las etiquetas reales del conjunto de test "y_test".
El cálculo de estas métricas permite evaluar el rendimiento del modelo en términos de exactitud, precisión, 
exhaustividad y medida F1."""

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, pred)
precision = precision_score(y_test, pred)
recall = recall_score(y_test, pred)
f1 = f1_score(y_test, pred)

accuracy, precision, recall, f1 # Muestra las métricas de evaluación del modelo.
```

```python
# Bloque 7 — Validación cruzada
"""Este bloque aplica validación cruzada para evaluar la estabilidad del modelo.
Este bloque importa "cross_val_score" de "sklearn.model_selection" para realizar validación cruzada del modelo.
Aplicamos la función "cross_val_score()" al modelo, los datos de entrada "X" y las etiquetas "y", especificando 
el número de particiones con el parámetro "cv=4".
El dataset tiene ocho muestras, cuatro por clase, y se divide en cuatro particiones estratificadas.
La variable "cv_scores" almacena la exactitud obtenida en cada partición.
La validación cruzada permite evaluar la estabilidad del modelo al entrenarlo y evaluarlo en diferentes particiones 
del conjunto de datos completo, incluidos los datos de prueba de la división anterior."""

from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(modelo, X, y, cv=4)

cv_scores # Muestra las puntuaciones obtenidas en cada partición.   
```

```python
# Bloque 8 — Estructura de resultados
"""Este bloque organiza los resultados de evaluación en una estructura.
Utilizamos un diccionario para organizar las métricas y resultados de validación cruzada.

Cada clave del diccionario representa una métrica o resultado específico del modelo y su valor corresponde 
al resultado obtenido para esa métrica o evaluación específica.
En el caso de la validación cruzada utilizamos el método ".tolist()" para convertir los resultados en una lista.

La variable "evaluacion" almacena métricas básicas y resultados de validación cruzada en forma de diccionario.
La estructura permite interpretar el rendimiento y la estabilidad del modelo."""

evaluacion = {
    "accuracy": accuracy,
    "precision": precision,
    "recall": recall,
    "f1_score": f1,
    "validacion_cruzada": cv_scores.tolist()
}

evaluacion # Muestra la estructura de evaluación completa.
```
