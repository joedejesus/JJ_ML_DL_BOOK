```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena un dataset con columnas adecuadas para entrenar un modelo supervisado.
El dataset permite calcular la curva ROC y el AUC comparando probabilidades predichas con valores reales."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700],
    "compra": [0, 1, 1, 0, 1, 1, 0, 0]
})

df # Muestra el dataset creado para verificar su contenido.
```

```python
# Bloque 2 — Separación de características y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para entrenar y evaluar el modelo."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y la variable objetivo separadas.
```

```python
# Bloque 3 — División en entrenamiento y test
"""Este bloque divide los datos en entrenamiento y test.
Este bloque importa "train_test_split" de "sklearn.model_selection" para dividir los datos en conjuntos de entrenamiento y prueba.

La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los argumentos "X" e "y" representan las características y las etiquetas de los datos, respectivamente.
Los parámetros "test_size=" y "random_state=" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.

Las variables "X_train" y "y_train" contienen los datos de entrenamiento y se usan para ajustar el modelo.
Las variables "X_test" y "y_test" contienen los datos de prueba y se usan para evaluar su rendimiento.
La división permite calcular la curva ROC y el AUC sobre datos no vistos."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

X_train, X_test, y_train, y_test # Muestra los conjuntos de entrenamiento y prueba para verificar la división.
```

```python
# Bloque 4 — Entrenamiento del modelo
"""Este bloque entrena un modelo de clasificación.
Este bloque importa "DecisionTreeClassifier" de "sklearn.tree" para crear un clasificador basado en árboles.

Aplicamos la clase "DecisionTreeClassifier()" a la variable "modelo" para crear un clasificador basado en árboles.
El parámetro "random_state=" asegura la reproducibilidad del modelo al fijar la semilla de aleatoriedad.

Aplicamos el método ".fit()" al modelo utilizando los datos de entrenamiento "X_train" y "y_train".
La variable "modelo" almacena un clasificador basado en árboles ajustado a los datos de entrenamiento."""        

from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(random_state=0)
modelo.fit(X_train, y_train)

modelo # Muestra el modelo entrenado con los datos de entrenamiento.
```

```python
# Bloque 5 — Generación de probabilidades
"""Este bloque genera probabilidades de pertenecer a la clase positiva.
Aplicamos el método ".predict_proba()" al modelo utilizando los datos de prueba "X_test".
Utilizamos el índice [:, 1] para seleccionar las probabilidades de la clase positiva correspondiente a la segunda 
columna del array devuelto por "predict_proba()".
La variable "prob" almacena las probabilidades predichas por el modelo.
La generación permite obtener las probabilidades necesarias para calcular la curva ROC y el AUC."""

prob = modelo.predict_proba(X_test)[:, 1]

prob # Muestra las probabilidades predichas por el modelo.
```

```python
# Bloque 6 — Cálculo de la curva ROC
"""Este bloque calcula los valores necesarios para la curva ROC.
Este bloque importa "roc_curve" de "sklearn.metrics" para calcular la curva ROC.

Aplicamos la función "roc_curve()" a los valores reales "y_test" y las probabilidades predichas "prob"
para generar los valores de la curva ROC.

La variable "fpr" almacena la tasa de falsos positivos.
La variable "tpr" almacena la tasa de verdaderos positivos.
La variable "thresholds" almacena los umbrales utilizados.
El cálculo de la curva ROC permite evaluar el rendimiento del modelo en términos de su capacidad para distinguir entre 
clases positivas y negativas."""

from sklearn.metrics import roc_curve

fpr, tpr, thresholds = roc_curve(y_test, prob)

fpr, tpr, thresholds # Muestra los valores de la curva ROC.
```

```python
# Bloque 7 — Cálculo del AUC
"""Este bloque calcula el área bajo la curva ROC.
Este bloque importa "roc_auc_score" de "sklearn.metrics" para calcular el AUC.

Aplicamos la función "roc_auc_score()" a los valores reales "y_test" y las probabilidades predichas "prob" para calcular el AUC.
La variable "auc" almacena el valor numérico del área bajo la curva ROC.
El cálculo del AUC permite evaluar el rendimiento del modelo en términos de su capacidad para distinguir entre 
clases positivas y negativas."""

from sklearn.metrics import roc_auc_score

auc = roc_auc_score(y_test, prob)

auc # Muestra el valor del AUC.
```

```python
# Bloque 8 — Estructura de resultados
"""Este bloque organiza los valores de la curva ROC y el AUC en una estructura.
La variable "resultado" almacena fpr, tpr, thresholds y el valor del AUC en forma de diccionario.
Para los valores múltiples usamos el método ".tolist()" para convertir los arrays de NumPy en listas de Python.
La estructura permite interpretar la capacidad discriminativa del modelo."""

resultado = {
    "fpr": fpr.tolist(),
    "tpr": tpr.tolist(),
    "thresholds": thresholds.tolist(),
    "auc": auc
}

resultado # Muestra la estructura que contiene los valores de la curva ROC y el AUC.
```
