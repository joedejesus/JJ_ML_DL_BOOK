```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena un dataset con columnas adecuadas para entrenar un modelo supervisado.
El dataset permite calcular precisión, recall y f1-score comparando predicciones con valores reales."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700],
    "compra": [0, 1, 1, 0, 1, 1, 0, 0]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Separación de características y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para entrenar y evaluar el modelo."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y la variable objetivo.
```

```python
# Bloque 3 — División en entrenamiento y test
"""Este bloque divide los datos en entrenamiento y test.
Este bloque importa "train_test_split" de "sklearn.model_selection" para dividir los datos en conjuntos de entrenamiento y prueba.

La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los argumentos "X" e "y" representan las características y las etiquetas de los datos, respectivamente.
Los parámetros "test_size=" y "random_state=" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.

Las variables "X_train" y "y_train" contienen los datos de entrenamiento y se usan para ajustar el modelo.
Las variables "X_test" y "y_test" contienen los datos de prueba y se usan para evaluar su rendimiento.
La división permite calcular métricas sobre datos no vistos."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

X_train, X_test, y_train, y_test # Muestra los conjuntos de entrenamiento y prueba.
```

```python
# Bloque 4 — Entrenamiento del modelo
"""Este bloque entrena un modelo de clasificación.
Este bloque importa "DecisionTreeClassifier" de "sklearn.tree" para crear un modelo de clasificación basado en árboles.

Aplicamos la clase "DecisionTreeClassifier()" a la variable "modelo" para crear un clasificador basado en árboles.
El parámetro "random_state=" asegura la reproducibilidad del modelo.

Aplicamos el método ".fit()" al modelo con los datos de entrenamiento "X_train" y "y_train".
La variable "modelo" almacena un clasificador basado en árboles ajustado con los datos de entrenamiento."""

from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(random_state=0)
modelo.fit(X_train, y_train)

modelo # Muestra el modelo ajustado con los datos de entrenamiento.
```

```python
# Bloque 5 — Generación de predicciones
"""Este bloque genera predicciones sobre el conjunto de test.
Aplicamos el método ".predict()" al modelo con los datos de test "X_test" para generar las predicciones.
La variable "pred" almacena las etiquetas predichas por el modelo.
Las predicciones permiten comparar resultados estimados con valores reales."""

pred = modelo.predict(X_test)

pred # Muestra las etiquetas predichas por el modelo.
```

```python
# Bloque 6 — Cálculo de precisión, recall y f1-score
"""Este bloque calcula precisión, recall y f1-score del modelo.
Este bloque importa las funciones "precision_score", "recall_score" y "f1_score" de "sklearn.metrics" para 
calcular las métricas de evaluación del modelo.
La variable "precision" almacena la proporción de predicciones positivas correctas.
La variable "recall" almacena la proporción de instancias positivas detectadas.
La variable "f1" almacena la media armónica entre precisión y recall.
En todos los casos estas métricas se calculan comparando las predicciones del modelo "pred" con los valores reales "y_test".
La medición de estas métricas permite evaluar el desempeño del modelo en términos de precisión, recall y f1-score."""

from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(y_test, pred)
recall = recall_score(y_test, pred)
f1 = f1_score(y_test, pred)

precision, recall, f1 # Muestra las métricas de evaluación del modelo.
```

```python
# Bloque 7 — Estructura de resultados
"""Este bloque organiza las métricas en una estructura.
La variable "resultado" almacena precisión, recall y f1-score en forma de diccionario.
La estructura permite interpretar el rendimiento del modelo mediante estas métricas."""

resultado = {
    "precision": precision,
    "recall": recall,
    "f1_score": f1
}

resultado # Muestra la estructura con las métricas de evaluación del modelo.
```
