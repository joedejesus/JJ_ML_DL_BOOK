```python
#  Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena un dataset con columnas adecuadas para entrenar un modelo supervisado.
El dataset permite aplicar GridSearchCV evaluando múltiples configuraciones de hiperparámetros."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700],
    "compra": [0, 1, 1, 0, 1, 1, 0, 0]
})

df  # Muestra el dataset completo.
```

```python
# Bloque 2 — Separación de características y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para entrenar y evaluar el modelo mediante GridSearchCV."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y  # Muestra las características y la variable objetivo.
```

```python
# Bloque 3 — Definición del modelo base
"""Este bloque define un modelo de clasificación.
Este bloque importa "DecisionTreeClassifier" desde "sklearn.tree" para crear un modelo de clasificación basado en árboles.

Aplicamos la clase "DecisionTreeClassifier()" a la variable "modelo" para crear un clasificador basado en árboles.
El parámetro "random_state=" asegura la reproducibilidad del modelo al fijar la semilla del generador de números aleatorios.
La variable "modelo" almacena un clasificador basado en árboles."""

from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(random_state=0)

modelo # Muestra la configuración actual del clasificador basado en árboles.
```

```python
# Bloque 4 — Definición del espacio de hiperparámetros
"""Este bloque define el espacio de hiperparámetros a explorar.
La variable "parametros" almacena combinaciones de profundidad máxima y criterio de división en forma de diccionario.
La clave "max_depth" corresponde a la profundidad máxima del árbol y su valor es una lista de posibles profundidades a evaluar. 
La clave "criterion" corresponde al criterio de división y su valor es una lista de criterios posibles.
La definición del espacio de hiperparámetros permite a GridSearchCV explorar diferentes configuraciones de manera sistemática."""

parametros = {
    "max_depth": [2, 3, 4],
    "criterion": ["gini", "entropy"]
}
     
parametros # Muestra el espacio de hiperparámetros definido.
```

```python
# Bloque 5 — Aplicación de GridSearchCV
"""Este bloque aplica GridSearchCV sobre el dataset completo.
Este bloque importa "GridSearchCV" de "sklearn.model_selection" para realizar la búsqueda de hiperparámetros con validación cruzada.

Aplicamos la clase "GridSearchCV()" a la variable "grid" para realizar la búsqueda de hiperparámetros con validación cruzada.
El primer argumento es el clasificador base que se desea optimizar, en este caso "modelo".
El segundo argumento es el espacio de hiperparámetros definido anteriormente, en este caso "parametros".
El parámetro "cv=" especifica el número de particiones para la validación cruzada, en este caso 3.

Aplicamos el método ".fit()" sobre la variable "grid" para ejecutar el entrenamiento y evaluación de cada combinación de hiperparámetros.
Los argumentos "X" e "y" corresponden a las características y etiquetas del dataset completo, respectivamente.

La variable "grid" almacena el objeto configurado con validación cruzada.
La aplicación de grid search permite encontrar la mejor combinación entre las evaluadas para el modelo."""

from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(modelo, parametros, cv=3)
grid.fit(X, y)

grid # Muestra la configuración del objeto GridSearchCV ajustado.
```

```python
# Bloque 6 — Obtención del mejor modelo
"""Este bloque obtiene el mejor modelo encontrado por GridSearchCV.
Accedemos al atributo "best_estimator_" de la variable "grid" para obtener el mejor modelo.
La variable "mejor_modelo" almacena el estimador con la mejor combinación de hiperparámetros.
El atributo "best_estimator_" permite acceder al modelo seleccionado y reajustado con el dataset completo."""

mejor_modelo = grid.best_estimator_

mejor_modelo # Muestra el mejor modelo encontrado por GridSearchCV.
```

```python
# Bloque 7 — Obtención de los mejores hiperparámetros
"""Este bloque obtiene los mejores hiperparámetros seleccionados por GridSearchCV.
Accedemos al atributo "best_params_" de la variable "grid" para obtener los mejores hiperparámetros.
La variable "mejores_params" almacena la mejor configuración de hiperparámetros evaluada.
El atributo "best_params_" permite interpretar qué valores obtuvieron la mejor puntuación media en validación cruzada."""

mejores_params = grid.best_params_

mejores_params # Muestra los mejores hiperparámetros encontrados por GridSearchCV.
```

```python
# Bloque 8 — Obtención del mejor score
"""Este bloque obtiene el mejor score obtenido durante la validación cruzada.
Accedemos al atributo "best_score_" de la variable "grid" para obtener la mejor puntuación media de validación cruzada.
La variable "mejor_score" almacena la exactitud media de la mejor configuración en las particiones de validación.
Esta puntuación se usa para seleccionar la configuración y no es una evaluación independiente del modelo."""

mejor_score = grid.best_score_

mejor_score # Muestra el mejor score obtenido durante la validación cruzada.
```

```python
# Bloque 9 — Estructura de resultados
"""Este bloque organiza los resultados de GridSearchCV en una estructura.
La variable "resultado" almacena el mejor modelo, los mejores hiperparámetros y el mejor score en forma de diccionario.
En cada caso, cada clave contiene el valor correspondiente al dato obtenido durante la optimización.
Para el caso de "mejor_modelo" utilizamos el constructor "str()" para convertir el objeto del modelo en una 
representación de cadena de texto.
La estructura permite interpretar el proceso de optimización realizado."""

resultado = {
    "mejor_modelo": str(mejor_modelo),
    "mejores_parametros": mejores_params,
    "mejor_score": mejor_score
}

resultado # Muestra la estructura con los resultados de la optimización.
```
