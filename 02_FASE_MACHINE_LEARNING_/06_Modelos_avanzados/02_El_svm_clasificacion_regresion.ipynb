```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y dos etiquetas: una binaria y una continua.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena columnas adecuadas para entrenar modelos SVM de clasificación y regresión.
El dataset sirve como base para aplicar funciones de decisión basadas en márgenes."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "compra": [0, 1, 1, 0, 1],
    "gasto_total": [700, 1200, 1600, 900, 1400]
})

df # Muestra el DataFrame creado.
```

```python
# Bloque 2 — Separación de features y variables objetivo
"""Este bloque separa las características y las variables objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y_clf" almacena la etiqueta binaria para clasificación.
La variable "y_reg" almacena el valor continuo para regresión.
La separación permite preparar los datos para ambos modelos SVM."""

X = df[["edad", "ingresos"]]
y_clf = df["compra"]
y_reg = df["gasto_total"]

X, y_clf, y_reg # Muestra las variables separadas.
```

```python
# Bloque 3 — División en entrenamiento y test
"""Este bloque divide los datos en entrenamiento y test.
Este bloque importa "train_test_split" de "sklearn.model_selection" para dividir los datos en conjuntos de entrenamiento y prueba.

La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba para cada caso.
Los argumentos "X" e "y_clf" o "y_reg" representan las características y las etiquetas de los datos, respectivamente.
Los parámetros "test_size=" y "random_state=" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.

Las variables "X_train" y "y_clf_train" contienen los datos de entrenamiento y se usan para ajustar el modelo de clasificación.
Las variables "X_test" y "y_clf_test" contienen los datos de prueba y se usan para evaluar su rendimiento en clasificación.

La variable "y_reg_train" contiene los valores objetivo de entrenamiento para el modelo de regresión.
La variable "y_reg_test" contiene los valores objetivo de prueba para evaluar su rendimiento en regresión.

En este caso, la segunda división usa las mismas características, el mismo tamaño de prueba y la misma semilla, 
por lo que se pueden reutilizar "X_train" y "X_test" con las variables objetivo de regresión.
La división evita evaluar los modelos con datos vistos previamente."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_clf_train, y_clf_test = train_test_split(X, y_clf, test_size=0.4, random_state=0)
_, _, y_reg_train, y_reg_test = train_test_split(X, y_reg, test_size=0.4, random_state=0)

X_train, X_test, y_clf_train, y_clf_test, y_reg_train, y_reg_test # Muestra las variables de entrenamiento y prueba.
```

```python
# Bloque 4 — Entrenamiento del SVM para clasificación
"""Este bloque entrena un modelo SVM para clasificación.
Este bloque importa "SVC" de "sklearn.svm" para crear un clasificador SVM.

Aplicamos la clase "SVC" a la variable "modelo_clf" para crear un clasificador SVM.
El parámetro "kernel=" define el tipo de función de núcleo utilizada para calcular similitudes.
El parámetro "C=" controla la penalización por infracciones del margen.
El parámetro "gamma=" define el alcance de influencia de un solo ejemplo de entrenamiento.

Aplicamos el método ".fit()" al modelo con los datos de entrenamiento "X_train" y "y_clf_train" para ajustar la frontera 
de decisión del clasificador SVM.
La variable "modelo_clf" almacena un clasificador SVM con kernel RBF ajustado a los datos de entrenamiento."""

from sklearn.svm import SVC

modelo_clf = SVC(kernel="rbf", C=1.0, gamma="scale")
modelo_clf.fit(X_train, y_clf_train)

modelo_clf # Muestra el clasificador SVM entrenado.
```

```python
# Bloque 5 — Evaluación del SVM de clasificación
"""Este bloque evalúa el modelo SVM de clasificación.
Este bloque importa "accuracy_score" de "sklearn.metrics" para evaluar la exactitud del modelo.

Aplicamos el método ".predict()" al modelo con los datos de test "X_test" para generar las predicciones del clasificador SVM.
La variable "pred_clf" almacena las predicciones generadas sobre el conjunto de test.

Aplicamos la función "accuracy_score" a las etiquetas verdaderas "y_clf_test" y a las predicciones "pred_clf" para calcular 
la exactitud del modelo.
La variable "precision" almacena la métrica de exactitud del modelo.
La evaluación permite medir el desempeño del clasificador basado en márgenes sobre dos ejemplos de prueba."""

from sklearn.metrics import accuracy_score

pred_clf = modelo_clf.predict(X_test)
precision = accuracy_score(y_clf_test, pred_clf)

precision # Muestra la exactitud del modelo SVM de clasificación.   
```

```python
# Bloque 6 — Entrenamiento del SVM para regresión
"""Este bloque entrena un modelo SVM para regresión.
Este bloque importa "SVR" de "sklearn.svm" para crear un regresor SVM.

Aplicamos la clase "SVR" a la variable "modelo_reg" para crear un regresor SVM.
El parámetro "kernel=" define el tipo de función de núcleo utilizada para calcular similitudes.
El parámetro "C=" controla la penalización por errores de aproximación fuera del margen de tolerancia.
El parámetro "gamma=" define el alcance de influencia de un solo ejemplo de entrenamiento.

Aplicamos el método ".fit()" al modelo con los datos de entrenamiento "X_train" y "y_reg_train" para ajustar el 
modelo a los datos.
La variable "modelo_reg" almacena un regresor SVR con kernel RBF ajustado a los datos de entrenamiento."""

from sklearn.svm import SVR

modelo_reg = SVR(kernel="rbf", C=1.0, gamma="scale")
modelo_reg.fit(X_train, y_reg_train)

modelo_reg # Muestra el regresor SVM entrenado.
```

```python
# Bloque 7 — Evaluación del SVM de regresión
"""Este bloque evalúa el modelo SVM de regresión.
Este bloque importa "mean_squared_error" de "sklearn.metrics" para calcular el error cuadrático medio del modelo.

Aplicamos el método ".predict()" al modelo con los datos de test "X_test" para generar las predicciones del modelo.
La variable "pred_reg" almacena las predicciones generadas sobre el conjunto de test.

Aplicamos la función "mean_squared_error" a las etiquetas verdaderas "y_reg_test" y a las predicciones "pred_reg" para 
calcular el error cuadrático medio.
La variable "error" almacena el error cuadrático medio del modelo.
La evaluación permite medir el error de la aproximación continua sobre dos ejemplos de prueba."""

from sklearn.metrics import mean_squared_error

pred_reg = modelo_reg.predict(X_test)
error = mean_squared_error(y_reg_test, pred_reg)

error # Muestra el error cuadrático medio del modelo SVM de regresión.  
```
