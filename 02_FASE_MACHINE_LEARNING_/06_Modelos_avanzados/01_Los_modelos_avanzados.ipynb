```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena un dataset con columnas adecuadas para entrenar un modelo avanzado basado en márgenes.
El dataset sirve como base para aplicar un clasificador con funciones de decisión no lineales."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "compra": [0, 1, 1, 0, 1]
})

df # Muestra el DataFrame completo.
```

```python
# Bloque 2 — Separación de features y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para el modelo avanzado."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y la variable objetivo separadas.
```

```python
# Bloque 3 — División en entrenamiento y test
"""Este bloque divide los datos en entrenamiento y test.
Este bloque importa "train_test_split" de "sklearn.model_selection" para dividir los datos en conjuntos de entrenamiento y prueba.

La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los argumentos "X" e "y" representan las características y las etiquetas de los datos, respectivamente.
Los parámetros "test_size=" y "random_state=" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.

Las variables "X_train" y "y_train" contienen los datos de entrenamiento y se usan para ajustar el modelo.
Las variables "X_test" y "y_test" contienen los datos de prueba y se usan para evaluar su rendimiento.
La división evita evaluar el modelo con datos vistos previamente."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=0)

X_train, X_test, y_train, y_test # Muestra los conjuntos de entrenamiento y prueba.
```

```python
# Bloque 4 — Entrenamiento del modelo avanzado (SVM con kernel RBF)
"""Este bloque entrena un modelo avanzado basado en márgenes.
Este bloque importa "SVC" de "sklearn.svm" para crear un clasificador SVM.

Aplicamos la clase "SVC" a la variable "modelo" para crear un clasificador SVM con kernel RBF.
El parámetro "kernel=" define el tipo de kernel a utilizar.
El parámetro "C=" controla la penalización por infracciones del margen.
El parámetro "gamma=" ajusta la influencia de cada punto de entrenamiento.

Aplicamos el método ".fit()" al modelo con los datos de entrenamiento "X_train" y "y_train" para ajustar la frontera 
de decisión del clasificador SVM con kernel RBF.
La variable "modelo" almacena un clasificador SVM con kernel RBF ajustado a los datos de entrenamiento."""

from sklearn.svm import SVC

modelo = SVC(kernel="rbf", C=1.0, gamma="scale")
modelo.fit(X_train, y_train)

modelo # Muestra el modelo entrenado.
```

```python
# Bloque 5 — Predicción y evaluación del modelo
"""Este bloque evalúa el modelo entrenado.
Este bloque importa "accuracy_score" de "sklearn.metrics" para evaluar la exactitud del modelo.

Aplicamos el método ".predict()" al modelo con los datos de prueba "X_test" para generar las predicciones.
La variable "pred" almacena las predicciones generadas sobre el conjunto de test.

Aplicamos la función "accuracy_score()" con los valores reales "y_test" y las predicciones "pred" para calcular la exactitud del modelo.
La variable "precision" almacena la métrica de exactitud del modelo.
La evaluación permite medir el desempeño del clasificador avanzado sobre dos ejemplos de prueba."""

from sklearn.metrics import accuracy_score

pred = modelo.predict(X_test)
precision = accuracy_score(y_test, pred)

precision # Muestra la exactitud del modelo.
```
