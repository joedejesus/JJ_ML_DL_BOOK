```python
# Bloque 1 — Creación del dataset para clasificación binaria
"""Este bloque crea un dataset con características numéricas y una etiqueta binaria, en este caso "compra".
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena columnas adecuadas para entrenar un modelo de regresión logística.
El dataset sirve como base para estimar probabilidades de pertenencia a una clase."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "compra": [0, 1, 1, 0, 1]
})

df # Muestra el dataset creado para verificar su contenido.
```

```python
# Bloque 2 — Separación de features y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la etiqueta binaria a predecir.
La separación permite preparar los datos para el entrenamiento del modelo."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las variables separadas para verificar su contenido.
```

```python
# Bloque 3 — División en entrenamiento y test
"""Este bloque divide los datos en entrenamiento y test.
Este bloque importa "train_test_split" de "sklearn.model_selection" para dividir los datos en conjuntos de entrenamiento y prueba.

La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los argumentos "X" e "y" representan las características y las etiquetas de los datos, respectivamente.
Los parámetros "test_size=" y "random_state=" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.

Las variables "X_train" y "y_train" contienen los datos de entrenamiento y se usan para ajustar el modelo.
Las variables "X_test" y "y_test" contienen los datos de prueba y se usan para evaluar su rendimiento.
La división evita evaluar el modelo con datos vistos durante el entrenamiento."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=0)

X_train, X_test, y_train, y_test # Muestra los conjuntos de entrenamiento y prueba para verificar su contenido.
```

```python
# Bloque 4 — Entrenamiento del modelo de regresión logística
"""Este bloque entrena un modelo de regresión logística.
Este bloque importa "LogisticRegression" de "sklearn.linear_model" para entrenar un modelo de regresión logística.
Aplicamos la clase "LogisticRegression()" a la variable "modelo" para crear un modelo de regresión logística.
Aplicamos el método ".fit()" al modelo con los datos de entrenamiento "X_train" y "y_train" como argumentos 
para ajustar los parámetros del clasificador a los datos de entrenamiento.
La variable "modelo" almacena el clasificador ajustado con los datos de entrenamiento."""

from sklearn.linear_model import LogisticRegression

modelo = LogisticRegression()
modelo.fit(X_train, y_train)

modelo # Muestra el modelo entrenado para verificar su contenido.
```

```python
# Bloque 5 — Predicción y evaluación del modelo
"""Este bloque evalúa el modelo entrenado.
Este bloque importa "accuracy_score" de "sklearn.metrics" para evaluar la exactitud del modelo.

Aplicamos el método ".predict()" al modelo con los datos de prueba "X_test" como argumento para generar las predicciones.
La variable "pred" almacena las predicciones generadas sobre el conjunto de test.

Aplicamos la función "accuracy_score" con las etiquetas verdaderas "y_test" y las predicciones "pred" como argumentos 
para calcular la exactitud del modelo.
La variable "precision" almacena la métrica de exactitud del modelo.
La evaluación permite medir el desempeño del algoritmo de clasificación binaria sobre dos ejemplos de prueba."""

from sklearn.metrics import accuracy_score

pred = modelo.predict(X_test)
precision = accuracy_score(y_test, pred)

precision # Muestra la exactitud del modelo sobre el conjunto de prueba.
```
