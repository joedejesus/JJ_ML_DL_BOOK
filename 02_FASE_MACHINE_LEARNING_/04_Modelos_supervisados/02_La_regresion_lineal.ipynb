```python
# Bloque 1 — Creación del dataset para regresión
"""Este bloque crea un dataset con una columna dependiente continua, en este caso "ingresos".
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena columnas numéricas adecuadas para regresión lineal.
El dataset sirve como base para ajustar un modelo que predice un valor continuo."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "gastos": [700, 1200, 1600, 900, 1400]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Separación de features y variable objetivo
"""Este bloque separa las características y la variable objetivo.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la columna a predecir.
La separación permite preparar los datos para el entrenamiento del modelo."""

X = df[["edad", "gastos"]]
y = df["ingresos"]

X, y # Muestra las características y la variable objetivo.
```

```python
# Bloque 3 — División en entrenamiento y test
"""Este bloque divide los datos en entrenamiento y test.
Este bloque importa "train_test_split" de "sklearn.model_selection" para dividir los datos en conjuntos de entrenamiento y prueba.

La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los argumentos "X" e "y" representan las características y las etiquetas de los datos, respectivamente.
Los parámetros "test_size=" y "random_state=" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.

Las variables "X_train" y "y_train" contienen los datos de entrenamiento y se usan para ajustar el modelo.
Las variables "X_test" y "y_test" contienen los datos de prueba y se usan para evaluar su rendimiento.

La división evita evaluar el modelo con datos vistos durante el entrenamiento."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=0)

X_train, X_test, y_train, y_test # Muestra los conjuntos de entrenamiento y prueba.
```

```python
# Bloque 4 — Entrenamiento del modelo de regresión lineal
"""Este bloque entrena un modelo de regresión lineal.
Este bloque importa "LinearRegression" de "sklearn.linear_model" para crear un modelo de regresión lineal.
Aplicamos la clase "LinearRegression()" a la variable "modelo" para crear un modelo de regresión lineal.

Aplicamos el método ".fit()" al modelo con los datos de entrenamiento "X_train" y "y_train" como argumentos para ajustarlo.
El método ".fit()" calcula los coeficientes que minimizan el error cuadrático.
La variable "modelo" almacena el regresor ajustado con los datos de entrenamiento."""

from sklearn.linear_model import LinearRegression

modelo = LinearRegression()
modelo.fit(X_train, y_train)

modelo # Muestra el modelo entrenado.
```

```python
# Bloque 5 — Predicción y evaluación del modelo
"""Este bloque evalúa el modelo entrenado.
Este bloque importa "mean_squared_error" de "sklearn.metrics" para evaluar el rendimiento del modelo.

Aplicamos el método ".predict()" al modelo con los datos de prueba "X_test" como argumento para generar predicciones.
La variable "pred" almacena las predicciones generadas sobre el conjunto de test.

Aplicamos la función "mean_squared_error()" con las etiquetas verdaderas "y_test" y las predicciones "pred" como 
argumentos para calcular el error cuadrático medio.
La variable "error" almacena el error cuadrático medio del modelo.
La evaluación permite medir el error de las predicciones continuas sobre dos ejemplos de prueba."""

from sklearn.metrics import mean_squared_error

pred = modelo.predict(X_test)
error = mean_squared_error(y_test, pred)

error # Muestra el error cuadrático medio del modelo.   
```
