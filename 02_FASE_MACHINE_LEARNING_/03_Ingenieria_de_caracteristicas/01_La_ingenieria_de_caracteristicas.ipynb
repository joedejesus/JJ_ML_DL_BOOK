```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con columnas numéricas y categóricas.
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena un dataset con datos que permiten generar nuevas características.
El dataset sirve como base para aplicar ingeniería de características."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "ciudad": ["A", "B", "A", "C", "B"]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Creación de una característica combinada
"""Este bloque crea una característica derivada.
La variable "df["ratio_ingresos_edad"]" almacena la relación entre las columnas "ingresos" y "edad".
La combinación permite capturar información que no está explícita en las columnas originales."""

df["ratio_ingresos_edad"] = df["ingresos"] / df["edad"]

df # Muestra el dataset con la nueva característica.
```

```python
# Bloque 3 — Discretización de una variable numérica
"""Este bloque discretiza la variable edad.
La variable df["edad_grupos"] almacena la discretización de la columna "edad".
La función "pd.cut()" se utiliza para dividir la variable en intervalos definidos.
El argumento df["edad"] especifica la columna que se va a discretizar.
Los parámetros "bins" y "labels" definen los intervalos y las etiquetas correspondientes, respectivamente.
La discretización permite representar valores continuos como categorías."""
     
df["edad_grupos"] = pd.cut(df["edad"], bins=[0, 30, 40, 50], labels=["joven", "adulto", "senior"])

df # Muestra el dataset con la nueva característica discretizada.
```

```python
# Bloque 4 — Codificación de la nueva variable categórica
"""Este bloque codifica la variable discretizada.
Este bloque importa "OrdinalEncoder" de "sklearn.preprocessing" para realizar la codificación ordinal.
Aplicamos la clase "OrdinalEncoder()" a la variable "ordinal" para crear un objeto codificador de columnas categóricas.
El parámetro "categories=[["joven", "adulto", "senior"]]" indica el orden de las categorías.

Aplicamos el método "fit_transform()" al codificador "ordinal" indicando la columna a codificar.
El argumento "df[["edad_grupos"]]" indica la columna a codificar.

La variable "cod_edad" almacena códigos ordinales para la columna "edad_grupos" en un array de valores flotantes.
La codificación permite usar la discretización en modelos numéricos."""

from sklearn.preprocessing import OrdinalEncoder

ordinal = OrdinalEncoder(categories=[["joven", "adulto", "senior"]])       
cod_edad = ordinal.fit_transform(df[["edad_grupos"]])

cod_edad # Muestra la codificación ordinal de la columna "edad_grupos".
```

```python
# Bloque 5 — Incorporación de la codificación al dataset
"""Este bloque incorpora la codificación ordinal al dataset.
El método "copy()" se utiliza para crear una copia del dataset original en la variable "df_final".
Asignamos la codificación "cod_edad" a la nueva columna "edad_grupos_ordinal" del dataset "df_final".
La incorporación permite disponer de un conjunto ampliado para el modelado."""

df_final = df.copy()
df_final["edad_grupos_ordinal"] = cod_edad

df_final # Muestra el dataset final con la nueva columna "edad_grupos_ordinal" incorporada.
```
