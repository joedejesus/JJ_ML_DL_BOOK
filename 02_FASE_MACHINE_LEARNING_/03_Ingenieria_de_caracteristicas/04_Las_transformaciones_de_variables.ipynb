```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con columnas numéricas.
Este bloque importa "pandas" para trabajar con estructuras de datos tabulares.
La variable "df" almacena datos que permiten aplicar transformaciones.
El dataset sirve como base para modificar la representación de las variables."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "gastos": [700, 1200, 1600, 900, 1400]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Transformación logarítmica
"""Este bloque aplica una transformación logarítmica.
Este bloque importa "numpy" para trabajar con datos numéricos.
La variable "df["log_ingresos"]" almacena el logaritmo de los ingresos.
Aplicamos la función "np.log()" a la columna "df["ingresos"]" para obtener el logaritmo de los ingresos.
La transformación reduce la influencia de valores grandes."""

import numpy as np

df["log_ingresos"] = np.log(df["ingresos"])

df # Muestra el dataset con la nueva columna transformada.
```

```python
# Bloque 3 — Transformación de raíz cuadrada
"""Este bloque aplica una transformación de raíz cuadrada.
La variable "df["sqrt_gastos"]" almacena la raíz cuadrada de los gastos.
Aplicamos la función "np.sqrt()" a la columna "df["gastos"]" para obtener la raíz cuadrada de los gastos.
La transformación suaviza variaciones en distribuciones sesgadas."""

df["sqrt_gastos"] = np.sqrt(df["gastos"])

df # Muestra el dataset con la nueva columna transformada.           
```

```python
# Bloque 4 — Transformación de estandarización
"""Este bloque aplica estandarización a la columna edad.
Este bloque importa "StandardScaler" de "sklearn.preprocessing" para realizar la estandarización.
Aplicamos la clase "StandardScaler()" a la variable "scaler" para crear un objeto estandarizador.

Aplicamos el método "fit_transform()" al objeto "scaler" para estandarizar la columna.
El argumento del método "fit_transform()" es la columna que queremos estandarizar, en este caso "df[["edad"]]".

La variable "df["edad_std"]" almacena valores centrados y escalados.
La transformación ajusta los valores de la columna "edad" a media 0 y desviación estándar poblacional 1."""

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df["edad_std"] = scaler.fit_transform(df[["edad"]])

df # Muestra el dataset con la nueva columna transformada.
```

```python
# Bloque 5 — Transformación min-max
"""Este bloque aplica normalización min-max a la columna ingresos.
Este bloque importa "MinMaxScaler" de "sklearn.preprocessing" para realizar la normalización min-max.
Aplicamos la clase "MinMaxScaler()" a la variable "mm" para crear un objeto normalizador.

Aplicamos el método "fit_transform()" al objeto "mm" para normalizar la columna.
El argumento del método "fit_transform()" es la columna que queremos normalizar, en este caso "df[["ingresos"]]".

La variable "df["ingresos_minmax"]" almacena valores reescalados entre 0 y 1.
La transformación ajusta la variable según sus límites mínimo y máximo."""

from sklearn.preprocessing import MinMaxScaler

mm = MinMaxScaler()
df["ingresos_minmax"] = mm.fit_transform(df[["ingresos"]])

df # Muestra el dataset con la nueva columna transformada.
```
