```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con variables numéricas.
Este bloque importa "pandas" para trabajar con datos tabulares.
La variable "df" almacena un dataset con columnas que permiten evaluar relevancia.
El dataset sirve como base para aplicar selección de características."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "gastos": [700, 1200, 1600, 900, 1400],
    "compra": [0, 1, 1, 0, 1]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Separación de features y label
"""Este bloque separa las características y la etiqueta.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la columna objetivo.
La separación permite definir el material para la selección."""

X = df[["edad", "ingresos", "gastos"]]
y = df["compra"]

X, y # Muestra las variables separadas de features y label.
```

```python
# Bloque 3 — Selección basada en importancia del modelo
"""Este bloque calcula la importancia de las características mediante un modelo.
Este bloque importa "RandomForestClassifier" de "sklearn.ensemble" para entrenar un modelo de clasificación.
Aplicamos la clase "RandomForestClassifier()" a la variable modelo para crear un modelo de clasificación.
El parámetro "random_state=0" fija la semilla aleatoria del modelo.

Aplicamos el método ".fit()" al modelo para entrenarlo.
Los argumentos "X" e "y" representan las características y la etiqueta, respectivamente.
La variable "modelo" almacena un clasificador entrenado.

Accedemos al atributo ".feature_importances_" del modelo para obtener la importancia de cada característica.
La variable "importancia" almacena la relevancia de cada característica.
La importancia permite identificar las variables más relevantes según el modelo entrenado."""

from sklearn.ensemble import RandomForestClassifier

modelo = RandomForestClassifier(random_state=0)
modelo.fit(X, y)
importancia = modelo.feature_importances_

importancia # Muestra la importancia de cada característica.
```

```python
# Bloque 4 — Conversión de importancias a un DataFrame
"""Este bloque convierte la importancia en un DataFrame.
Aplicamos la clase "pd.DataFrame()" a la variable "df_importancia" para convertir las importancias en un DataFrame.

El primer par "clave-valor" del DataFrame corresponde a la columna "caracteristica" y su valor asociado.
Aplicamos la propiedad ".columns" de "X" para obtener los nombres de las características de "X" como valor asociado.

El segundo par "clave-valor" del DataFrame corresponde a la columna "importancia" y su valor asociado es el 
array de numpy de importancias obtenido del modelo.

La variable "df_importancia" almacena cada característica con su relevancia.
La conversión permite inspeccionar y ordenar los resultados."""

df_importancia = pd.DataFrame({
    "caracteristica": X.columns,
    "importancia": importancia
})

df_importancia # Muestra el DataFrame con la importancia de cada característica.
```

```python
# Bloque 5 — Selección de características relevantes
"""Este bloque selecciona características según un umbral.
La variable "umbral" define el límite de importancia para seleccionar características.

Aplicamos el método ".tolist()" a la columna "caracteristica" del DataFrame para crear una lista con los 
valores de la columna que cumplen con el umbral.

El umbral se compone de un valor numérico que la importancia de una característica debe superar para que sea 
seleccionada.

La variable "seleccionadas" contendrá únicamente las características cuya importancia supere el umbral.
La selección permite reducir dimensionalidad según la importancia estimada por el modelo."""

umbral = 0.3
seleccionadas = df_importancia[df_importancia["importancia"] > umbral]["caracteristica"].tolist()

seleccionadas # Muestra la lista de características seleccionadas.
```
