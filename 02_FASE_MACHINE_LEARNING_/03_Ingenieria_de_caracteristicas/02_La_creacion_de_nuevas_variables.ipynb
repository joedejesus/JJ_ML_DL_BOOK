```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset numérico y categórico.
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena un dataset con columnas que permiten generar nuevas características.
El dataset sirve como base para aplicar transformaciones derivadas."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "ciudad": ["A", "B", "A", "C", "B"]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Creación de una variable ratio
"""Este bloque crea una variable derivada mediante una operación matemática.
La variable "df["ratio_ingresos_edad"]" almacena la relación entre ingresos y edad.
La creación permite capturar una interacción entre ambas columnas."""

df["ratio_ingresos_edad"] = df["ingresos"] / df["edad"]

df # Muestra el dataset con la nueva variable creada.
```

```python
# Bloque 3 — Creación de una variable binaria
"""Este bloque crea una variable binaria basada en una condición.
La variable "df["ingresos_altos"]" almacena un indicador según un umbral definido.
Aplicamos el método ".astype(int)" al resultado de la condición para convertir sus valores booleanos en enteros.
La condición se compone de la variable "df["ingresos"]" y el umbral definido "2500".
La creación permite representar una característica categórica derivada de valores numéricos."""

df["ingresos_altos"] = (df["ingresos"] > 2500).astype(int)


df # Muestra el dataset con la nueva variable binaria creada.
```

```python
# Bloque 4 — Creación de una variable combinada categórica
"""Este bloque crea una variable categórica combinada.
La variable "df["ciudad_ingresos"]" almacena la unión entre ciudad y el nivel de ingresos.
Aplicamos el método ".astype(str)" a la variable "df["ingresos_altos"]" para convertir los valores binarios 
de la columna en cadenas de texto (str) antes de la concatenación.
Se concatenan las variables "df["ciudad"]" y "df["ingresos_altos"]" con el separador "_" para formar la nueva variable categórica.
La combinación permite capturar relaciones entre columnas heterogéneas."""

df["ciudad_ingresos"] = df["ciudad"] + "_" + df["ingresos_altos"].astype(str)

df # Muestra el dataset con la nueva variable combinada creada.
```

```python
# Bloque 5 — Generación de una variable transformada
"""Este bloque genera una variable transformada mediante una función matemática.
Este bloque importa "numpy" para trabajar con datos numéricos.
La variable "df["log_ingresos"]" almacena el logaritmo de los ingresos.
Aplicamos la función "np.log()" a la columna "df["ingresos"]" para obtener el logaritmo de los ingresos.
La transformación permite reducir la influencia de valores grandes."""

import numpy as np

df["log_ingresos"] = np.log(df["ingresos"])

df # Muestra el dataset con la nueva variable transformada creada.
```
