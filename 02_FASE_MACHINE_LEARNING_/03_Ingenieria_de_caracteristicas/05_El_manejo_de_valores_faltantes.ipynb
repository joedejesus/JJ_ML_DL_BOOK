```python
# Bloque 1 — Creación del dataset con valores faltantes
"""Este bloque crea un dataset con valores faltantes.
Este bloque importa "pandas" para trabajar con datos tabulados y "numpy" para manejar datos numéricos.
La variable "df" almacena un dataset con columnas con datos incompletos.
El dataset sirve como base para aplicar técnicas de imputación."""

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "edad": [22, np.nan, 47, 29, np.nan],
    "ingresos": [1500, 2300, np.nan, 1800, 2900],
    "gastos": [700, np.nan, 1600, 900, 1400]
})

df # Muestra el dataset con valores faltantes.
```

```python
# Bloque 2 — Detección de valores faltantes
"""Este bloque detecta valores faltantes.
La variable "faltantes" almacena el conteo por columna.
Aplicamos los métodos ".isnull()" y ".sum()" al DataFrame "df" para contar los valores faltantes por columna.
La detección permite identificar qué columnas requieren imputación."""

faltantes = df.isnull().sum()

faltantes # Muestra el conteo de valores faltantes por columna.
```

```python
# Bloque 3 — Imputación con la media
"""Este bloque aplica imputación con la media.
La variable "df_media" almacena el dataset con valores numéricos completados.
Aplicamos el método ".fillna()" al DataFrame con la media de cada columna numérica como argumento.
Calculamos la media de cada columna numérica aplicando el método ".mean()" con el parámetro "numeric_only=True".
El parámetro "numeric_only=True" limita el cálculo a las columnas numéricas.
La imputación sustituye valores faltantes por el promedio de cada columna."""

df_media = df.fillna(df.mean(numeric_only=True))

df_media # Muestra el dataset con valores faltantes imputados con la media.
```

```python
# Bloque 4 — Imputación con SimpleImputer
"""Este bloque aplica imputación mediante SimpleImputer.
Este bloque importa "SimpleImputer" de "sklearn.impute" para realizar la imputación de valores faltantes.

La variable "imputer" almacena el objeto imputador configurado con estrategia de media.
Aplicamos la clase "SimpleImputer()" a la variable "imputer" para crear un objeto imputador de valores faltantes.
El parámetro "strategy=" define la estrategia de imputación, en este caso "mean" para usar la media de cada columna.

Aplicamos el método "fit_transform()" al objeto "imputer" para completar los valores faltantes en el dataset.
El argumento "df" pasado al método "fit_transform()" es el dataset con valores faltantes que se desea completar.
La variable "datos_imputados" almacena los valores completados automáticamente."""

from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="mean")
datos_imputados = imputer.fit_transform(df)

datos_imputados # Muestra el array con valores faltantes imputados con la media.  
```

```python
# Bloque 5 — Conversión del resultado imputado
"""Este bloque convierte los datos imputados en un DataFrame.
La variable "df_imputado" almacena el dataset sin valores faltantes.
Aplicamos la clase "pd.DataFrame()" a la variable "df_imputado" para convertirlos en un DataFrame.
El argumento "datos_imputados" contiene el array de datos que han sido imputados y ya no presentan valores faltantes.
El parámetro "columns=df.columns" asegura que las columnas del DataFrame imputado tengan los mismos nombres que las del DataFrame original.
La conversión permite inspeccionar el resultado final de la imputación."""

df_imputado = pd.DataFrame(datos_imputados, columns=df.columns)

df_imputado # Muestra el DataFrame imputado.
```
