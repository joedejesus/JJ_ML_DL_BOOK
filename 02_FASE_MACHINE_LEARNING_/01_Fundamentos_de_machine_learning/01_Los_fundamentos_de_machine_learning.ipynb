```python
# Bloque 1 — Separación de variables y etiquetas
"""Este bloque separa variables y etiquetas.
Este bloque importa "pandas" para manejar datos estructurados.
La variable "df" almacena un DataFrame con datos de ejemplo.
La variable "X" almacena características.
La variable "y" almacena las etiquetas objetivo.
La separación permite preparar los datos para el entrenamiento."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29],
    "ingresos": [1500, 2300, 3200, 1800],
    "compra": [0, 1, 1, 0]
})

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y las etiquetas objetivo para verificar la separación.
```

```python
# Bloque 2 — División en entrenamiento y prueba
"""Este bloque divide los datos en entrenamiento y prueba.
Este bloque importa la función "train_test_split()" de "sklearn.model_selection" para dividir los datos.
Las variables "X_train" y "X_test" almacenan las características de entrenamiento y prueba, respectivamente.
Las variables "y_train" y "y_test" almacenan las etiquetas de entrenamiento y prueba, respectivamente.
La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los parámetros "test_size" y "random_state" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.
La división permite evaluar la capacidad de generalización del modelo."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

X_train, X_test, y_train, y_test # Muestra los conjuntos de entrenamiento y prueba para verificar la división.
```

```python
# Bloque 3 — Entrenamiento de un modelo supervisado
"""Este bloque entrena un modelo supervisado.
Este bloque importa la clase "LogisticRegression" de "sklearn.linear_model" para crear un clasificador.
Aplicamos la clase "LogisticRegression()" a la variable "modelo" para crear un clasificador de regresión logística.
Aplicamos el método ".fit()" del clasificador para entrenar el modelo de regresión logística con los datos de entrenamiento.
Los argumentos "X_train" y "y_train" representan las características y etiquetas de entrenamiento, respectivamente.
El entrenamiento permite aprender patrones a partir de los datos."""

from sklearn.linear_model import LogisticRegression

modelo = LogisticRegression()
modelo.fit(X_train, y_train)

modelo # Muestra el modelo después del entrenamiento.
```

```python
# Bloque 4 — Predicción
"""Este bloque genera predicciones.
La variable "predicciones" almacena valores estimados por el modelo.
Aplicamos el método ".predict()" del modelo para generar predicciones sobre los datos de prueba.
El argumento "X_test" representa las características de prueba.
La predicción permite evaluar el comportamiento del modelo sobre datos nuevos."""

predicciones = modelo.predict(X_test)

predicciones # Muestra las predicciones para verificar los resultados.
```

```python
# Bloque 5 — Métrica de evaluación
"""Este bloque calcula una métrica de evaluación.
Este bloque importa la función "accuracy_score()" de "sklearn.metrics" para calcular la exactitud.
La variable "precision" almacena la proporción de aciertos.
La función "accuracy_score()" compara las etiquetas verdaderas con las predicciones del modelo.
Los argumentos "y_test" y "predicciones" representan las etiquetas verdaderas y las predicciones del modelo, respectivamente.
La métrica permite medir el rendimiento del modelo, aunque este conjunto de prueba contiene solo un ejemplo."""

from sklearn.metrics import accuracy_score

precision = accuracy_score(y_test, predicciones)

precision # Muestra la exactitud para evaluar el rendimiento del modelo.
```
