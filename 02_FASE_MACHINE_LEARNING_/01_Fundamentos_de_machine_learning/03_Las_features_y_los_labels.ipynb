```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características y una etiqueta.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena instancias con columnas de entrada y salida.
Las columnas edad e ingresos son features y compra es el label."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29],
    "ingresos": [1500, 2300, 3200, 1800],
    "compra": [0, 1, 1, 0]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Separación de features
"""Este bloque separa las features.
La variable "X" almacena las columnas de entrada.
La separación permite preparar los datos para el modelo."""

X = df[["edad", "ingresos"]]

X # Muestra las features separadas.
```

```python
# Bloque 3 — Separación del label
"""Este bloque separa el label.
La variable "y" almacena la columna objetivo.
La separación permite definir la salida que el modelo debe predecir."""

y = df["compra"]

y # Muestra el label separado.
```

```python
# Bloque 4 — Verificación de dimensiones
"""Este bloque verifica las dimensiones de features y label.
Las variables "dim_X" y "dim_y" almacenan las formas de cada estructura.
La propiedad "shape" devuelve una tupla con el número de filas y columnas de "X" y el número de elementos de "y".
La verificación permite comparar el número de filas, pero no garantiza la alineación de los índices."""

dim_X = X.shape
dim_y = y.shape

dim_X, dim_y # Muestra las dimensiones de features y label.
```

```python
# Bloque 5 — Visualización simple de features y label
"""Este bloque genera una visualización simple.
Este bloque importa "matplotlib" para crear gráficos.
La función "scatter" genera un diagrama de dispersión con una feature y el label.
La función "xlabel" define la etiqueta del eje x.
La función "ylabel" define la etiqueta del eje y.
La figura muestra la relación entre ingresos y el label compra.
La visualización permite observar cómo se distribuye la salida respecto a una feature."""

import matplotlib.pyplot as plt

plt.scatter(X["ingresos"], y)
plt.xlabel("Ingresos")
plt.ylabel("Compra")

plt.show() # Muestra la figura generada.
```
