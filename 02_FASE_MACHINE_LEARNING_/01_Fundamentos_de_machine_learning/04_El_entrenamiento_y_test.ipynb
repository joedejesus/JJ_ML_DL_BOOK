```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características y etiqueta.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena instancias con dos columnas de entradas y una columna de salida.
El dataset permite definir el material para entrenamiento y test."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "compra": [0, 1, 1, 0, 1]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Separación de features y label
"""Este bloque separa las características y la etiqueta.
La variable "X" almacena las columnas de entrada.
La variable "y" almacena la columna objetivo.
La separación permite preparar los datos para el modelo."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y la etiqueta separadas.
```

```python
# Bloque 3 — División en entrenamiento y test
"""Este bloque divide los datos en entrenamiento y prueba.
Este bloque importa la función "train_test_split()" de "sklearn.model_selection" para dividir los datos.
Las variables "X_train" y "X_test" almacenan las características de entrenamiento y prueba, respectivamente.
Las variables "y_train" y "y_test" almacenan las etiquetas de entrenamiento y prueba, respectivamente.
La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los parámetros "test_size" y "random_state" controlan el tamaño del conjunto de prueba y la aleatoriedad de la división.
La división evita que el modelo se evalúe con datos ya vistos."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=0)

X_train, X_test, y_train, y_test # Muestra los conjuntos de entrenamiento y prueba para verificar la división.
```

```python
# Bloque 4 — Entrenamiento del modelo
"""Este bloque entrena un modelo supervisado.
Este bloque importa la clase "LogisticRegression" de "sklearn.linear_model" para crear un clasificador.
Aplicamos la clase "LogisticRegression()" a la variable "modelo" para crear un clasificador de regresión logística.
Aplicamos el método ".fit()" del clasificador para entrenar el modelo de regresión logística con los datos de entrenamiento.
Los argumentos "X_train" y "y_train" representan las características y etiquetas de entrenamiento, respectivamente.
El entrenamiento permite aprender patrones entre features y label."""

from sklearn.linear_model import LogisticRegression

modelo = LogisticRegression()
modelo.fit(X_train, y_train)

modelo # Muestra el modelo después del entrenamiento.
```

```python
# Bloque 5 — Evaluación en test
"""Este bloque evalúa el modelo en el conjunto de test.
Este bloque importa la función "accuracy_score()" de "sklearn.metrics" para calcular la exactitud del modelo.
La variable "pred" almacena las predicciones del modelo para el conjunto de test.
La función "predict()" genera predicciones basadas en las características de test.
El argumento "X_test" representa las características de test.
La variable "precision" almacena la proporción de aciertos.
La función "accuracy_score()" compara las etiquetas verdaderas con las predicciones del modelo.
Los argumentos "y_test" y "pred" representan las etiquetas verdaderas y las predicciones del modelo, respectivamente.
La evaluación permite estimar la capacidad de generalización, aunque el conjunto de prueba contiene solo dos ejemplos."""
     
from sklearn.metrics import accuracy_score

pred = modelo.predict(X_test)
precision = accuracy_score(y_test, pred)

pred, precision # Muestra las predicciones y la exactitud del modelo para evaluar su desempeño en el conjunto de test.
```
