```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset tabular.
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena datos que se organizan en filas y columnas.
Cada columna representa una característica del conjunto de datos."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29],
    "ingresos": [1500, 2300, 3200, 1800],
    "ciudad": ["A", "B", "A", "C"]
})

df # Muestra el DataFrame para verificar su contenido.
```

```python
# Bloque 2 — Inspección de las primeras filas
"""Este bloque muestra las primeras filas del dataset.
La variable "vista" almacena una porción inicial del conjunto.
El método ".head()" permite obtener las primeras filas del DataFrame.
La inspección permite verificar la estructura general."""

vista = df.head()

vista # Muestra las primeras filas del DataFrame para inspección.
```

```python
# Bloque 3 — Tipos de datos
"""Este bloque obtiene los tipos de datos del dataset.
La variable "tipos" almacena el tipo de cada columna.
La propiedad ".dtypes" permite acceder a los tipos de datos de un DataFrame.
La inspección permite identificar variables numéricas y categóricas."""

tipos = df.dtypes

tipos # Muestra los tipos de datos de cada columna para inspección.
```

```python
# Bloque 4 — Conteo de valores únicos
"""Este bloque calcula valores únicos por columna.
La variable "unicos" almacena el número de valores distintos por columna.
El método ".nunique()" permite contar valores únicos en cada columna del DataFrame.
El conteo permite analizar la variabilidad de cada característica."""

unicos = df.nunique()

unicos # Muestra el número de valores únicos por columna para inspección.
```

```python
# Bloque 5 — Estadísticas descriptivas
"""Este bloque genera estadísticas descriptivas.
La variable "estadisticas" almacena medidas de resumen.
El método ".describe()" permite obtener estadísticas de las columnas numéricas del DataFrame.
Las estadísticas permiten evaluar la distribución de las variables numéricas."""

estadisticas = df.describe()

estadisticas # Muestra las estadísticas descriptivas para inspección.

```
