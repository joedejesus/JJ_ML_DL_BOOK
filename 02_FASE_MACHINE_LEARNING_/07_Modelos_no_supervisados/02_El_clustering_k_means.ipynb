```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas sin etiquetas.
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena un dataset con columnas adecuadas para aplicar el algoritmo k-means.
El dataset sirve como base para agrupar instancias según similitud."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700]
})

df # Muestra el dataset creado para verificar su contenido.
```

```python
# Bloque 2 — Selección de características
"""Este bloque selecciona las características del dataset.
La variable "X" almacena las columnas de entrada.
La selección permite preparar los datos para el algoritmo k-means. Los ingresos tienen una escala mayor y pueden dominar las distancias."""

X = df[["edad", "ingresos"]]

X # Muestra las características seleccionadas para el algoritmo k-means.
```

```python
# Bloque 3 — Aplicación de k-means
"""Este bloque aplica el algoritmo k-means.
Este bloque importa "KMeans" de "sklearn.cluster" para crear un modelo de agrupamiento k-means.

Aplicamos la clase "KMeans()" a la variable "modelo" para crear un modelo agrupador k-means.
El parámetro "n_clusters=" define el número de clusters que se desean formar.
El parámetro "random_state=" fija la semilla del generador de números aleatorios para reproducir la inicialización.
La variable "modelo" almacena el agrupador configurado con un número específico de clusters.

Aplicamos el método ".fit_predict()" al modelo con las características seleccionadas "X" para obtener las 
etiquetas de los clusters asignados a cada instancia.
La variable "clusters" almacena las etiquetas de los clusters asignados a cada instancia.
La aplicación permite asignar cada instancia a un cluster basado en la distancia al centroide correspondiente."""

from sklearn.cluster import KMeans

modelo = KMeans(n_clusters=2, random_state=0)
clusters = modelo.fit_predict(X)

clusters # Muestra las etiquetas de los clusters asignados a cada instancia.
```

```python
# Bloque 4 — Incorporación de los clusters al dataset
"""Este bloque añade los clusters generados al dataset original.
La variable "df["cluster"]" almacena una nueva columna con las etiquetas de los clusters asignadas a cada instancia.
La incorporación de los clusters al dataset permite analizar los grupos obtenidos por el algoritmo."""

df["cluster"] = clusters

df # Muestra el dataset con la nueva columna de clusters.
```

```python
# Bloque 5 — Obtención de los centroides
"""Este bloque obtiene los centroides del modelo k-means.
Accedemos al atributo "cluster_centers_" del modelo para obtener las coordenadas de los centroides.
La variable "centroides" almacena las coordenadas de cada centroide.
La extracción permite interpretar la posición representativa de cada grupo."""

centroides = modelo.cluster_centers_

centroides # Muestra las coordenadas de los centroides.
```
