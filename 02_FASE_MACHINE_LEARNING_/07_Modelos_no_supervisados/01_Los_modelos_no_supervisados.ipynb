```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas sin etiquetas.
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena un dataset con columnas adecuadas para aplicar algoritmos no supervisados.
El dataset sirve como base para identificar estructuras internas sin información de clases."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700]
})

df # Muestra el dataset creado para verificar su contenido.
```

```python
# Bloque 2 — Selección de características
"""Este bloque selecciona las características del dataset.
La variable "X" almacena las columnas de entrada.
La selección permite preparar los datos para algoritmos no supervisados. Los ingresos tienen una escala mayor y pueden dominar las distancias."""

X = df[["edad", "ingresos"]]

X # Muestra las características seleccionadas para verificar su contenido.
```

```python
# Bloque 3 — Aplicación de KMeans
"""Este bloque aplica un algoritmo de clustering KMeans.
Este bloque importa "KMeans" de "sklearn.cluster" para crear un modelo de clustering.

Aplicamos la clase "KMeans()" a la variable "modelo_kmeans" para crear un modelo de clustering.
El parámetro "n_clusters=" define el número de grupos que el modelo intentará identificar en los datos.
El parámetro "random_state=" fija la semilla para reproducir la inicialización de los centroides.
La variable "modelo_kmeans" almacena el agrupador configurado con un número específico de clusters.

Aplicamos el método ".fit_predict()" al modelo con las características seleccionadas "X" para asignar cada instancia a un 
grupo basado en similitud.
La variable "clusters" almacena las etiquetas de los grupos asignados a cada instancia.
La aplicación de "KMeans" permite identificar grupos de instancias similares dentro del dataset."""

from sklearn.cluster import KMeans

modelo_kmeans = KMeans(n_clusters=2, random_state=0)
clusters = modelo_kmeans.fit_predict(X)

clusters # Muestra las etiquetas de los grupos asignados a cada instancia.
```

```python
# Bloque 4 — Incorporación de los clusters al dataset
"""Este bloque añade los clusters generados al dataset original.
La variable "df["cluster"]" almacena una nueva columna con las etiquetas de los clusters asignadas a cada instancia.
La incorporación de los clusters al dataset permite analizar los grupos obtenidos por el algoritmo."""

df["cluster"] = clusters 

df # Muestra el dataset con la nueva columna de clusters.
```

```python
# Bloque 5 — Aplicación de PCA
"""Este bloque aplica PCA a las características del dataset.
Este bloque importa "PCA" de "sklearn.decomposition" para calcular componentes principales.

Aplicamos la clase "PCA()" a la variable "modelo_pca" para crear un transformador de componentes principales.
El parámetro "n_components=" define el número de componentes principales que se conservarán.
La variable "modelo_pca" almacena el transformador configurado para dos componentes principales.

Aplicamos el método ".fit_transform()" al modelo con las características seleccionadas "X" para obtener la representación 
transformada del dataset.
La variable "pca_resultado" almacena la representación del dataset en componentes principales.
En este caso se conservan dos componentes para dos características, por lo que no se reduce la dimensionalidad."""

from sklearn.decomposition import PCA

modelo_pca = PCA(n_components=2)
pca_resultado = modelo_pca.fit_transform(X)

pca_resultado # Muestra la representación del dataset tras la aplicación de PCA.
```

```python
# Bloque 6 — Incorporación de los componentes principales
"""Este bloque añade los componentes principales al dataset.
La variable "df["comp1"]" almacena la primera columna generada por PCA.
La variable "df["comp2"]" almacena la segunda columna generada por PCA.
La incorporación permite visualizar el dataset en el espacio de los componentes principales."""

df["comp1"] = pca_resultado[:, 0]
df["comp2"] = pca_resultado[:, 1]

df # Muestra el dataset con los componentes principales incorporados.
```
