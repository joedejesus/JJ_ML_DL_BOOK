```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características numéricas sin etiquetas.
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena un dataset con columnas adecuadas para aplicar clustering.
El dataset sirve como base para interpretar agrupaciones generadas por k-means."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41, 50, 31, 28],
    "ingresos": [1500, 2300, 3200, 1800, 2900, 4100, 2000, 1700]
})

df # Muestra el dataset creado para su revisión.
```

```python
# Bloque 2 — Selección de características
"""Este bloque selecciona las características del dataset.
La variable "X" almacena las columnas de entrada.
La selección permite preparar los datos para el algoritmo de clustering. Los ingresos tienen una escala mayor y pueden dominar las distancias."""

X = df[["edad", "ingresos"]]

X # Muestra las características seleccionadas para su revisión.
```

```python
# Bloque 3 — Aplicación de k-means
"""Este bloque aplica el algoritmo k-means.
Este bloque importa "KMeans" de "sklearn.cluster" para crear el modelo de clustering.

Aplicamos la clase "KMeans()" a la variable "modelo" para crear un modelo de clustering.
El parámetro "n_clusters=" define el número de clusters que se desean formar.
El parámetro "random_state=" fija la semilla del generador de números aleatorios para reproducir la inicialización.
La variable "modelo" almacena el agrupador configurado con un número específico de clusters.

Aplicamos el método ".fit_predict()" al modelo con las características seleccionadas "X" para obtener las etiquetas 
de los clusters asignados a cada instancia.
La variable "clusters" almacena las etiquetas de los clusters asignados a cada instancia.
La aplicación del algoritmo k-means permite identificar grupos de instancias similares dentro del conjunto de datos."""

from sklearn.cluster import KMeans

modelo = KMeans(n_clusters=2, random_state=0)
clusters = modelo.fit_predict(X)

clusters # Muestra las etiquetas de los clusters asignados a cada instancia.
```

```python
# Bloque 4 — Incorporación de los clusters al dataset
"""Este bloque añade los clusters generados al dataset original.
La variable "df["cluster"]" almacena una nueva columna con las etiquetas de los clusters asignadas a cada instancia.
La incorporación de los clusters al dataset permite analizar los grupos obtenidos por el algoritmo."""

df["cluster"] = clusters

df # Muestra el dataset con la nueva columna de clusters.
```

```python
# Bloque 5 — Obtención de los centroides
"""Este bloque obtiene los centroides del modelo k-means.
Accedemos al atributo "cluster_centers_" del modelo para obtener los centroides.
La variable "centroides" almacena las coordenadas de cada centroide.
La extracción permite interpretar la posición representativa de cada grupo."""

centroides = modelo.cluster_centers_

centroides # Muestra los centroides para su interpretación.
```

```python
# Bloque 6 — Cálculo de estadísticas por cluster
"""Este bloque calcula estadísticas descriptivas por cluster.
Calculamos la media de cada columna dentro de cada cluster con el método ".mean()" de pandas.
Aplicamos el método ".groupby()" de pandas para agrupar los datos por cluster.
La variable "resumen" almacena medias de cada característica dentro de cada grupo.
El cálculo permite identificar diferencias relevantes entre clusters."""

resumen = df.groupby("cluster")[["edad", "ingresos"]].mean()

resumen # Muestra el resumen para su interpretación.
```

```python
# Bloque 7 — Interpretación estructural
"""Este bloque construye un diccionario llamado "interpretacion" para resumir cada cluster.
Cada clave "c" representa la etiqueta de un cluster obtenida previamente en "resumen.index".

Para cada cluster, se crea como valor otro diccionario con tres elementos descriptivos.
"centroide" toma "centroides[c].tolist()" para convertir a lista las coordenadas del centroide del cluster "c".
"media_edad" toma "resumen.loc[c, \"edad\"]" para recuperar la edad media de ese cluster.
"media_ingresos" toma "resumen.loc[c, \"ingresos\"]" para recuperar el ingreso medio de ese cluster.

La comprensión de diccionario recorre todos los clusters con "for c in resumen.index" y genera una entrada por cada uno.
El resultado final permite leer de forma compacta la posición y las características promedio de cada grupo."""

interpretacion = {
    c: {
        "centroide": centroides[c].tolist(),
        "media_edad": resumen.loc[c, "edad"],
        "media_ingresos": resumen.loc[c, "ingresos"]
    }
    for c in resumen.index
}

interpretacion # Muestra la interpretación estructural de los clusters.
```
