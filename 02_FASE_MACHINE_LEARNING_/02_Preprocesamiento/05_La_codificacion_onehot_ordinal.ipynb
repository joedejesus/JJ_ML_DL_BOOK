```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con variables categóricas.
Este bloque importa "pandas" para manipular datos en forma de DataFrame.
La variable "df" almacena un DataFrame con columnas que requieren codificación.
El dataset sirve como entrada para las transformaciones."""

import pandas as pd

df = pd.DataFrame({
    "ciudad": ["A", "B", "A", "C", "B"],
    "nivel": ["bajo", "medio", "alto", "medio", "alto"]
})

df # Muestra el DataFrame.
```

```python
# Bloque 2 — Codificación one-hot
"""Este bloque aplica codificación one-hot.
Este bloque importa "OneHotEncoder" de "sklearn.preprocessing" para realizar la codificación one-hot.
Aplicamos la clase "OneHotEncoder()" a la variable "onehot" para crear un objeto codificador de columnas categóricas.
El parámetro "sparse_output=False" indica que la salida debe ser un array denso en lugar de una matriz dispersa.

Aplicamos el método "fit_transform()" al codificador "onehot" indicando la columna a codificar.
El argumento "df[["ciudad"]]" indica la columna a codificar.

La variable "cod_onehot" almacena columnas binarias para la columna "ciudad".
La codificación permite representar categorías sin orden implícito."""

from sklearn.preprocessing import OneHotEncoder

onehot = OneHotEncoder(sparse_output=False)
cod_onehot = onehot.fit_transform(df[["ciudad"]])

cod_onehot # Muestra la codificación one-hot resultante.
```

```python
# Bloque 3 — Conversión del resultado one-hot
"""Este bloque convierte los datos one-hot en un DataFrame.
La clase "pd.DataFrame()" se utiliza para convertir el array numpy en un DataFrame.
Los argumentos de "pd.DataFrame()" son el array numpy que se desea convertir y las columnas que se desean asignar al DataFrame "columns=".
El valor de "columns=" se obtiene a partir del método "get_feature_names_out()" aplicado al codificador one-hot con el argumento que indica 
las columnas originales que se codificaron, en este caso la columna "ciudad".
La variable "df_onehot" almacena los datos codificados en forma de DataFrame.
La conversión permite inspeccionar la representación codificada."""

df_onehot = pd.DataFrame(cod_onehot, columns=onehot.get_feature_names_out(["ciudad"]))

df_onehot # Muestra el DataFrame con la codificación one-hot.
```

```python
# Bloque 4 — Codificación ordinal
"""Este bloque aplica codificación ordinal.
Este bloque importa "OrdinalEncoder" de "sklearn.preprocessing" para realizar la codificación ordinal.
Aplicamos la clase "OrdinalEncoder()" a la variable "ordinal" para crear un objeto codificador de columnas categóricas.
El parámetro "categories=[["bajo", "medio", "alto"]]" indica el orden de las categorías.

Aplicamos el método "fit_transform()" al codificador "ordinal" indicando la columna a codificar.
El argumento "df[["nivel"]]" indica la columna a codificar.

La variable "cod_ordinal" almacena códigos ordinales para la columna "nivel" en un array de valores flotantes.
La codificación permite representar categorías con jerarquía explícita."""

from sklearn.preprocessing import OrdinalEncoder

ordinal = OrdinalEncoder(categories=[["bajo", "medio", "alto"]])
cod_ordinal = ordinal.fit_transform(df[["nivel"]])

cod_ordinal # Muestra los códigos ordinales correspondientes a la columna "nivel".
```

```python
# Bloque 5 — Conversión del resultado ordinal
"""Este bloque convierte los datos ordinales en un DataFrame.
La clase "pd.DataFrame()" se utiliza para convertir el array numpy en un DataFrame.
Los argumentos de "pd.DataFrame()" son el array numpy que se desea convertir y las columnas que se desean asignar al DataFrame "columns=".
El valor de "columns=" se asigna manualmente según las columnas codificadas, en este caso ["nivel_ordinal"].
La variable "df_ordinal" almacena los códigos ordinales asignados en forma de DataFrame.
La conversión permite observar la estructura jerárquica aplicada."""

df_ordinal = pd.DataFrame(cod_ordinal, columns=["nivel_ordinal"])

df_ordinal # Muestra el DataFrame con los valores ordinales asignados.
```
