```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset con características y etiqueta.
Este bloque importa "pandas" para trabajar con datos estructurados.
La variable "df" almacena instancias que serán divididas en entrenamiento y test.
El dataset sirve como entrada para la partición controlada."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900],
    "compra": [0, 1, 1, 0, 1]
})

df # Muestra el DataFrame con las instancias del dataset.
```

```python
# Bloque 2 — Separación de features y label
"""Este bloque separa las características y la etiqueta.
La variable "X" almacena las columnas de entrada "características".
La variable "y" almacena la columna objetivo "etiqueta".
La separación permite definir el material para la división."""

X = df[["edad", "ingresos"]]
y = df["compra"]

X, y # Muestra las características y la etiqueta separadas.
```

```python
# Bloque 3 — División con train_test_split
"""Este bloque aplica train_test_split.
Este bloque importa la función "train_test_split" de "sklearn.model_selection" para dividir los datos en conjuntos de entrenamiento y prueba.
La función "train_test_split()" divide los datos en conjuntos de entrenamiento y prueba.
Los parámetros de "train_test_split()" incluyen los datos de entrada X, las etiquetas y, el tamaño del conjunto de prueba "test_size" y la semilla aleatoria "random_state".
La división evita que el modelo sea evaluado con datos ya vistos."""

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=0)

X_train, X_test, y_train, y_test # Muestra las particiones de entrenamiento y test.
```

```python
# Bloque 4 — Verificación de tamaños
"""Este bloque verifica los tamaños de cada subconjunto.
Usamos la propiedad "shape" para obtener las dimensiones de cada subconjunto y almacenarlas en las variables correspondientes.
Las variables "tam_train" y "tam_test" almacenan las dimensiones resultantes.
La verificación permite comprobar los tamaños de las particiones."""

tam_train = (X_train.shape, y_train.shape)
tam_test = (X_test.shape, y_test.shape)

tam_train, tam_test # Muestra los tamaños de los subconjuntos de entrenamiento y prueba.
```

```python
# Bloque 5 — Entrenamiento simple para validar la división
"""Este bloque entrena un modelo básico para validar la división.
Este bloque importa "LogisticRegression" de "sklearn.linear_model" y "accuracy_score" de "sklearn.metrics" para entrenar y evaluar un modelo básico.
La clase "LogisticRegression()" crea un clasificador de regresión logística almacenado en la variable "modelo".

Aplicamos el método ".fit()" al clasificador con los datos de entrenamiento "X_train" y "y_train" como argumentos.
La variable "modelo" almacena el clasificador ajustado.

Aplicamos el método ".predict()" al clasificador con los datos de prueba "X_test" para obtener las predicciones.
La variable "pred" almacena las predicciones del modelo en el conjunto de test.

Aplicamos la función "accuracy_score()" con las etiquetas verdaderas "y_test" y las predicciones "pred" para evaluar la exactitud del modelo.
La variable "precision" almacena el rendimiento en el conjunto de test.
La validación demuestra el uso práctico de la partición."""

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

modelo = LogisticRegression()
modelo.fit(X_train, y_train)
pred = modelo.predict(X_test)
precision = accuracy_score(y_test, pred)

precision # Muestra la exactitud del modelo en el conjunto de test.
```
