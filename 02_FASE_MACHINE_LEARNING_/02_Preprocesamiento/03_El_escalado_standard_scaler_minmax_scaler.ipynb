```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset numérico.
Este bloque importa "pandas" para trabajar con estructuras de datos tabulares.
La variable "df" almacena columnas que requieren escalado.
El dataset sirve como entrada para las transformaciones."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900]
})

df # Muestra el dataset creado.
```

```python
# Bloque 2 — Aplicación de StandardScaler
"""Este bloque aplica StandardScaler.
Este bloque importa "StandardScaler" de "sklearn.preprocessing" para estandarizar los datos.
La variable "scaler_std" almacena el objeto de estandarización mediante la clase "StandardScaler".
Aplicamos el método "fit_transform()" con el argumento "df" al objeto "scaler_std" para obtener los datos estandarizados.
La variable "datos_std" almacena valores centrados y escalados en forma de array numpy."""

from sklearn.preprocessing import StandardScaler

scaler_std = StandardScaler()
datos_std = scaler_std.fit_transform(df)

datos_std # Muestra los datos estandarizados.
```

```python
# Bloque 3 — Conversión del resultado estandarizado
"""Este bloque convierte los datos estandarizados en un DataFrame.
La clase "pd.DataFrame()" se utiliza para convertir el array numpy en un DataFrame.
Los argumentos de "pd.DataFrame()" son el array numpy que se desea convertir y las columnas que se desean asignar al DataFrame.
La variable "df_std" almacena los valores con media cero y desviación estándar poblacional uno en forma de DataFrame.
La conversión permite inspeccionar el resultado del escalado."""

df_std = pd.DataFrame(datos_std, columns=df.columns)

df_std # Muestra el DataFrame con los datos estandarizados.
```

```python
# Bloque 4 — Aplicación de MinMaxScaler
"""Este bloque aplica MinMaxScaler.
Este bloque importa "MinMaxScaler" de "sklearn.preprocessing" para normalizar los datos.
La variable "scaler_mm" almacena el objeto de normalización mediante la clase "MinMaxScaler()".
Aplicamos el método "fit_transform()" con el argumento "df" al objeto "scaler_mm" para obtener los datos normalizados.
La variable "datos_mm" almacena valores reescalados entre 0 y 1 en forma de array numpy."""

from sklearn.preprocessing import MinMaxScaler

scaler_mm = MinMaxScaler()
datos_mm = scaler_mm.fit_transform(df)

datos_mm # Muestra los datos normalizados.
```

```python
# Bloque 5 — Conversión del resultado normalizado
"""Este bloque convierte los datos normalizados en un DataFrame.
La clase "pd.DataFrame()" se utiliza para convertir el array numpy en un DataFrame.
Los argumentos de "pd.DataFrame()" son el array numpy que se desea convertir y las columnas que se desean asignar al DataFrame.
La variable "df_mm" almacena los valores ajustados al rango 0-1.     
La conversión permite comparar ambos métodos de escalado."""

df_mm = pd.DataFrame(datos_mm, columns=df.columns)     

df_mm # Muestra el DataFrame con los datos normalizados.
```
