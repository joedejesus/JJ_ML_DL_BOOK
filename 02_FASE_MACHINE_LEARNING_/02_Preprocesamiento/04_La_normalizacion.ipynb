```python
# Bloque 1 — Creación del dataset
"""Este bloque crea un dataset numérico.
Este bloque importa "pandas" para manipular datos en forma de DataFrame.
La variable "df" almacena un DataFrame con columnas que requieren normalización.
El dataset sirve como entrada para el reescalado."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, 47, 29, 41],
    "ingresos": [1500, 2300, 3200, 1800, 2900]
})

df # Muestra el DataFrame.
```

```python
# Bloque 2 — Cálculo de mínimos y máximos
"""Este bloque calcula mínimos y máximos por columna.
El método ".min()" calcula el valor mínimo por columna, mientras que el método ".max()" calcula el valor máximo por columna.
Las variables "minimos" y "maximos" almacenan los límites de cada característica del DataFrame.
El cálculo permite definir el rango para la normalización."""

minimos = df.min()
maximos = df.max()

minimos, maximos # Muestra los valores mínimos y máximos por columna.
```

```python
# Bloque 3 — Aplicación de normalización min-max
"""Este bloque aplica normalización min-max.
La variable "df_normalizado" almacena valores reescalados entre 0 y 1 en forma de DataFrame.
Aplicamos la fórmula de normalización min-max: (valor - mínimo) / (máximo - mínimo).
En este caso, el valor es el DataFrame original "df", ya que queremos normalizar todas sus columnas.
La normalización ajusta cada columna según sus límites."""

df_normalizado = (df - minimos) / (maximos - minimos)

df_normalizado # Muestra el DataFrame normalizado.

# En este caso, la conversión no tendría sentido, ya que "df_normalizado" ya es un DataFrame y no un "array" de numpy.
```

```python
# Bloque 4 — Comparación entre valores originales y normalizados
"""Este bloque compara valores originales y normalizados.
Utilizamos la función "pd.concat()" para concatenar el DataFrame original y el DataFrame normalizado.
Los argumentos de la función "pd.concat()" son una lista de DataFrames a concatenar.
El parámetro "axis=1" indica que la concatenación se realiza a lo largo de las columnas.
La variable "comparacion" almacena ambas versiones del dataset.
La comparación permite observar el efecto del reescalado."""     
 
comparacion = pd.concat([df, df_normalizado], axis=1)

comparacion # Muestra la comparación entre los valores originales y normalizados.
```
