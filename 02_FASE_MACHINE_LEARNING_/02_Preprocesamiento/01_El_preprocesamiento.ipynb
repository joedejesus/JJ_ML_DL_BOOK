```python
# Bloque 1 — Carga del dataset
"""Este bloque carga un dataset tabular.
Este bloque importa "pandas" para trabajar con datos estructurados en forma de DataFrame.
La variable "df" almacena instancias con columnas numéricas y categóricas que contienen valores nulos.
El dataset sirve como base para aplicar operaciones de preprocesamiento."""

import pandas as pd

df = pd.DataFrame({
    "edad": [22, 35, None, 29, 41],
    "ingresos": [1500, None, 3200, 1800, 2900],
    "ciudad": ["A", "B", "A", None, "C"]
})

df # Muestra el DataFrame para visualizar los datos cargados, incluyendo valores nulos.
```

```python
# Bloque 2 — Detección de valores faltantes
"""Este bloque detecta valores faltantes.
La variable "faltantes" almacena el conteo de valores nulos por columna.
El método ".isnull()" identifica los valores nulos en el DataFrame.
El método ".sum()" cuenta la cantidad de valores nulos por columna.
La detección permite identificar problemas de calidad en el dataset."""

faltantes = df.isnull().sum()

faltantes # Muestra el conteo de valores nulos por columna para evaluar la calidad del dataset.
```

```python
# Bloque 3 — Imputación de valores faltantes
"""Este bloque imputa valores faltantes.
La variable "df_imputado" almacena el dataset con los valores faltantes rellenados.
El método ".copy()" crea una copia del DataFrame original para evitar modificarlo directamente.
El método ".fillna()" reemplaza los valores nulos numéricos con la media y los de "ciudad" con "Desconocido".
La media se calcula utilizando el método ".mean()" aplicado a la columna numérica correspondiente.
La imputación permite tratar valores nulos antes de entrenar modelos que no los aceptan."""

df_imputado = df.copy()

df_imputado["edad"] = df_imputado["edad"].fillna(df_imputado["edad"].mean())
df_imputado["ingresos"] = df_imputado["ingresos"].fillna(df_imputado["ingresos"].mean())
df_imputado["ciudad"] = df_imputado["ciudad"].fillna("Desconocido")

df_imputado # Muestra el DataFrame imputado para visualizar los datos después de la imputación de valores faltantes.
```

```python
# Bloque 4 — Normalización
"""Este bloque aplica normalización min-max.
La variable "normalizado" almacena columnas reescaladas entre 0 y 1.
La variable "minimos" almacena los valores mínimos de las columnas numéricas mediante el método ".min()".
La variable "maximos" almacena los valores máximos de las columnas numéricas mediante el método ".max()".
Se aplica la fórmula de normalización: (valor - mínimo) / (máximo - mínimo).
En este caso, el valor es "df_imputado[["edad", "ingresos"]]", ya que se normalizan las columnas "edad" e "ingresos".
La normalización ajusta las variables numéricas a un rango uniforme."""

minimos = df_imputado[["edad", "ingresos"]].min()
maximos = df_imputado[["edad", "ingresos"]].max()

normalizado = (df_imputado[["edad", "ingresos"]] - minimos) / (maximos - minimos)

normalizado # Muestra el DataFrame normalizado para visualizar los datos reescalados entre 0 y 1.
```

```python
# Bloque 5 — Codificación categórica
"""Este bloque aplica codificación one-hot.
La variable "codificado" almacena columnas categóricas transformadas en variables binarias.
La función "pd.get_dummies()" convierte cada categoría en una columna separada con valores indicadores.
El argumento "df_imputado["ciudad"]" indica la columna a codificar.
La codificación permite que los modelos procesen variables no numéricas."""

codificado = pd.get_dummies(df_imputado["ciudad"])

codificado # Muestra el DataFrame codificado para visualizar las columnas binarias resultantes de la codificación one-hot.
```
