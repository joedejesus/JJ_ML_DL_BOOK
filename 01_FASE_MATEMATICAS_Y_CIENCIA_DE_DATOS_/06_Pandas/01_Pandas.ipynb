```python
# Bloque 1 — Creación de Series y DataFrames
"""Este bloque crea una Serie y un DataFrame.
Este bloque importa "pandas" para manejar datos estructurados.
La variable "serie" almacena datos unidimensionales con etiquetas.
La variable "df" almacena datos tabulares con filas y columnas.
Las estructuras permiten representar datos estructurados de forma flexible."""

import pandas as pd

serie = pd.Series([10, 20, 30], index=["a", "b", "c"])

df = pd.DataFrame({
    "edad": [25, 30, 45],
    "altura": [170, 180, 175]
})

serie, df # Muestra la Serie y el DataFrame para visualización.
```

```python
# Bloque 2 — Selección de datos
"""Este bloque selecciona datos mediante etiquetas y posiciones.
La variable "fila_label" almacena una fila seleccionada por etiqueta.
La variable "fila_pos" almacena una fila seleccionada por posición.
La selección permite acceder a subconjuntos específicos de datos."""

fila_label = df.loc[0]
fila_pos = df.iloc[1]

fila_label, fila_pos # Muestra las filas seleccionadas para visualización.
```

```python
# Bloque 3 — Filtrado de datos
"""Este bloque filtra filas según una condición.
La variable "filtrado_1" almacena las filas donde la edad es mayor que 30.
La variable "filtrado_2" almacena las filas donde la altura es menor que 180.
El filtrado permite extraer información relevante según criterios definidos."""

filtrado_1 = df[df["edad"] > 30]
filtrado_2 = df[df["altura"] < 180]

filtrado_1, filtrado_2 # Muestra los DataFrames filtrados para visualización.
```

```python
# Bloque 4 — Operaciones sobre columnas
"""Este bloque aplica operaciones vectorizadas sobre columnas.
La variable "nueva_columna" almacena una Serie con la altura convertida a metros.
Las operaciones sobre columnas permiten transformar datos de forma eficiente."""

nueva_columna = df["altura"] / 100

nueva_columna # Muestra la nueva Serie para visualización.
```

```python
# Bloque 5 — Estadísticas básicas
"""Este bloque calcula estadísticas sobre columnas del DataFrame.
La variable "media_edad" almacena la media de la edad.
La variable "max_altura" almacena la altura máxima.
Las estadísticas permiten analizar distribuciones de datos."""

media_edad = df["edad"].mean()
max_altura = df["altura"].max()

media_edad, max_altura # Muestra las estadísticas para visualización.
```
