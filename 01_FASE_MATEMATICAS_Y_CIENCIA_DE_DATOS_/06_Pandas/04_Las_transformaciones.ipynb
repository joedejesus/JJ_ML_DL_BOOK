```python
# Bloque 1 — Escalado Min‑Max
"""Este bloque aplica una transformación de escalado min-max.
Este bloque importa "pandas" y "numpy" para trabajar con estructuras de datos y datos numéricos.
La variable "datos" almacena valores numéricos.
Las variables "minimo" y "maximo" almacenan los extremos de cada columna.
La variable "escalado" almacena los datos reescalados entre 0 y 1.
La transformación ajusta las diferencias de escala entre columnas."""

import pandas as pd
import numpy as np

datos = pd.DataFrame({
    "x": [10, 20, 30],
    "y": [100, 150, 200]
})

minimo = datos.min()
maximo = datos.max()

escalado = (datos - minimo) / (maximo - minimo)

escalado # Muestra el DataFrame reescalado entre 0 y 1.
```

```python
# Bloque 2 — Estandarización
"""Este bloque aplica estandarización.
Las variables "media" y "desviacion" almacenan estadísticas por columna; la desviación es muestral.
La variable "estandarizado" almacena los datos con media cero y desviación estándar muestral uno.
La transformación facilita el uso de algoritmos sensibles a la escala."""

media = datos.mean()
desviacion = datos.std()

estandarizado = (datos - media) / desviacion

estandarizado # Muestra el DataFrame estandarizado con media cero y desviación estándar muestral uno.
```

```python
# Bloque 3 — Codificación One‑Hot
"""Este bloque aplica codificación one-hot.
La variable "categorias" almacena valores categóricos.
La variable "codificado" almacena las columnas binarias generadas.
La transformación convierte categorías en columnas indicadoras."""

categorias = pd.DataFrame({
    "color": ["rojo", "azul", "rojo"]
})

codificado = pd.get_dummies(categorias)

codificado # Muestra el DataFrame con codificación one-hot.
```

```python
# Bloque 4 — Transformación logarítmica
"""Este bloque aplica una transformación logarítmica.
La variable "valores" almacena números positivos.
La variable "logaritmo" almacena la transformación logarítmica.
La transformación reduce la influencia de valores grandes."""

valores = pd.Series([1, 10, 100, 1000])

logaritmo = np.log(valores)

logaritmo # Muestra la Serie con la transformación logarítmica.
```

```python
# Bloque 5 — Creación de nuevas variables      
"""Este bloque genera una nueva variable a partir de columnas existentes.
La variable "df" almacena datos numéricos.
La variable "nueva_columna" almacena una Serie con la suma de dos columnas.
La transformación permite crear características derivadas."""        
     
df = pd.DataFrame({
    "a": [1, 2, 3],
    "b": [4, 5, 6]
})

nueva_columna = df["a"] + df["b"]

nueva_columna # Muestra la Serie creada a partir de las dos columnas.
```
