```python
# Bloque 1 — Creación de Series
"""Este bloque crea una Serie con etiquetas.
Este bloque importa "pandas" para manejar datos estructurados.
La variable "serie" almacena valores unidimensionales asociados a etiquetas.
La Serie permite representar datos con identificación explícita."""

import pandas as pd

serie = pd.Series([100, 200, 300], index=["x", "y", "z"])

serie # Muestra la Serie para visualización.
```

```python
# Bloque 2 — Creación de DataFrames
"""Este bloque crea un DataFrame con filas y columnas.
La variable "df" almacena datos tabulares estructurados.
El DataFrame permite organizar información en columnas etiquetadas."""

df = pd.DataFrame({
    "ciudad": ["A", "B", "C"],
    "poblacion": [12000, 50000, 32000]
})

df # Muestra el DataFrame para visualización.
```

```python
# Bloque 3 — Selección de elementos
"""Este bloque selecciona elementos de Series y DataFrames.
La variable "valor_serie" almacena un valor de la Serie mediante etiqueta.
La variable "fila_df" almacena una fila del DataFrame mediante etiqueta.
La selección permite acceder a datos específicos de forma controlada."""

valor_serie = serie["y"]
fila_df = df.loc[0]

valor_serie, fila_df # Muestra el valor de la Serie y la fila del DataFrame para visualización.
```

```python
# Bloque 4 — Operaciones sobre columnas
"""Este bloque aplica operaciones vectorizadas sobre columnas del DataFrame.
La variable "poblacion_miles" almacena la población expresada en miles.
Las operaciones permiten transformar columnas de forma eficiente."""

poblacion_miles = df["poblacion"] / 1000

poblacion_miles # Muestra la población en miles para visualización.
```

```python
# Bloque 5 — Filtrado de filas
"""Este bloque filtra filas del DataFrame según una condición.
La variable "filtrado" almacena las filas donde la población supera un umbral.
El filtrado permite extraer subconjuntos relevantes de datos."""

filtrado = df[df["poblacion"] > 30000]

filtrado # Muestra el DataFrame filtrado para visualización.
```
