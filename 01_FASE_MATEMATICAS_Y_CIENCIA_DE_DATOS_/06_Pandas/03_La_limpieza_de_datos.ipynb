```python
# Bloque 1 — Detección de valores faltantes
"""Este bloque detecta valores faltantes en un DataFrame.
Este bloque importa "pandas" y "numpy" para trabajar con estructuras de datos y datos numéricos.
La variable "df" almacena datos con valores faltantes.
La variable "faltantes" almacena un conteo por columna.
La detección permite identificar problemas antes de procesar los datos."""

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "edad": [25, np.nan, 40],
    "altura": [170, 180, np.nan]
})

faltantes = df.isnull().sum()

df, faltantes # Muestra el DataFrame y el conteo de valores faltantes por columna.
```

```python
# Bloque 2 — Eliminación de filas con valores faltantes
"""Este bloque elimina filas que contienen valores faltantes.
La variable "df_limpio" almacena el DataFrame sin filas incompletas.
La eliminación permite trabajar solo con datos completos cuando es necesario."""

df_limpio = df.dropna()

df_limpio # Muestra el DataFrame sin filas con valores faltantes.
```

```python
# Bloque 3 — Imputación de valores faltantes
"""Este bloque imputa valores faltantes con la media de cada columna.
La variable "medias" almacena la media por columna.
La variable "df_imputado" almacena el DataFrame con valores imputados.
La imputación permite conservar filas sustituyendo los valores faltantes por estimaciones."""

medias = df.mean()
df_imputado = df.fillna(medias)

medias, df_imputado # Muestra la media por columna y el DataFrame con valores imputados.
```

```python
# Bloque 4 — Detección de duplicados
"""Este bloque detecta filas duplicadas.
La variable "duplicados" almacena un indicador por fila.
La detección permite identificar registros repetidos que pueden afectar el análisis; en este DataFrame no hay duplicados."""

duplicados = df.duplicated()

duplicados # Muestra un indicador de duplicados por fila.
```

```python
# Bloque 5 — Eliminación de duplicados
"""Este bloque elimina filas duplicadas.
La variable "df_sin_duplicados" almacena el DataFrame sin registros repetidos.
La eliminación de duplicados puede evitar que los registros repetidos sesguen el procesamiento posterior."""

df_sin_duplicados = df.drop_duplicates()

df_sin_duplicados # Muestra el DataFrame sin registros repetidos. 
```
