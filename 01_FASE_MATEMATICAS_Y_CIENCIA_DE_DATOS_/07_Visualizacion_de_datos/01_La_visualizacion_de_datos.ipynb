```python
# Bloque 1 — Gráfico de líneas
"""Este bloque genera un gráfico de líneas.
Este bloque importa "matplotlib" para crear visualizaciones.
La variable "valores" almacena datos numéricos.
La función "plt.plot()" dibuja el gráfico de líneas a partir de los datos.
La figura representa los valores según su posición en la lista.
La visualización permite identificar tendencias."""

import matplotlib.pyplot as plt

valores = [10, 15, 20, 18, 25]

plt.plot(valores)
plt.show() # Muestra el gráfico de líneas generado.
```

```python
# Bloque 2 — Gráfico de barras
"""Este bloque genera un gráfico de barras.
La variable "categorias" almacena etiquetas.
La variable "cantidades" almacena valores asociados.
La función "plt.bar()" dibuja el gráfico de barras a partir de los datos.
La visualización permite comparar magnitudes entre categorías."""

categorias = ["A", "B", "C"]
cantidades = [30, 50, 40]

plt.bar(categorias, cantidades)
plt.show() # Muestra el gráfico de barras generado.
            
```

```python
# Bloque 3 — Histograma
"""Este bloque genera un histograma.
La variable "datos" almacena valores numéricos.
La función "plt.hist()" dibuja el histograma con 5 intervalos a partir de los datos.
La figura muestra la distribución de frecuencias.
La visualización permite analizar la forma de la distribución."""

datos = [1, 2, 2, 3, 3, 3, 4, 4, 5]

plt.hist(datos, bins=5)
plt.show() # Muestra el histograma generado.
```

```python
# Bloque 4 — Gráfico de dispersión
"""Este bloque genera un gráfico de dispersión.
La variable "x" almacena valores del eje horizontal.
La variable "y" almacena valores del eje vertical.
La función "plt.scatter()" dibuja el gráfico de dispersión a partir de los datos.
La visualización permite identificar relaciones entre variables."""

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

plt.scatter(x, y)
plt.show() # Muestra el gráfico de dispersión generado.
```

```python
# Bloque 5 — Visualización con DataFrame
"""Este bloque genera un gráfico a partir de un DataFrame.
Este bloque importa "pandas" para manejar datos tabulares.
La variable "df" almacena datos tabulares.
La función "df["valor"].plot()" dibuja un gráfico a partir de la columna seleccionada del DataFrame.
La figura representa una columna del DataFrame mediante un gráfico de líneas.
La visualización permite explorar columnas de forma directa."""

import pandas as pd

df = pd.DataFrame({
    "valor": [5, 10, 15, 20, 25]
})

df["valor"].plot()
plt.show() # Muestra el gráfico generado a partir del DataFrame.
```
