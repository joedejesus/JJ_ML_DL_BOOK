```python
# Bloque 1 — Estilo y configuración inicial
"""Este bloque configura el estilo visual de Seaborn.
Este bloque importa "seaborn" y "matplotlib" para la visualización de datos.
La función "set_style()" establece un estilo predefinido.
La configuración permite controlar la apariencia general de los gráficos."""

import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("whitegrid") # Establece el estilo de fondo blanco con cuadrícula para los gráficos.
```

```python
# Bloque 2 — Gráfico de dispersión
"""Este bloque genera un gráfico de dispersión.
La variable "df" almacena el conjunto de datos Iris incluido en scikit-learn, sin necesidad de descargarlo.
La función "scatterplot()" representa relaciones entre dos variables.
El parámetro "data" indica el DataFrame que contiene los datos a graficar, en este caso, el DataFrame "df".
Los parámetros "x" e "y" especifican las columnas del DataFrame que se utilizarán para los ejes.
La visualización permite identificar patrones y posibles relaciones entre variables."""

from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.data.rename(columns={
    "sepal length (cm)": "sepal_length", "sepal width (cm)": "sepal_width",
    "petal length (cm)": "petal_length", "petal width (cm)": "petal_width"
})
df["species"] = iris.target.map(dict(enumerate(iris.target_names)))

sns.scatterplot(data=df, x="sepal_length", y="sepal_width")
plt.show() # Muestra el gráfico de dispersión generado.
```

```python
# Bloque 3 — Histograma con densidad
"""Este bloque genera un histograma con curva de densidad.
La función "histplot()" muestra la distribución de una variable.
El parámetro "data" indica el DataFrame que contiene los datos a graficar, en este caso, el DataFrame "df".
El parámetro "x" especifica la columna del DataFrame que se utilizará para el eje "x".
El parámetro "kde" añade la estimación de densidad a la visualización.
La visualización permite analizar la forma de la distribución."""

sns.histplot(data=df, x="petal_length", kde=True)
plt.show() # Muestra el histograma con curva de densidad generado.
```

```python
# Bloque 4 — Boxplot
"""Este bloque genera un boxplot.
La función "boxplot()" muestra la distribución y valores atípicos.
El parámetro "data" indica el DataFrame que contiene los datos a graficar, en este caso, el DataFrame "df".
Los parámetros "x" e "y" especifican las columnas del DataFrame que se utilizarán para los ejes.
La visualización permite comparar variaciones entre categorías."""

sns.boxplot(data=df, x="species", y="sepal_length")
plt.show() # Muestra el boxplot generado.
```

```python
# Bloque 5 — Heatmap
"""Este bloque genera un mapa de calor.
La variable "corr" almacena la matriz de correlación a partir del DataFrame "df" mediante el método "corr()".
El parámetro "numeric_only=True" asegura que solo se consideren columnas numéricas del DataFrame.
La función "heatmap()" representa correlaciones mediante colores.
Los parámetros "annot=True" y "cmap='coolwarm'" añaden anotaciones y un esquema de colores.
La visualización permite identificar correlaciones fuertes entre variables."""

corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show() # Muestra el mapa de calor generado.
```
