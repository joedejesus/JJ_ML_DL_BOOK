```python
# Bloque 1 — Figura y ejes
"""Este bloque crea una figura y un eje.
Este bloque importa "matplotlib" para crear visualizaciones.
Las variables "fig" y "ax" almacenan la figura y el eje, respectivamente, mediante la función "plt.subplots()".
La variable "ax" almacena el eje donde se dibujan los gráficos mediante la función "ax.plot()".
La estructura permite controlar la composición del gráfico."""

import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot([1, 2, 3], [2, 4, 6])  # Agrega una línea al eje con los puntos (1,2), (2,4) y (3,6).
plt.show()                     # Muestra la figura con el gráfico dibujado.
```

```python
# Bloque 2 — Configuración de títulos y etiquetas
"""Este bloque configura título y etiquetas de un gráfico.
Las variables "fig" y "ax" almacenan la figura y el eje, respectivamente, mediante la función "plt.subplots()".
La variable "ax" almacena el eje donde se dibujan los gráficos mediante la función "ax.plot()".
Configura el título del gráfico mediante la función "ax.set_title()".
Configura la etiqueta del eje "X" mediante la función "ax.set_xlabel()".
Configura la etiqueta del eje "Y" mediante la función "ax.set_ylabel()".
La configuración permite describir el contenido del gráfico."""

fig, ax = plt.subplots()
ax.plot([0, 1, 2], [3, 1, 4])

ax.set_title("Ejemplo de línea")
ax.set_xlabel("Eje X")
ax.set_ylabel("Eje Y")

plt.show() # Muestra la figura con el gráfico dibujado y las configuraciones aplicadas.
```

```python
# Bloque 3 — Gráfico de barras
"""Este bloque genera un gráfico de barras.
La variable "categorias" almacena etiquetas.
La variable "valores" almacena magnitudes asociadas.
La función "plt.bar()" genera un gráfico de barras con las categorías y valores.
El gráfico permite comparar cantidades entre categorías."""

categorias = ["A", "B", "C"]
valores = [5, 7, 3]

plt.bar(categorias, valores)
plt.show() # Muestra la figura con el gráfico de barras dibujado.
```

```python
# Bloque 4 — Gráfico de dispersión
"""Este bloque genera un gráfico de dispersión.
La variable "x" almacena valores horizontales.
La variable "y" almacena valores verticales.
La función "plt.scatter()" genera un gráfico de dispersión con los valores de "x" y "y".
El gráfico permite observar relaciones entre variables."""

x = [1, 2, 3, 4]
y = [2, 3, 5, 4]

plt.scatter(x, y)
plt.show() # Muestra la figura con el gráfico de dispersión dibujado.
```

```python
# Bloque 5 — Subplots
"""Este bloque crea múltiples subgráficos.
Las variables "fig" y "axes" almacenan la figura y los ejes, respectivamente, mediante la función "plt.subplots()".
La expresión "axes[0]" accede al primer eje, donde se dibuja un gráfico mediante "axes[0].plot()".
La expresión "axes[1]" accede al segundo eje, donde se dibuja un gráfico mediante "axes[1].bar()".
Los subplots permiten mostrar varias visualizaciones en una sola figura."""

fig, axes = plt.subplots(1, 2)

axes[0].plot([1, 2, 3], [1, 4, 9])       # Agrega una línea al primer eje con los puntos (1,1), (2,4) y (3,9).
axes[1].bar(["X", "Y", "Z"], [3, 6, 2])  # Agrega un gráfico de barras al segundo eje con las categorías "X", "Y" y "Z".

plt.show() # Muestra la figura con los subplots dibujados.
```
