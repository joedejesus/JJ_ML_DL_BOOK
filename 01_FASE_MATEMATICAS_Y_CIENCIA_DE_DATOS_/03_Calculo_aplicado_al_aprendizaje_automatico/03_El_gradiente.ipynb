<h1 style="margin:0 0 14px 0; font-size:34px; line-height:1.2; font-weight:700; color:#ffffff;">El Gradiente</h1>
<p style="margin:0; color:#cbd5e1; font-size:18px;">El gradiente reúne varias derivadas parciales en un único vector que resume cómo cambia una función en torno a un punto. En aprendizaje automático, esa información permite decidir cómo ajustar parámetros para reducir el error.</p>

```python
# Bloque 1 — Gradiente de una función simple
"""Este bloque calcula las derivadas parciales de la función escalar "f(x, y) = x**2 + y**2".
Este bloque importa "numpy" para manejar valores numéricos.
La variable "X" almacena valores para la primera variable.
La variable "Y" almacena valores para la segunda variable.
La variable "grad_X" almacena la derivada parcial respecto a "X".
La variable "grad_Y" almacena la derivada parcial respecto a "Y".
El gradiente reúne ambas derivadas para indicar la dirección de máximo incremento."""

import numpy as np

X = np.linspace(-3, 3, 100)
Y = np.linspace(-3, 3, 100)

grad_X = (2 * X)
grad_Y = (2 * Y)

grad_X, grad_Y # Muestra los valores de las derivadas parciales respecto a "X" e "Y", respectivamente.
```

```python
# Bloque 2 — Gradiente numérico mediante diferencia finita
"""Este bloque calcula el gradiente numérico usando diferencia finita.
La variable "h" almacena un incremento pequeño.
La variable "f" almacena los valores de la función "X**2 + Y**2".
La variable "grad_X_num" almacena la aproximación de la derivada parcial respecto a "X".
La variable "grad_Y_num" almacena la aproximación de la derivada parcial respecto a "Y".
El gradiente numérico permite estimar variaciones cuando no se dispone de la expresión analítica."""

h = 1e-5

f = (X**2) + (Y**2)

grad_X_num = ((X + h)**2 + Y**2 - f) / h
grad_Y_num = (X**2 + (Y + h)**2 - f) / h    

grad_X_num, grad_Y_num # Muestra los valores aproximados de las derivadas parciales respecto a "X" e "Y", respectivamente.
```

```python
# Bloque 3 — Gradiente en un punto específico
"""Este bloque evalúa el gradiente en un punto concreto.
La variable "x0" almacena el valor de "X" en el punto.
La variable "y0" almacena el valor de "Y" en el punto.
La variable "grad_x0" almacena la derivada parcial respecto a "X" en el punto.
La variable "grad_y0" almacena la derivada parcial respecto a "Y" en el punto.
La evaluación puntual permite analizar la sensibilidad local de la función."""

x0 = 1.0
y0 = -2.0

grad_x0 = (2 * x0)
grad_y0 = (2 * y0)

grad_x0, grad_y0 # Muestra los valores de las derivadas parciales respecto a "X" e "Y" en el punto específico.
```

```python
# Bloque 4 — Gradiente de una función de pérdida
"""Este bloque calcula expresiones para el gradiente de una función de pérdida cuadrática.
La variable "w" almacena el peso del modelo.
La variable "b" almacena el sesgo del modelo.
La variable "y_real" almacena el valor objetivo.
La variable "y_pred" almacena la predicción generada mediante "w*(X + b)".
La variable "perdida" almacena el error cuadrático.
La variable "grad_w" almacena una expresión para la derivada parcial respecto a "w".
La variable "grad_b" almacena una expresión para la derivada parcial respecto a "b".
Estas expresiones no corresponden a las derivadas de la pérdida calculada con la predicción "w*(X + b)"."""

w = 2.0
b = 1.0
y_real = 10.0

y_pred = w * (X + b)
perdida = (y_real - y_pred)**2

grad_w = -2 * (y_real - y_pred) * X
grad_b = -2 * (y_real - y_pred)

grad_w, grad_b # Muestra los valores calculados para las derivadas parciales respecto a "w" y "b", respectivamente.
```

```python
# Bloque 5 — Paso de actualización usando el gradiente
"""Este bloque aplica un paso de actualización basado en el gradiente.
La variable "tasa_aprendizaje" almacena el tamaño del ajuste.
La variable "w_actualizado" almacena el nuevo valor del parámetro "w".
La variable "b_actualizado" almacena el nuevo valor del parámetro "b".
El ajuste utiliza las expresiones calculadas en el bloque anterior, que no corresponden a la pérdida calculada con la predicción utilizada."""

tasa_aprendizaje = 0.01

w_actualizado = w - (tasa_aprendizaje * np.mean(grad_w))
b_actualizado = b - (tasa_aprendizaje * np.mean(grad_b))


w_actualizado, b_actualizado # Muestra los valores actualizados de los parámetros "w" y "b" después del paso de ajuste.

     
```
