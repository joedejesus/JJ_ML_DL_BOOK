```python
# Bloque 1 — Derivada de una función escalar
"""Este bloque importa "numpy" para manejar valores numéricos.
La variable "x" almacena un arreglo de puntos generado por "np.linspace()" donde se evaluará la función cuadrática.
La variable "f" almacena los valores de la función cuadrática "x**2".
La variable "derivada" almacena la derivada analítica "2*x".
La derivada cuantifica la variación local de la función en cada punto."""

import numpy as np

x = np.linspace(-5, 5, 100)
f = (x**2) 

derivada = (2 * x)

derivada # Muestra la derivada de la función cuadrática evaluada en los puntos de "x".
```

```python
# Bloque 2 — Aproximación numérica de la derivada
"""Este bloque calcula una aproximación numérica de la derivada.
La variable "h" almacena un incremento pequeño.
La variable "derivada_numerica" almacena el cociente incremental aplicado a la función x**2.
La aproximación permite estimar derivadas cuando no se dispone de la expresión analítica."""

h = 1e-5
derivada_numerica = ((x + h)**2 - x**2) / h

derivada_numerica # Muestra la aproximación numérica de la derivada evaluada en los puntos de "x".
```

```python
# Bloque 3 — Gradiente de una función de dos variables
"""Este bloque calcula el gradiente de una función de dos variables.
La variable "X" almacena un arreglo bidimensional con valores para la primera variable.
La variable "Y" almacena un arreglo bidimensional con valores para la segunda variable.
La variable "grad_X" almacena la derivada parcial respecto a "X".
La variable "grad_Y" almacena la derivada parcial respecto a "Y".
El gradiente indica la dirección de máximo incremento de la función."""

X, Y = np.meshgrid(np.linspace(-3, 3, 50), np.linspace(-3, 3, 50))

grad_X = (2 * X)
grad_Y = (2 * Y)

grad_X, grad_Y # Muestra las derivadas parciales respecto a "X" y "Y" evaluadas en la malla de puntos.
```

```python
# Bloque 4 — Función de pérdida cuadrática
"""Este bloque define una función de pérdida cuadrática.
La variable "y_real" almacena un valor objetivo.
La variable "y_pred" almacena un valor predicho.
La variable "perdida" almacena el error cuadrático entre ambos valores.
La pérdida cuantifica la discrepancia entre predicción y objetivo."""

y_real = 10
y_pred = 7

perdida = (y_real - y_pred)**2

perdida # Muestra el error cuadrático entre la predicción y el valor objetivo.
```

```python
# Bloque 5 — Gradiente de la pérdida respecto a la predicción
"""Este bloque calcula el gradiente de la pérdida cuadrática respecto a la predicción.
La variable "grad_perdida" almacena la derivada de "(y_real - y_pred)**2" con respecto a "y_pred".
El gradiente indica cómo debe ajustarse la predicción para reducir la pérdida."""

grad_perdida = -2 * (y_real - y_pred)

grad_perdida # Muestra el gradiente de la pérdida respecto a la predicción.
```

```python
# Bloque 6 — Paso de descenso de gradiente
"""Este bloque aplica un paso de descenso de gradiente.
La variable "tasa_aprendizaje" almacena el tamaño del paso.
La variable "y_pred_actualizado" almacena la predicción ajustada mediante "y_pred - tasa_aprendizaje*grad_perdida".
El descenso de gradiente permite minimizar funciones de pérdida ajustando parámetros."""

tasa_aprendizaje = 0.1
y_pred_actualizado = y_pred - (tasa_aprendizaje * grad_perdida)

y_pred_actualizado # Muestra la predicción ajustada tras un paso de descenso de gradiente.
```
