<h1 style="margin:0 0 14px 0; font-size:34px; line-height:1.2; font-weight:700; color:#ffffff;">Las Derivadas Parciales</h1>
<p style="margin:0; color:#cbd5e1; font-size:18px;">Cuando una función depende de varias variables, ya no basta con una sola derivada para describir cómo cambia. Las derivadas parciales permiten estudiar el efecto de variar una variable mientras las demás se mantienen fijas.</p>
</div>

<div style="display:grid; gap:14px;">
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Por qué importa</div>
    <div style="color:#f8fafc;">Los modelos reales tienen varios parámetros y cada uno contribuye de forma distinta al error. Entender esa contribución es clave para optimizar.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Qué vas a aprender aquí</div>
    <div style="color:#f8fafc;">A calcular derivadas parciales, aproximarlas numéricamente, reunirlas en un gradiente y relacionarlas con una función de pérdida sencilla.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Regla práctica que debes recordar</div>
    <div style="color:#f8fafc;">Una derivada parcial analiza el cambio respecto a una variable concreta manteniendo las otras constantes durante ese cálculo.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Contexto mínimo</div>
    <div style="color:#f8fafc;">Si una función tiene varios parámetros, cada derivada parcial mide la sensibilidad local de la salida frente a uno de ellos.</div>
  </div>

```python
# Bloque 1 — Derivadas parciales de una función simple
"""Este bloque calcula las derivadas parciales analíticas de la función "x**2 + 1.5*y**2".
Este bloque importa "numpy" para manejar valores numéricos.
La variable "X" almacena valores para la primera variable.
La variable "Y" almacena valores para la segunda variable.
La variable "derivada_X" almacena la derivada parcial respecto a "X".
La variable "derivada_Y" almacena la derivada parcial respecto a "Y".
Las derivadas parciales cuantifican la variación de la función en cada dirección."""

import numpy as np

X = np.linspace(-3, 3, 100)
Y = np.linspace(-3, 3, 100)

derivada_X = (2 * X)
derivada_Y = (3 * Y)

derivada_X, derivada_Y # Muestra las derivadas parciales respecto a "X" y "Y".
```

```python
# Bloque 2 — Derivadas parciales mediante diferencia finita
"""Este bloque calcula derivadas parciales numéricas usando diferencia finita.
La variable "h" almacena un incremento pequeño.
La variable "f" almacena los valores de la función "X**2 + Y**2" evaluada en los pares de valores de "X" e "Y".
La variable "derivada_X_num" almacena la aproximación de la derivada parcial respecto a "X".
La variable "derivada_Y_num" almacena la aproximación de la derivada parcial respecto a "Y".
La aproximación permite estimar derivadas cuando no se dispone de la expresión analítica. Esta función es distinta de la usada en el bloque anterior."""

h = 1e-5

f = (X**2) + (Y**2)

derivada_X_num = ((X + h)**2 + Y**2 - f) / h
derivada_Y_num = (X**2 + (Y + h)**2 - f) / h   

derivada_X_num, derivada_Y_num # Muestra las aproximaciones de las derivadas parciales respecto a "X" y "Y".
```

```python
# Bloque 3 — Gradiente a partir de derivadas parciales
"""Este bloque construye el gradiente de una función usando derivadas parciales.
La variable "gradiente" almacena un arreglo bidimensional con las derivadas parciales analíticas respecto a "X" y "Y".
El gradiente indica la dirección de máximo incremento de la función."""

gradiente = np.array([derivada_X, derivada_Y])

gradiente # Muestra el gradiente de la función en cada par de valores de "X" e "Y".
```

```python
# Bloque 4 — Derivadas parciales en un punto específico
"""Este bloque evalúa derivadas parciales en un punto concreto.
La variable "x0" almacena el valor de "X" en el punto.
La variable "y0" almacena el valor de "Y" en el punto.
La variable "dX" almacena la derivada parcial respecto a "X" evaluada en el punto.
La variable "dY" almacena la derivada parcial respecto a "Y" evaluada en el punto.
La evaluación puntual permite analizar la sensibilidad local de la función."""

x0 = 1.5
y0 = -2.0

dX = (2 * x0)
dY = (3 * y0)

dX, dY # Muestra las derivadas parciales evaluadas en el punto (x0, y0).
```

```python
# Bloque 5 — Derivadas parciales en una función de pérdida
"""Este bloque calcula derivadas parciales de una función de pérdida de dos parámetros.
La variable "w" almacena un parámetro del modelo.
La variable "b" almacena un segundo parámetro.
La variable "y_real" almacena el valor objetivo.
La variable "y_pred" almacena la predicción generada mediante "w*X + b".
La variable "perdida" almacena el error cuadrático.
La variable "d_w" almacena la derivada parcial de la pérdida respecto a "w".
La variable "d_b" almacena la derivada parcial de la pérdida respecto a "b".
Las derivadas parciales permiten ajustar parámetros mediante optimización."""

w = 2.0
b = 1.0
y_real = 10.0

y_pred = (w * X) + b
perdida = (y_real - y_pred)**2

d_w = -2 * (y_real - y_pred) * X
d_b = -2 * (y_real - y_pred)

d_w, d_b # Muestra las derivadas parciales de la pérdida respecto a "w" y "b".
```
