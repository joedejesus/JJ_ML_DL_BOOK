```python
# Bloque 1 — Curvas de nivel y gradiente
"""Este bloque define una función escalar "f(x, y) = x**2 + y**2".
Este bloque importa "numpy" para manejar valores numéricos.
La variable "X" y la variable "Y" almacenan una cuadrícula bidimensional.
La variable "f" almacena los valores de la función en la cuadrícula.
La variable "grad_X" almacena la derivada parcial respecto a "X".
La variable "grad_Y" almacena la derivada parcial respecto a "Y".
El gradiente es perpendicular a las curvas de nivel y señala la dirección de máximo incremento."""

import numpy as np

X, Y = np.meshgrid(np.linspace(-3, 3, 100), np.linspace(-3, 3, 100))
f = (X**2) + (Y**2)

grad_X = (2 * X)
grad_Y = (2 * Y)

grad_X, grad_Y # Muestra los valores de las derivadas parciales en cada punto de la cuadrícula.
```

```python
# Bloque 2 — Magnitud del gradiente
"""Este bloque calcula la magnitud del gradiente.
La variable "magnitud" almacena la raíz cuadrada de la suma de los cuadrados de "grad_X" y "grad_Y".
La magnitud indica la intensidad del cambio de la función en cada punto."""

magnitud = np.sqrt(grad_X**2 + grad_Y**2)

magnitud # Muestra la magnitud del gradiente en cada punto de la cuadrícula.
```

```python
# Bloque 3 — Dirección del gradiente
"""Este bloque normaliza el gradiente para obtener su dirección.
La variable "direccion_X" almacena la componente "X" del gradiente normalizado.
La variable "direccion_Y" almacena la componente "Y" del gradiente normalizado.
La dirección del gradiente indica hacia dónde crece más rápidamente la función."""

direccion_X = (grad_X / magnitud)
direccion_Y = (grad_Y / magnitud)

direccion_X, direccion_Y # Muestra las componentes del gradiente normalizado en cada punto de la cuadrícula.    
```

```python
# Bloque 4 — Gradiente en un punto y vector perpendicular
"""Este bloque evalúa el gradiente en un punto específico.
La variable "x0" y la variable "y0" almacenan las coordenadas del punto.
La variable "grad_x0" y la variable "grad_y0" almacenan las derivadas parciales en ese punto.
La variable "vector_perpendicular" almacena un vector perpendicular al gradiente.
El vector es tangente a la curva de nivel en ese punto y perpendicular al gradiente."""

x0 = 1.0
y0 = 2.0

grad_x0 = (2 * x0)
grad_y0 = (2 * y0)

vector_perpendicular = np.array([-grad_y0, grad_x0])

vector_perpendicular # Muestra el vector perpendicular al gradiente en el punto (x0, y0).
```

```python
# Bloque 5 — Interpretación geométrica en descenso de gradiente
"""Este bloque aplica un paso de descenso de gradiente para ilustrar su interpretación geométrica.
La variable "tasa_aprendizaje" almacena el tamaño del paso.
La variable "punto" almacena el punto inicial.
La variable "gradiente_punto" almacena el gradiente evaluado en el punto.
La variable "punto_actualizado" almacena el punto movido en la dirección opuesta al gradiente.
El descenso de gradiente se mueve en la dirección de máximo decrecimiento de la función."""

tasa_aprendizaje = 0.1

punto = np.array([1.0, -1.0])
gradiente_punto = np.array([2 * punto[0], 2 * punto[1]])

punto_actualizado = punto - (tasa_aprendizaje * gradiente_punto)

punto_actualizado # Muestra el nuevo punto después de aplicar un paso de descenso de gradiente.
```
