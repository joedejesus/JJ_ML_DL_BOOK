```python
# Bloque 1 — Espacio muestral y eventos
"""Este bloque importa "numpy" para manejar valores numéricos.
La variable "espacio_muestral" almacena un arreglo con los posibles resultados de un experimento simple.
La variable "evento" almacena un subconjunto del espacio muestral representado mediante un arreglo.
Estas estructuras permiten definir formalmente los resultados posibles y los eventos de interés."""

import numpy as np

espacio_muestral = np.array([1, 2, 3, 4, 5, 6])
evento = np.array([2, 4, 6])

espacio_muestral, evento # Muestra el espacio muestral y el evento definido.
```

```python
# Bloque 2 — Probabilidad de un evento
"""Este bloque calcula la probabilidad de un evento dentro del espacio muestral.
La variable "prob_evento" almacena el cociente entre el número de elementos del evento y el número de elementos del espacio muestral.
El cálculo representa la probabilidad clásica basada en casos favorables y casos posibles, suponiendo que todos los resultados son equiprobables."""

prob_evento = (len(evento) / len(espacio_muestral))

prob_evento # Muestra la probabilidad del evento definido.
```

```python
# Bloque 3 — Probabilidad complementaria
"""Este bloque calcula la probabilidad complementaria de un evento.
La variable "prob_complementaria" almacena el valor 1 menos la probabilidad del evento.
La probabilidad complementaria representa la posibilidad de que el evento no ocurra."""

prob_complementaria = (1 - prob_evento)

prob_complementaria # Muestra la probabilidad complementaria del evento definido.
```

```python
# Bloque 4 — Probabilidad conjunta para eventos independientes
"""Este bloque calcula la probabilidad conjunta de dos eventos independientes.
La variable "prob_A" almacena la probabilidad de un primer evento.
La variable "prob_B" almacena la probabilidad de un segundo evento.
La variable "prob_conjunta" almacena el producto entre "prob_A" y "prob_B".
La probabilidad conjunta representa la ocurrencia simultánea de ambos eventos en el mismo espacio muestral."""

prob_A = 0.3
prob_B = 0.5

prob_conjunta = (prob_A * prob_B)

prob_conjunta # Muestra la probabilidad conjunta de los eventos definidos.  
```

```python
# Bloque 5 — Probabilidad condicional
"""Este bloque calcula una probabilidad condicional.
La variable "prob_A" almacena la probabilidad de un evento A.
La variable "prob_B" almacena la probabilidad de un evento B.
La variable "prob_A_y_B" almacena la probabilidad conjunta de A y B.
La variable "prob_condicional_A" almacena el cociente entre "prob_A_y_B" y "prob_B".
La probabilidad condicional cuantifica la probabilidad de A dado que B ha ocurrido."""

prob_A = 0.4
prob_B = 0.6
prob_A_y_B = 0.24

prob_condicional_A = (prob_A_y_B / prob_B)

prob_condicional_A # Muestra la probabilidad condicional de A dado B.
```
