```python
# Bloque 1 — Cálculo de medidas descriptivas básicas
"""Este bloque importa "numpy" para manejar datos numéricos.
La variable "datos" almacena un arreglo unidimensional con valores representativos.
La variable "media" almacena el promedio de los valores mediante "np.mean()".
La variable "mediana" almacena el valor central mediante "np.median()".
La variable "varianza" almacena la variabilidad de los datos mediante "np.var()".
La variable "desviacion_estandar" almacena la raíz cuadrada de la varianza mediante "np.std()".
Estas medidas permiten describir el comportamiento general de un conjunto de datos."""

import numpy as np

datos = np.array([12, 15, 14, 10, 18, 20, 13])

media = np.mean(datos)
mediana = np.median(datos)
varianza = np.var(datos)
desviacion_estandar = np.std(datos)

media, mediana, varianza, desviacion_estandar # Muestra los resultados de las medidas descriptivas básicas.
```

```python
# Bloque 2 — Cálculo de cuantiles
"""Este bloque calcula cuantiles para describir la distribución de los datos.
La variable "q25" almacena el cuantil 0.25 mediante "np.quantile()".
La variable "q50" almacena el cuantil 0.50, equivalente a la mediana.
La variable "q75" almacena el cuantil 0.75 mediante "np.quantile()".
Los cuantiles permiten identificar puntos clave dentro de la distribución."""

q25 = np.quantile(datos, 0.25)
q50 = np.quantile(datos, 0.50)
q75 = np.quantile(datos, 0.75)

q25, q50, q75 # Muestra los resultados de los cuantiles.
```

```python
# Bloque 3 — Cálculo de rango intercuartílico
"""Este bloque calcula el rango intercuartílico.
La variable "ric" almacena la diferencia entre "q75" y "q25".
El rango intercuartílico cuantifica la dispersión central de los datos."""

ric = (q75 - q25)
ric # Muestra el resultado del rango intercuartílico.
```

```python
# Bloque 4 — Detección básica de valores atípicos
"""Este bloque identifica valores atípicos mediante el criterio del rango intercuartílico.
La variable "limite_inferior" almacena el umbral inferior calculado como "q25 - 1.5*ric".
La variable "limite_superior" almacena el umbral superior calculado como "q75 + 1.5*ric".
La variable "atipicos" almacena los valores de "datos" que quedan fuera de los límites.
Este procedimiento permite detectar observaciones que se alejan del comportamiento general."""

limite_inferior = (q25 - (1.5 * ric))
limite_superior = (q75 + (1.5 * ric))

atipicos = datos[(datos < limite_inferior) | (datos > limite_superior)]

atipicos # Muestra los valores atípicos detectados en el conjunto de datos.
```

```python
# Bloque 5 — Distribución de frecuencias
"""Este bloque calcula la distribución de frecuencias de los datos.
La variable "valores_unicos" almacena los valores distintos mediante "np.unique()".
La variable "frecuencias" almacena la cantidad de apariciones de cada valor mediante "return_counts=True".
La distribución de frecuencias permite analizar la repetición de valores dentro del conjunto."""

valores_unicos, frecuencias = np.unique(datos, return_counts=True)

valores_unicos, frecuencias # Muestra los valores únicos y sus frecuencias correspondientes.
```
