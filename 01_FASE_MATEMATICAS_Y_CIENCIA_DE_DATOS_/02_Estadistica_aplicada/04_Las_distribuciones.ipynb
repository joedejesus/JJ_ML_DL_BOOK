```python
# Bloque 1 — Distribución empírica
"""Este bloque importa "numpy" para manejar datos numéricos.
La variable "datos" almacena un conjunto de valores representativos.
La variable "valores_unicos" almacena los valores distintos mediante "np.unique()".
La variable "frecuencias" almacena la cantidad de apariciones de cada valor único mediante "np.unique()".
La distribución empírica describe cómo se repiten los valores dentro del conjunto."""

import numpy as np

datos = np.array([3, 4, 4, 5, 6, 6, 6, 7, 8, 8])

valores_unicos, frecuencias = np.unique(datos, return_counts=True)

valores_unicos, frecuencias # Muestra los valores únicos y sus frecuencias correspondientes.
```

```python
# Bloque 2 — Distribución de probabilidad empírica
"""Este bloque convierte la distribución empírica en una distribución de probabilidad.
La variable "probabilidades" almacena las frecuencias normalizadas dividiendo cada frecuencia entre el total de datos.
La distribución de probabilidad asigna a cada valor la proporción de veces que aparece."""

probabilidades = (frecuencias / len(datos))

probabilidades # Muestra las probabilidades asociadas a cada valor único.
```

```python
# Bloque 3 — Función de distribución acumulada (FDA)
"""Este bloque calcula la función de distribución acumulada.
La variable "fda" almacena la suma acumulada de las probabilidades mediante "np.cumsum()".
La FDA empírica representa la proporción de datos menores o iguales a cada valor observado."""

fda = np.cumsum(probabilidades)

fda # Muestra la función de distribución acumulada (FDA).   
```

```python
# Bloque 4 — Simulación de una distribución uniforme
"""Este bloque genera valores siguiendo una distribución uniforme.
La variable "uniforme" almacena números aleatorios entre 0 y 1 mediante "np.random.uniform()".
La distribución uniforme continua asigna la misma densidad a los valores dentro del intervalo."""

uniforme = np.random.uniform(0, 1, size=10)

uniforme # Muestra los valores generados siguiendo una distribución uniforme.

```

```python
# Bloque 5 — Simulación de una distribución normal
"""Este bloque genera valores siguiendo una distribución normal.
La variable "normal" almacena números aleatorios con media 0 y desviación estándar 1 mediante "np.random.normal()".
La distribución normal modela fenómenos que se concentran alrededor de un valor central con dispersión simétrica."""

normal = np.random.normal(loc=0, scale=1, size=10)

normal # Muestra los valores generados siguiendo una distribución normal.
```

```python
# Bloque 6 — Comparación de medias entre distribuciones simuladas
"""Este bloque compara las medias de dos muestras simuladas.
La variable "media_uniforme" almacena la media de la muestra uniforme.
La variable "media_normal" almacena la media de la muestra normal.
La comparación permite observar las medias de estas muestras, que pueden variar entre ejecuciones."""

media_uniforme = np.mean(uniforme)
media_normal = np.mean(normal)

media_uniforme, media_normal # Muestra las medias de las muestras uniforme y normal.  
```
