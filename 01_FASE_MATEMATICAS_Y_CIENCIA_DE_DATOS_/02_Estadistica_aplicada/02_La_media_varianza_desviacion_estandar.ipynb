```python
# Bloque 1 — Cálculo de la media
"""Este bloque importa "numpy" para manejar datos numéricos.
La variable "datos" almacena un arreglo unidimensional con valores representativos.
La variable "media" almacena el promedio de los valores mediante "np.mean()".
El cálculo de la media resume la tendencia central del conjunto de datos."""

import numpy as np

datos = np.array([10, 12, 15, 18, 20, 22, 25])
media = np.mean(datos)

media # Muestra el valor de la media.
```

```python
# Bloque 2 — Cálculo de la varianza
"""Este bloque calcula la varianza del conjunto de datos.
La variable "varianza" almacena el promedio de las diferencias cuadráticas respecto a la media mediante "np.var()".
La varianza cuantifica la dispersión interna del conjunto de datos."""

varianza = np.var(datos)    

varianza # Muestra el valor de la varianza.
```

```python
# Bloque 3 — Cálculo de la desviación estándar
"""Este bloque calcula la desviación estándar del conjunto de datos.
La variable "desviacion_estandar" almacena la raíz cuadrada de la varianza mediante "np.std()".
La desviación estándar expresa la dispersión en las mismas unidades que los datos originales."""

desviacion_estandar = np.std(datos)

desviacion_estandar # Muestra el valor de la desviación estándar.
```

```python
# Bloque 4 — Comparación entre varianza y desviación estándar
"""Este bloque compara la varianza y la desviación estándar.
La variable "relacion" almacena la diferencia entre ambas medidas.
La desviación estándar es la raíz cuadrada de la varianza; sus unidades son distintas, por lo que la diferencia no compara directamente ambas medidas."""

relacion = (desviacion_estandar - varianza)

relacion # Muestra la diferencia entre la desviación estándar y la varianza.
```

```python
# Bloque 5 — Efecto de un cambio en los datos
"""Este bloque analiza cómo cambian las medidas al modificar los datos.
La variable "datos_modificados" almacena un nuevo arreglo con valores más dispersos.
La variable "media_mod" almacena la media de los datos modificados.
La variable "varianza_mod" almacena la varianza de los datos modificados.
La variable "desviacion_estandar_mod" almacena la desviación estándar de los datos modificados.
Este análisis permite observar cómo la dispersión afecta las medidas estadísticas."""

datos_modificados = np.array([5, 12, 20, 30, 45, 60])
media_mod = np.mean(datos_modificados)
varianza_mod = np.var(datos_modificados)
desviacion_estandar_mod = np.std(datos_modificados)

media_mod, varianza_mod, desviacion_estandar_mod # Muestra los valores de la media, varianza y desviación estándar de los datos modificados.
      
```
