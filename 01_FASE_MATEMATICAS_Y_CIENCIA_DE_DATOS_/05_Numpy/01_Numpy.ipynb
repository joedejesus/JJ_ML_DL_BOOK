```python
# Bloque 1 — Creación de arreglos
"""Este bloque crea arreglos unidimensionales y bidimensionales.
Este bloque importa "numpy" para manejar datos numéricos.
La variable "a" almacena un arreglo unidimensional.
La variable "b" almacena un arreglo bidimensional.
La creación de arreglos permite representar datos de forma compacta."""

import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([[1, 2], 
              [3, 4]])

a, b # Muestra los arreglos contenidos en las variables "a" y "b".
```

```python
# Bloque 2 — Operaciones vectorizadas
"""Este bloque aplica operaciones vectorizadas sobre arreglos.
La variable "suma" almacena la suma elemento a elemento.
La variable "producto" almacena el producto elemento a elemento.
Las operaciones vectorizadas evitan bucles explícitos y pueden mejorar el rendimiento."""

suma = (a + 10)
producto = (a * 2)

suma, producto # Muestra los resultados de las operaciones vectorizadas.
```

```python
# Bloque 3 — Indexación y slicing
"""Este bloque muestra indexación y slicing en arreglos.
La variable "primero" almacena el primer elemento del arreglo con el índice 0.
La variable "segmento" almacena una porción del arreglo desde el índice 1 hasta el índice 3 (excluyendo el índice 3).
La indexación permite acceder a posiciones específicas y extraer subconjuntos."""

primero = a[0]
segmento = a[1:3]

primero, segmento # Muestra los elementos extraídos mediante indexación y slicing.
```

```python
# Bloque 4 — Operaciones matriciales
"""Este bloque realiza operaciones matriciales.
La variable "matriz" almacena una matriz cuadrada.
La variable "transpuesta" almacena la matriz transpuesta.
La variable "producto_matriz" almacena el resultado de la multiplicación matricial entre la matriz y su transpuesta.
Las operaciones matriciales permiten cálculos algebraicos esenciales en ML."""

matriz = np.array([[1, 2], 
                   [3, 4]])

transpuesta = matriz.T
producto_matriz = matriz @ transpuesta

transpuesta, producto_matriz # Muestra la matriz transpuesta y el resultado de la multiplicación matricial.
```

```python
# Bloque 5 — Funciones estadísticas
"""Este bloque calcula estadísticas básicas sobre un arreglo.
La variable "media" almacena la media.
La variable "desviacion" almacena la desviación estándar.
Las variables "minimo" y "maximo" almacenan los valores extremos.
Las funciones estadísticas permiten analizar distribuciones de datos."""

media = np.mean(a)
desviacion = np.std(a)
minimo = np.min(a)
maximo = np.max(a)

media, desviacion, minimo, maximo # Muestra los resultados de las funciones estadísticas.
```
