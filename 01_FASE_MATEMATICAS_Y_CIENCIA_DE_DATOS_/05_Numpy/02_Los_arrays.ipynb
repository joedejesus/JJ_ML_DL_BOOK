```python
# Bloque 1 — Creación de arrays
"""Este bloque crea arrays unidimensionales y bidimensionales.
Este bloque importa "numpy" para manejar valores numéricos.
La variable "a" almacena un array unidimensional.
La variable "b" almacena un array bidimensional.
Los arrays permiten representar datos de forma estructurada y eficiente."""

import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([[1, 2], 
              [3, 4]])

a, b # Muestra los arrays unidimensional y bidimensional creados.
```

```python
# Bloque 2 — Propiedades de los arrays
"""Este bloque consulta propiedades fundamentales de los arrays.
La variable "dimensiones" almacena el número de dimensiones del array.
La variable "forma" almacena la forma del array.
La variable "tipo" almacena el tipo de datos del array.
Las propiedades permiten comprender la estructura interna del array."""

dimensiones = a.ndim
forma = a.shape
tipo = a.dtype

dimensiones, forma, tipo # Muestra las dimensiones, la forma y el tipo de datos del array "a".
```

```python
# Bloque 3 — Operaciones elementales
"""Este bloque aplica operaciones elementales sobre arrays.
La variable "suma" almacena la suma elemento a elemento.
La variable "multiplicacion" almacena la multiplicación elemento a elemento.
Las operaciones vectorizadas permiten cálculos rápidos sin bucles explícitos."""

suma = (a + 5)
multiplicacion = (a * 3)

suma, multiplicacion # Muestra la suma y la multiplicación elemento a elemento del array "a".
```

```python
# Bloque 4 — Indexación y slicing
"""Este bloque muestra indexación y slicing en arrays.
La variable "elemento" almacena un valor específico del array.
La variable "subarray" almacena una porción del array.
La indexación permite acceder y manipular posiciones concretas."""

elemento = a[2]
subarray = a[1:4]

elemento, subarray # Muestra el elemento específico y el subarray obtenido del array "a" mediante slicing.
```

```python
# Bloque 5 — Arrays multidimensionales
"""Este bloque manipula arrays multidimensionales.
La variable "matriz" almacena una matriz de dos dimensiones.
La variable "fila" almacena una fila extraída mediante slicing.
La variable "columna" almacena una columna extraída mediante slicing.
Los arrays multidimensionales permiten representar datos estructurados en varias dimensiones."""

matriz = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

fila = matriz[1, :]      # Extrae la fila que se encuentra en el índice 1 de la matriz. "[índice de fila, :]"
columna = matriz[:, 2]   # Extrae la columna que se encuentra en el índice 2 de la matriz. "[:, índice de columna]"

fila, columna # Muestra la fila y la columna extraídas de la matriz mediante slicing.   
```
