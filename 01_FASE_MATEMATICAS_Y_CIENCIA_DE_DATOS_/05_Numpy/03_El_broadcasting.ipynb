<h1 style="margin:0 0 14px 0; font-size:34px; line-height:1.2; font-weight:700; color:#ffffff;">El Broadcasting</h1>
<p style="margin:0; color:#cbd5e1; font-size:18px;">El broadcasting es un mecanismo de NumPy que permite combinar arrays de formas distintas cuando sus dimensiones son compatibles. Gracias a él pueden escribirse operaciones compactas sin replicar manualmente los datos.</p>
</div>

<div style="display:grid; gap:14px;">
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Por qué importa</div>
    <div style="color:#f8fafc;">Este comportamiento simplifica código de normalización, centrado y muchas transformaciones habituales en ciencia de datos.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Qué vas a aprender aquí</div>
    <div style="color:#f8fafc;">A aplicar broadcasting entre escalares y arrays, entre vectores fila y columna y en operaciones por columnas sobre matrices.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Regla práctica que debes recordar</div>
    <div style="color:#f8fafc;">Dos dimensiones solo pueden alinearse si coinciden o si una de ellas vale uno. Conviene revisar siempre las formas antes de operar.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Contexto mínimo</div>
    <div style="color:#f8fafc;">Broadcasting no significa copiar físicamente todos los datos en memoria, sino extender su compatibilidad lógica durante la operación.</div>
  </div>

```python
# Bloque 1 — Broadcasting entre array y escalar
"""Este bloque aplica broadcasting entre un array y un escalar.
Este bloque importa "numpy" para manejar valores numéricos.
La variable "a" almacena un array unidimensional.
La variable "resultado" almacena la suma del array con un escalar.
El broadcasting permite operar con el escalar y cada elemento del array."""

import numpy as np

a = np.array([1, 2, 3, 4])
resultado = (a + 10)

resultado # Muestra el resultado de la operación de broadcasting entre el array y el escalar.
```

```python
# Bloque 2 — Broadcasting entre arrays de distinta forma
"""Este bloque aplica broadcasting entre arrays con formas compatibles.
La variable "fila" almacena un array unidimensional.
La variable "columna" almacena un array de una columna.
La variable "combinado" almacena la suma mediante broadcasting.
El broadcasting extiende dimensiones de tamaño uno para permitir la operación."""

fila = np.array([1, 2, 3])
columna = np.array([[10], 
                    [20], 
                    [30]])

combinado = (fila + columna)

combinado # Muestra el resultado de la operación de broadcasting entre los arrays.
```

```python
# Bloque 3 — Broadcasting en operaciones con matrices
"""Este bloque aplica broadcasting para escalar columnas de una matriz.
La variable "matriz" almacena una matriz bidimensional.
La variable "escala" almacena un array unidimensional compatible.
La variable "resultado" almacena la multiplicación mediante broadcasting.
El broadcasting permite aplicar un factor distinto a cada columna."""

matriz = np.array([[1, 2, 3],
                   [4, 5, 6]])

escala = np.array([10, 20, 30])

resultado = (matriz * escala)

resultado # Muestra el resultado de la multiplicación elemento a elemento mediante broadcasting.
```

```python
# Bloque 4 — Broadcasting para centrar datos
"""Este bloque aplica broadcasting para centrar columnas de una matriz.
La variable "datos" almacena valores distribuidos en columnas.
La variable "media_col" almacena la media por columna.
La variable "centrado" almacena los datos menos la media mediante broadcasting.
El broadcasting permite restar la media de cada columna sin replicar datos."""

datos = np.array([[10, 50],
                  [20, 60],
                  [30, 70]])

media_col = np.mean(datos, axis=0)

centrado = (datos - media_col)

centrado # Muestra el resultado de la operación de broadcasting para centrar los datos.
```

```python
# Bloque 5 — Broadcasting en normalización por columnas
"""Este bloque aplica normalización min-max usando broadcasting.
La variable "datos" almacena valores distribuidos en columnas.
La variable "minimo" almacena los mínimos por columna.
La variable "maximo" almacena los máximos por columna.
La variable "normalizado" almacena los datos reescalados mediante broadcasting.
El broadcasting permite dividir cada columna por su propio rango sin duplicación."""

datos = np.array([[2, 100],
                  [5, 150],
                  [8, 200]])

minimo = np.min(datos, axis=0)
maximo = np.max(datos, axis=0)

normalizado = (datos - minimo) / (maximo - minimo)

normalizado # Muestra el resultado de la operación de broadcasting para la normalización por columnas.  
```
