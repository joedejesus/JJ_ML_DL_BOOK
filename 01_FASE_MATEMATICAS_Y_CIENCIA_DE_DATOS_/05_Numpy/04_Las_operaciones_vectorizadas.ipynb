<h1 style="margin:0 0 14px 0; font-size:34px; line-height:1.2; font-weight:700; color:#ffffff;">Las Operaciones Vectorizadas</h1>

```python
# Bloque 1 — Suma vectorizada
"""Este bloque aplica una suma vectorizada sobre un array.
Este bloque importa "numpy" para manejar valores numéricos.
La variable "a" almacena valores numéricos.
La variable "resultado" almacena la suma elemento a elemento con un escalar.
La operación se ejecuta sin bucles explícitos."""

import numpy as np

a = np.array([1, 2, 3, 4])

resultado = (a + 5)

resultado # Muestra el resultado de la suma vectorizada, que es un array con cada elemento de "a" incrementado en 5.
```

```python
# Bloque 2 — Multiplicación vectorizada
"""Este bloque aplica multiplicación vectorizada sobre un array.
La variable "a" almacena valores numéricos.
La variable "producto" almacena el resultado de multiplicar cada elemento por un escalar.
La operación se aplica a todo el array sin bucles explícitos."""

producto = (a * 3)   

producto # Muestra el resultado de la multiplicación vectorizada, que es un array con cada elemento de "a" multiplicado por 3.
```

```python
# Bloque 3 — Operaciones entre arrays
"""Este bloque aplica operaciones vectorizadas entre dos arrays de igual forma.
La variable "x" y la variable "y" almacenan arreglos unidimensionales.
La variable "suma" almacena la suma elemento a elemento.
La variable "resta" almacena la resta elemento a elemento.
Las operaciones se realizan sin iteraciones manuales."""

x = np.array([10, 20, 30])     
y = np.array([1, 2, 3])

suma = (x + y)
resta = (x - y)

suma, resta # Muestra el resultado de la suma y la resta vectorizadas, que son arrays con los resultados de las operaciones elemento a elemento.
```

```python
# Bloque 4 — Funciones vectorizadas
"""Este bloque aplica funciones vectorizadas sobre un array.
La variable "valores" almacena números positivos en forma de array.
La variable "raiz" almacena la raíz cuadrada de cada elemento del array.
Las funciones vectorizadas permiten transformar datos de forma eficiente."""

valores = np.array([4, 9, 16, 25])
raiz = np.sqrt(valores)  

raiz # Muestra el resultado de la función vectorizada, que es un array con la raíz cuadrada de cada elemento del array.
```

```python
# Bloque 5 — Operaciones vectorizadas en matrices
"""Este bloque aplica operaciones vectorizadas sobre una matriz.
La variable "matriz" almacena valores bidimensionales.
La variable "doble" almacena la matriz multiplicada por dos.
La variable "cuadrado" almacena el cuadrado de cada elemento.
Las operaciones vectorizadas permiten cálculos algebraicos sobre estructuras multidimensionales."""

matriz = np.array([[1, 2, 3],
                   [4, 5, 6]])

doble = (matriz * 2)
cuadrado = (matriz ** 2)

doble, cuadrado # Muestra la matriz multiplicada por dos y la matriz con cada elemento al cuadrado. 
```
