```python
# Bloque 1 — Normalización min‑max
"""Este bloque aplica normalización min-max a un conjunto de datos.
Este bloque importa "numpy" para manejar valores numéricos.
La variable "datos" almacena valores numéricos representativos.
La variable "minimo" almacena el valor mínimo del conjunto.
La variable "maximo" almacena el valor máximo del conjunto.
La variable "normalizado" almacena los valores reescalados al intervalo [0, 1].
La normalización preserva las proporciones entre las diferencias de los valores."""

import numpy as np

datos = np.array([10, 15, 20, 30, 45, 60])

minimo = np.min(datos)
maximo = np.max(datos)

normalizado = (datos - minimo) / (maximo - minimo)

normalizado # Muestra el resultado de la normalización.
```

```python
# Bloque 2 — Estandarización
"""Este bloque aplica estandarización a un conjunto de datos.
La variable "media" almacena la media del conjunto.
La variable "desviacion" almacena la desviación estándar del conjunto.
La variable "estandarizado" almacena los valores transformados a media cero y desviación estándar uno.
La estandarización permite comparar variables con escalas distintas."""

media = np.mean(datos)
desviacion = np.std(datos)

estandarizado = (datos - media) / desviacion

estandarizado # Muestra el resultado de la estandarización.
```

```python
# Bloque 3 — Comparación entre normalización y estandarización
"""Este bloque compara los resultados de ambas transformaciones.
La variable "rango_normalizado" almacena el rango del conjunto normalizado.
La variable "media_estandarizado" almacena la media del conjunto estandarizado.
La variable "desviacion_estandarizado" almacena la desviación estándar del conjunto estandarizado.
La comparación muestra cómo cada técnica modifica la escala de los datos."""

rango_normalizado = np.max(normalizado) - np.min(normalizado)
media_estandarizado = np.mean(estandarizado)
desviacion_estandarizado = np.std(estandarizado)

rango_normalizado, media_estandarizado, desviacion_estandarizado # Muestra los resultados de la comparación.
```

```python
# Bloque 4 — Normalización aplicada a múltiples características
"""Este bloque aplica normalización min-max a una matriz con varias características.
La variable "matriz" almacena valores distribuidos en columnas.
La variable "minimo_col" almacena los mínimos por columna.
La variable "maximo_col" almacena los máximos por columna.
La variable "normalizado_col" almacena la matriz normalizada por columnas.
La normalización por características permite preparar datos multivariados para modelos sensibles a la escala."""

matriz = np.array([[1, 50],
                   [3, 60],
                   [5, 80],
                   [7, 90]])

minimo_col = np.min(matriz, axis=0)
maximo_col = np.max(matriz, axis=0)

normalizado_col = (matriz - minimo_col) / (maximo_col - minimo_col)

normalizado_col # Muestra la matriz normalizada por columnas.
```

```python
# Bloque 5 — Estandarización aplicada a múltiples características
"""Este bloque aplica estandarización a una matriz con varias características.
La variable "media_col" almacena las medias por columna.
La variable "desviacion_col" almacena las desviaciones estándar por columna.
La variable "estandarizado_col" almacena la matriz transformada.
La estandarización puede mejorar la estabilidad de métodos que dependen de magnitudes y gradientes."""

media_col = np.mean(matriz, axis=0)
desviacion_col = np.std(matriz, axis=0)

estandarizado_col = (matriz - media_col) / desviacion_col

estandarizado_col # Muestra la matriz estandarizada por columnas.
```
