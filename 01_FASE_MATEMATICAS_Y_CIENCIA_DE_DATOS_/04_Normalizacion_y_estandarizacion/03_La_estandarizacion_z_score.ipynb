<h1 style="margin:0 0 14px 0; font-size:34px; line-height:1.2; font-weight:700; color:#ffffff;">La Estandarización Z-Score</h1>
<p style="margin:0; color:#cbd5e1; font-size:18px;">La estandarización z-score transforma los datos usando su media y su desviación estándar para expresar cada valor en función de cuánto se aparta del centro. En este notebook verás cómo aplicar esta idea en distintos formatos y cómo recuperar luego la escala original.</p>
</div>

<div style="display:grid; gap:14px;">
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Por qué importa</div>
    <div style="color:#f8fafc;">Muchos modelos trabajan mejor cuando las variables quedan centradas y con una dispersión comparable, especialmente si el entrenamiento depende de gradientes o de distancias.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Qué vas a aprender aquí</div>
    <div style="color:#f8fafc;">A calcular z-score en vectores y matrices, revertir la transformación y usar parámetros de entrenamiento para transformar nuevos datos.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Regla práctica que debes recordar</div>
    <div style="color:#f8fafc;">Si usas z-score en un flujo real, la media y la desviación deben calcularse con los datos de entrenamiento y reutilizarse después.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">
    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Contexto mínimo</div>
    <div style="color:#f8fafc;">Un valor estandarizado indica cuántas desviaciones estándar se separa del valor medio del conjunto de referencia.</div>

```python
# Bloque 1 — Cálculo básico del z‑score
"""Este bloque aplica estandarización z-score a un conjunto de datos unidimensional.
Este bloque importa "numpy" para manejar valores numéricos.
La variable "datos" almacena un conjunto de valores numéricos.
La variable "media" almacena la media del conjunto.
La variable "desviacion" almacena la desviación estándar del conjunto.
La variable "estandarizado" almacena los valores transformados a media cero y desviación estándar uno.
La estandarización expresa los valores en desviaciones estándar respecto a la media."""

import numpy as np

datos = np.array([10, 14, 18, 22, 30])

media = np.mean(datos)
desviacion = np.std(datos)

estandarizado = (datos - media) / desviacion

estandarizado # Muestra el resultado de la estandarización z-score del conjunto de datos.
```

```python
# Bloque 2 — Estandarización en múltiples características
"""Este bloque aplica estandarización z-score a una matriz con varias características.
La variable "matriz" almacena valores distribuidos en columnas.
La variable "media_col" almacena las medias por columna.
La variable "desviacion_col" almacena las desviaciones estándar por columna.
La variable "estandarizado_col" almacena la matriz transformada.
La estandarización por características permite comparar variables con unidades distintas."""

matriz = np.array([[2, 100],
                   [4, 150],
                   [6, 200],
                   [8, 300]])

media_col = np.mean(matriz, axis=0)
desviacion_col = np.std(matriz, axis=0)

estandarizado_col = (matriz - media_col) / desviacion_col

estandarizado_col # Muestra el resultado de la estandarización z-score de la matriz de características.
```

```python
# Bloque 3 — Inversión del z‑score
"""Este bloque revierte la estandarización z-score.
La variable "estandarizado" almacena valores previamente transformados.
La variable "original" almacena los valores reconstruidos mediante "media + estandarizado*desviacion".
La inversión permite recuperar la escala original cuando es necesaria para interpretación."""

original = media + (estandarizado * desviacion)

original # Muestra los valores reconstruidos a partir de la estandarización z-score.
```

```python
# Bloque 4 — Estandarización aplicada a predicciones
"""Este bloque estandariza predicciones generadas por un modelo.
La variable "predicciones" almacena valores producidos por un modelo hipotético.
La variable "media_pred" almacena la media de las predicciones.
La variable "desv_pred" almacena la desviación estándar de las predicciones.
La variable "pred_est" almacena las predicciones transformadas.
La estandarización expresa estas predicciones en desviaciones estándar respecto a su media."""

predicciones = np.array([0.5, 1.0, 1.5, 2.0, 3.0])

media_pred = np.mean(predicciones)
desv_pred = np.std(predicciones)

pred_est = (predicciones - media_pred) / desv_pred

pred_est # Muestra las predicciones estandarizadas.
```

```python
# Bloque 5 — Estandarización para entrenamiento y test
"""Este bloque aplica estandarización z-score usando parámetros del conjunto de entrenamiento.
La variable "train" almacena datos de entrenamiento.
La variable "test" almacena datos de prueba.
Las variables "media_train" y "desv_train" almacenan los parámetros del entrenamiento.
Las variables "train_est" y "test_est" almacenan los conjuntos estandarizados.
El uso de parámetros del entrenamiento evita fugas de información desde el conjunto de prueba."""

train = np.array([10, 20, 30, 40])
test = np.array([15, 35, 50])

media_train = np.mean(train)    
desv_train = np.std(train)

train_est = (train - media_train) / desv_train
test_est = (test - media_train) / desv_train   # Usamos media y desviación estándar del entrenamiento para estandarizar el test.    

train_est, test_est # Muestra los conjuntos de entrenamiento y prueba estandarizados usando parámetros del entrenamiento.
```
