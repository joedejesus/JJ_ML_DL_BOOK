<h1 style="margin:0 0 14px 0; font-size:34px; line-height:1.2; font-weight:700; color:#ffffff;">La Normalización Min-Max</h1>
<p style="margin:0; color:#cbd5e1; font-size:18px;">La normalización min-max reescala una variable usando sus valores mínimo y máximo para llevarla a un intervalo controlado. En este notebook verás cómo se aplica sobre arrays y matrices, y cómo puede invertirse para recuperar la escala original.</p>
</div>

    <div style="color:#f8fafc;">Esta transformación es especialmente útil cuando interesa mantener proporciones relativas dentro de un rango fijo, por ejemplo entre 0 y 1.</div>
  </div>

    <div style="color:#f8fafc;">A aplicar la fórmula min-max, usarla por columnas, revertirla y trasladarla a predicciones o a conjuntos de entrenamiento y test.</div>
  </div>
  <div style="padding:16px 18px; border:1px solid #475569; border-radius:14px; background:rgba(148,163,184,0.10);">

    <div style="color:#f8fafc;">Los valores mínimo y máximo que definen la transformación deben salir del conjunto de entrenamiento si más adelante habrá datos nuevos.</div>
  </div>

    <div style="margin:0 0 6px 0; font-size:14px; font-weight:700; letter-spacing:0.08em; text-transform:uppercase; color:#fbbf24;">Contexto mínimo</div>
    <div style="color:#f8fafc;">Reescalar no significa perder el orden relativo de los datos; significa expresarlos en una escala común más fácil de manejar.</div>

```python
# Bloque 1 — Cálculo básico de min‑max
"""Este bloque aplica normalización min-max a un conjunto de datos unidimensional.
Este bloque importa "numpy" para manejar valores numéricos.
La variable "datos" almacena valores numéricos.
La variable "minimo" almacena el valor mínimo del conjunto.
La variable "maximo" almacena el valor máximo del conjunto.
La variable "normalizado" almacena los valores reescalados al intervalo [0, 1].
La transformación preserva las proporciones entre las diferencias de los valores."""

import numpy as np

datos = np.array([12, 18, 25, 40, 55])

minimo = np.min(datos)
maximo = np.max(datos)

normalizado = (datos - minimo) / (maximo - minimo)

normalizado # Muestra el resultado de la normalización min-max.

```

```python
# Bloque 2 — Normalización min‑max en múltiples características
"""Este bloque aplica normalización min-max a una matriz con varias características.
La variable "matriz" almacena valores distribuidos en columnas.
La variable "minimo_col" almacena los mínimos por columna.
La variable "maximo_col" almacena los máximos por columna.
La variable "normalizado_col" almacena la matriz transformada por columnas.
La normalización por características permite preparar datos multivariados para modelos sensibles a la escala."""

matriz = np.array([[2, 100],
                   [5, 150],
                   [8, 200],
                   [10, 300]])

minimo_col = np.min(matriz, axis=0)
maximo_col = np.max(matriz, axis=0)

normalizado_col = (matriz - minimo_col) / (maximo_col - minimo_col)

normalizado_col # Muestra la matriz normalizada por columnas.
```

```python
# Bloque 3 — Inversión de la normalización
"""Este bloque revierte la normalización min-max.
La variable "normalizado" almacena valores previamente normalizados.
La variable "original" almacena los valores reconstruidos mediante "minimo + normalizado * (maximo - minimo)".
La inversión permite recuperar la escala original cuando es necesaria para interpretación."""

original = minimo + (normalizado * (maximo - minimo))

original # Muestra los valores reconstruidos a partir de la normalización min-max.
         
```

```python
# Bloque 4 — Normalización min‑max aplicada a predicciones
"""Este bloque normaliza predicciones generadas por un modelo.
La variable "predicciones" almacena valores producidos por un modelo hipotético.
La variable "min_pred" almacena el mínimo de las predicciones.
La variable "max_pred" almacena el máximo de las predicciones.
La variable "pred_norm" almacena las predicciones reescaladas.
La normalización sitúa estas predicciones en una escala común."""

predicciones = np.array([0.8, 1.2, 2.5, 3.0, 4.1])    

min_pred = np.min(predicciones)
max_pred = np.max(predicciones)

pred_norm = (predicciones - min_pred) / (max_pred - min_pred)

pred_norm # Muestra las predicciones normalizadas al intervalo [0, 1].
```

```python
# Bloque 5 — Normalización min‑max para entrenamiento y test
"""Este bloque aplica normalización min-max usando parámetros del conjunto de entrenamiento.
La variable "train" almacena datos de entrenamiento.
La variable "test" almacena datos de prueba.
Las variables "min_train" y "max_train" almacenan los valores mínimo y máximo del entrenamiento.
Las variables "train_norm" y "test_norm" almacenan los conjuntos normalizados.
El uso de parámetros del entrenamiento evita fugas de información desde el conjunto de prueba. Los valores de prueba fuera del rango de entrenamiento pueden quedar fuera de [0, 1]."""

train = np.array([10, 20, 30, 40])
test = np.array([15, 35, 50]) 

min_train = np.min(train)
max_train = np.max(train) 

train_norm = (train - min_train) / (max_train - min_train)
test_norm = (test - min_train) / (max_train - min_train) # Usamos min y max del entrenamiento para normalizar el test.

train_norm, test_norm # Muestra los conjuntos de entrenamiento y prueba normalizados con los parámetros de entrenamiento.
```
