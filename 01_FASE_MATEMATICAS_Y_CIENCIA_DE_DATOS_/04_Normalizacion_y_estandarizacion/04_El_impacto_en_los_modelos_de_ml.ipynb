```python
# Bloque 1 — Impacto de la escala en el gradiente
"""Este bloque compara gradientes calculados con y sin estandarización.
Este bloque importa "numpy" para manejar valores numéricos.
La variable "datos" almacena valores numéricos.
Las variables "media" y "desviacion" almacenan los parámetros para estandarizar.
La variable "datos_est" almacena los valores estandarizados.
La variable "w" almacena un parámetro inicial.
Las variables "grad_original" y "grad_est" almacenan gradientes calculados sobre datos sin escalar y escalados.
La comparación muestra cómo la escala afecta la magnitud del gradiente."""

import numpy as np

datos = np.array([5, 20, 50, 80])

media = np.mean(datos)
desviacion = np.std(datos)

datos_est = (datos - media) / desviacion

w = 0.5

grad_original = -2 * (datos - w)
grad_est = -2 * (datos_est - w)

grad_original, grad_est # Muestra los gradientes calculados sobre datos sin escalar y escalados.    
```

```python
# Bloque 2 — Impacto en modelos basados en distancia
"""Este bloque muestra cómo la escala afecta distancias entre puntos.
La variable "puntos" almacena coordenadas en dos dimensiones.
Las variables "minimo" y "maximo" almacenan los valores mínimo y máximo de cada dimensión.
La variable "puntos_norm" almacena los puntos normalizados min-max.
La variable "dist_original" almacena la distancia euclidiana entre dos puntos sin escalar.
La variable "dist_norm" almacena la distancia entre los mismos puntos normalizados.
La comparación evidencia cómo la escala altera la geometría del espacio."""

puntos = np.array([[2, 100],
                   [8, 300]])

minimo = np.min(puntos, axis=0)
maximo = np.max(puntos, axis=0)

puntos_norm = (puntos - minimo) / (maximo - minimo)

dist_original = np.linalg.norm(puntos[0] - puntos[1]) 
dist_norm = np.linalg.norm(puntos_norm[0] - puntos_norm[1])

dist_original, dist_norm # Muestra las distancias calculadas entre puntos sin escalar y normalizados.
```

```python
# Bloque 3 — Impacto en la estabilidad numérica
"""Este bloque ilustra cómo la escala afecta operaciones internas de un modelo.
La variable "X" almacena valores con magnitudes grandes.
Las variables "media_X" y "desv_X" almacenan los parámetros para estandarizar.
La variable "X_est" almacena los valores estandarizados.
La variable "producto_original" almacena el resultado de "X.T @ X".
La variable "producto_est" almacena el resultado de "X_est.T @ X_est".
La comparación muestra cómo la estandarización reduce la magnitud del producto interno en este ejemplo."""

X = np.array([1000, 2000, 3000, 4000], dtype=float)

media_X = np.mean(X)
desv_X = np.std(X)

X_est = (X - media_X) / desv_X                   

producto_original = X.T @ X
producto_est = X_est.T @ X_est

producto_original, producto_est # Muestra los productos internos calculados sobre datos sin escalar y estandarizados.
```

```python
# Bloque 4 — Impacto en la regularización
"""Este bloque compara la penalización de parámetros originales y estandarizados.
La variable "w" almacena parámetros con magnitudes distintas.
Las variables "media_w" y "desv_w" almacenan los parámetros para estandarizar.
La variable "w_est" almacena los parámetros estandarizados.
La variable "penalizacion_original" almacena la suma de cuadrados de "w".
La variable "penalizacion_est" almacena la suma de cuadrados de "w_est".
La comparación muestra cómo cambia la suma de cuadrados al estandarizar los parámetros; no compara modelos equivalentes."""

w = np.array([0.1, 10.0, 50.0])

media_w = np.mean(w)
desv_w = np.std(w)

w_est = (w - media_w) / desv_w

penalizacion_original = np.sum(w**2)
penalizacion_est = np.sum(w_est**2)

penalizacion_original, penalizacion_est # Muestra las penalizaciones calculadas sobre parámetros sin escalar y estandarizados.
```

```python
# Bloque 5 — Impacto en la estabilidad del entrenamiento
"""Este bloque simula un paso de entrenamiento con datos sin escalar y escalados.
La variable "tasa" almacena la tasa de aprendizaje.
La variable "w_inicial" fija un parámetro escalar independiente del vector de parámetros del bloque anterior.
Las variables "grad_sin" y "grad_con" almacenan gradientes sobre datos sin escalar y estandarizados.
Las variables "w_sin" y "w_con" almacenan los parámetros actualizados.
Los datos son objetivos de este ejemplo simplificado; las dos actualizaciones no representan modelos equivalentes."""

tasa = 0.01
w_inicial = 0.5

grad_sin = -2 * (datos - w_inicial)
grad_con = -2 * (datos_est - w_inicial)

w_sin = w_inicial - tasa * np.mean(grad_sin)
w_con = w_inicial - tasa * np.mean(grad_con)

w_sin, w_con # Muestra los parámetros actualizados después de un paso de entrenamiento con datos sin escalar y estandarizados.

```
