```python
# Bloque 1 — Creación de vectores
"""Este bloque importa "numpy" para manejar estructuras numéricas.
La variable "v1" almacena un vector unidimensional con tres componentes.
La variable "v2" almacena un segundo vector con cuatro componentes.
Cada vector se representa mediante "np.array", que permite operar sobre sus componentes de forma vectorizada."""

import numpy as np

v1 = np.array([2, -1, 3])
v2 = np.array([5, 0, -2, 7])

v1, v2 # Muestra los vectores "v1" y "v2" para verificar su creación.
```

```python
# Bloque 2 — Suma y resta de vectores
"""Este bloque realiza operaciones básicas entre vectores.
La variable "suma_v1_v1" almacena la suma componente a componente entre "v1" y "v1".
La variable "resta_v1" almacena la resta entre "v1" y un vector constante creado mediante "np.array()".
Estas operaciones ilustran la manipulación directa de magnitudes y direcciones en espacios vectoriales."""

suma_v1_v1 = (v1 + v1)
resta_v1 = (v1 - np.array([1, 1, 1]))

suma_v1_v1, resta_v1 # Muestra los resultados de la suma y la resta para verificar las operaciones.
```

```python
# Bloque 3 — Producto escalar
"""Este bloque calcula el producto escalar entre dos vectores.
La variable "v3" almacena un vector compatible con "v1" en dimensión.
La variable "producto_escalar" almacena el resultado de "np.dot()" aplicado a "v1" y "v3".
El producto escalar depende de las magnitudes de los vectores y del ángulo entre ellos."""

v3 = np.array([1, 4, -2])
producto_escalar = np.dot(v1, v3)

producto_escalar # Muestra el resultado del producto escalar para verificar la operación.
```

```python
# Bloque 4 — Norma de un vector
"""Este bloque calcula la norma euclidiana de un vector.
La variable "norma_v1" almacena la magnitud de "v1" mediante "np.linalg.norm()".
La norma cuantifica la longitud del vector dentro del espacio vectorial."""

norma_v1 = np.linalg.norm(v1)

norma_v1 # Muestra la norma de "v1" para verificar la operación.
```

```python
# Bloque 5 — Normalización de un vector
"""Este bloque obtiene la versión normalizada de un vector.
La variable "direccion_v1" almacena el vector "v1" dividido entre su norma.
La operación produce un vector de magnitud unitaria que conserva la dirección original."""

direccion_v1 = (v1 / norma_v1)

direccion_v1 # Muestra la dirección normalizada de "v1" para verificar la operación.
```
