```python
# Bloque 1 — Suma y resta de vectores
"""Este bloque importa "numpy" para manejar vectores numéricos.
La variable "v1" almacena un vector de tres componentes.
La variable "v2" almacena un segundo vector compatible en dimensión.
La variable "suma" almacena la suma componente a componente entre "v1" y "v2".
La variable "resta" almacena la resta componente a componente entre "v1" y "v2".
Estas operaciones permiten combinar y comparar magnitudes dentro del espacio vectorial."""

import numpy as np

v1 = np.array([3, -1, 2])
v2 = np.array([1, 4, -3])

suma = (v1 + v2)
resta = (v1 - v2)

suma, resta # Muestra los resultados de la suma y la resta de los vectores "v1" y "v2".
```

```python
# Bloque 2 — Producto escalar
"""Este bloque calcula el producto escalar entre dos vectores.
La variable "v3" almacena un vector compatible con "v1" en dimensión.
La variable "producto_escalar" almacena el resultado de "np.dot()" aplicado a "v1" y "v3".
El producto escalar depende de las magnitudes de los vectores y del ángulo entre ellos."""

v3 = np.array([2, 0, -1])
producto_escalar = np.dot(v1, v3)

producto_escalar # Muestra el resultado del producto escalar entre los vectores "v1" y "v3".
```

```python
# Bloque 3 — Multiplicación por un escalar
"""Este bloque aplica un escalar a un vector.
La variable "escalar" almacena un valor numérico.
La variable "vector_escalado" almacena el resultado de multiplicar cada componente de "v1" por el escalar.
La operación ajusta la magnitud del vector sin modificar su dirección, ya que el escalar es positivo."""

escalar = 5
vector_escalado = (escalar * v1)

vector_escalado # Muestra el resultado de la multiplicación del vector "v1" por el escalar.
```

```python
# Bloque 4 — Multiplicación matricial
"""Este bloque realiza una multiplicación matricial entre una matriz y un vector.
La variable "M" almacena una matriz de dos filas y tres columnas.
La variable "v4" almacena un vector de tres componentes.
La variable "resultado" almacena el producto matricial mediante "np.dot()".
La multiplicación matricial aplica una transformación lineal sobre el vector."""

M = np.array([[1, 0, 2],
              [3, -1, 1]])

v4 = np.array([2, 1, -1])

resultado = np.dot(M, v4)

resultado # Muestra el resultado de la multiplicación matricial entre la matriz "M" y el vector "v4".
```

```python
# Bloque 5 — Combinación lineal
"""Este bloque genera una combinación lineal de dos vectores.
La variable "a" almacena un coeficiente escalar.
La variable "b" almacena un segundo coeficiente escalar.
La variable "combinacion" almacena la suma ponderada "a*v1 + b*v2".
Las combinaciones lineales permiten construir nuevas representaciones dentro del espacio vectorial."""

a = 2
b = -3
combinacion = (a * v1) + (b * v2)

combinacion # Muestra el resultado de la combinación lineal de los vectores "v1" y "v2".
```
