```python
# Bloque 1 — Creación de matrices      
"""Este bloque importa "numpy" para manejar estructuras bidimensionales.
La variable "M1" almacena una matriz de dos filas y tres columnas mediante "np.array()".
La variable "M2" almacena una matriz cuadrada de tres filas y tres columnas mediante "np.array()".
Cada matriz organiza datos en forma rectangular para permitir operaciones lineales."""

import numpy as np

M1 = np.array([[1, 2, 3],
               [4, 5, 6]])

M2 = np.array([[2, 0, 1],
               [1, 3, 4],
               [0, -1, 2]])

M1, M2 # Muestra las matrices "M1" y "M2" para verificar su creación.
```

```python
# Bloque 2 — Acceso a filas y columnas
"""Este bloque extrae elementos estructurales de una matriz.
La variable "primera_fila" almacena la fila 0 de "M1" mediante indexación.
La variable "segunda_columna" almacena la columna 1 de "M1" utilizando slicing.
Estas operaciones permiten acceder a componentes específicos dentro de la estructura matricial."""

primera_fila = M1[0]
segunda_columna = M1[:, 1]

primera_fila, segunda_columna # Muestra la primera fila y la segunda columna para verificar el acceso.
```

```python
# Bloque 3 — Suma y resta de matrices
"""Este bloque realiza operaciones básicas entre matrices del mismo tamaño.
La variable "suma_M1" almacena la suma componente a componente entre "M1" y "M1".
La variable "resta_M1" almacena la resta entre "M1" y una matriz constante creada mediante "np.array()".
Estas operaciones ilustran la manipulación directa de estructuras bidimensionales."""

suma_M1 = (M1 + M1)
resta_M1 = M1 - np.array([[1, 1, 1],
                          [1, 1, 1]])

suma_M1, resta_M1 # Muestra la suma y la resta para verificar las operaciones.
```

```python
# Bloque 4 — Multiplicación matricial
"""Este bloque calcula el producto matricial entre "M1" y la transpuesta de "M1".
La variable "M1_transpuesta" almacena la matriz transpuesta de "M1" mediante ".T".
La variable "producto_matricial" almacena el resultado de "np.dot()" aplicado a "M1" y "M1_transpuesta".
La multiplicación matricial permite combinar relaciones entre filas y columnas."""

M1_transpuesta = M1.T
producto_matricial = np.dot(M1, M1_transpuesta)

M1_transpuesta, producto_matricial # Muestra la transpuesta de "M1" y el "producto_matricial" para verificar las operaciones.
```

```python
# Bloque 5 — Determinante de una matriz
"""Este bloque calcula el determinante de una matriz cuadrada.
La variable "determinante_M2" almacena el valor del determinante de "M2" mediante "np.linalg.det()".
El determinante permite determinar si la matriz es invertible."""

determinante_M2 = np.linalg.det(M2)

determinante_M2 # Muestra el determinante de "M2" para verificar la operación.
```

```python
# Bloque 6 — Inversa de una matriz
"""Este bloque obtiene la matriz inversa de "M2".
La variable "inversa_M2" almacena el resultado de "np.linalg.inv()" aplicado a "M2".
La matriz inversa permite deshacer transformaciones lineales cuando la matriz es invertible."""

inversa_M2 = np.linalg.inv(M2)

inversa_M2 # Muestra la inversa de "M2" para verificar la operación.
```
