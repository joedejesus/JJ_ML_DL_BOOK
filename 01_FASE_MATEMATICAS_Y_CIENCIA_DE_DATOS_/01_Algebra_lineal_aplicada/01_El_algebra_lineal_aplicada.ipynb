```python
# Bloque 1 — Representación de vectores y matrices
"""Este bloque crea dos vectores y una matriz para ilustrar la representación básica de datos en álgebra lineal aplicada. 
Se importa "numpy" para manejar estructuras numéricas multidimensionales.
La variable "v1" almacena un vector de tres componentes.
La variable "v2" almacena un segundo vector de tres componentes.
La variable "M" almacena una matriz de dos filas y tres columnas.
Cada estructura representa datos que pueden ser manipulados mediante operaciones lineales."""

import numpy as np

v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
M = np.array([[1, 2, 3],
              [4, 5, 6]])

v1, v2, M # Muestra los vectores y la matriz para verificar su contenido.
```

```python
# Bloque 2 — Magnitudes y direcciones
"""Este bloque calcula la magnitud de un vector y su dirección normalizada.
La variable "magnitud_v1" almacena la norma euclidiana del vector "v1" mediante "np.linalg.norm()".
La variable "direccion_v1" almacena el vector normalizado dividiendo cada componente de "v1" entre su magnitud.
Estas operaciones permiten medir la intensidad y orientación de los datos representados por vectores."""

magnitud_v1 = np.linalg.norm(v1)
direccion_v1 = (v1 / magnitud_v1)

magnitud_v1, direccion_v1 # Muestra la magnitud y la dirección normalizada del vector "v1".
```

```python
# Bloque 3 — Combinaciones lineales
"""Este bloque genera una combinación lineal de los vectores "v1" y "v2".
La variable "a" almacena un coeficiente escalar.
La variable "b" almacena un segundo coeficiente escalar.
La variable "combinacion" almacena la suma ponderada (a*v1 + b*v2).
Las combinaciones lineales permiten construir nuevas representaciones a partir de vectores existentes."""

a = 2
b = -1    
combinacion = (a * v1) + (b * v2)

combinacion # Muestra la combinación lineal de los vectores "v1" y "v2".   
```

```python
# Bloque 4 — Transformaciones lineales con matrices
"""Este bloque aplica una transformación lineal a un vector mediante multiplicación matricial.
La variable "T" almacena una matriz cuadrada que actúa como transformación.
La variable "resultado" almacena el producto matricial entre "T" y "v1" utilizando "np.dot()".
Las transformaciones lineales permiten modificar datos preservando relaciones estructurales."""

T = np.array([[2, 0, 0],
              [0, 1, 0],
              [0, 0, 3]])

resultado = np.dot(T, v1)

resultado # Muestra el resultado de la transformación lineal aplicada al vector "v1".
```

```python
# Bloque 5 — Resolución de sistemas lineales
"""Este bloque resuelve un sistema lineal representado por "Ax = b".
La variable "A" almacena una matriz cuadrada que define las relaciones del sistema.
La variable "b_vec" almacena un vector de términos independientes.
La variable "solucion" almacena el vector "x" obtenido mediante "np.linalg.solve()".
La resolución de sistemas lineales permite encontrar valores que satisfacen relaciones entre variables."""

A = np.array([[3, 1],
              [1, 2]])

b_vec = np.array([9, 8])
solucion = np.linalg.solve(A, b_vec) 

solucion # Muestra la solución del sistema lineal.
```
